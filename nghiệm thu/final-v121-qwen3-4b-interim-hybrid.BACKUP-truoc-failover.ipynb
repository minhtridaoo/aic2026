
## v1.0.21 — Answer-flow correction from `test-vongsotuyen.ipynb`

The uploaded trial notebook makes the intended separation explicit:

```text
KIS / TRAKE = retrieval layer
QA          = the only branch that needs VLM image answering
```

So this final notebook no longer spends an LLM planner/verifier request on every KIS query.

Final remote-call policy:

```text
KIS   : 0 external API calls
TRAKE : 0 external API calls
QA    : 2 external calls/query
        1 grounded answer + 1 grounded verifier
```

For the 24-query trial pack (18 KIS, 3 TRAKE, 3 QA), the normal remote budget is therefore
**6 VLM calls total**, not an agent call chain across all 24 queries.

The log clock is now explicitly **ICT / Asia-Ho_Chi_Minh**.

# AIC26 preliminary wire format

## KIS

```csv
video_id,frame_id
```

up to 100 rows.

## QA

```csv
video_id,frame_id,answer
```

Answer <= 100 characters.

## TRAKE

```csv
video_id,frame_1,...,frame_N
```

N must equal the requested number of events.

## ZIP

```text
submission.zip
└── submission/
    ├── query-1-kis.csv
    ├── query-2-qa.csv
    ├── query-3-trake.csv
    └── ...
```

No header. UTF-8. Comma delimiter. Video ID has no `.mp4`. Frame IDs are integers.

````python
# ── F0 / CELL 2: INTEGRITY / RESOLVER HELPERS ─────────────────────────────────
# Mục tiêu: bounded dataset resolution + exact artifact hashing.
# INPUT: mounted Kaggle datasets.
# OUTPUT: helpers.
# Pre-check: known dataset handles only; no whole-corpus recursive guessing.
# Post-check: every accepted parent is verified by stage/build/manifest.
# History v0.2.0: serving-aware resolver replaces v0.1.0 filename capability guess.
# History v0.5.1: preflight all eight exact serving datasets before any heavy verification/model load.
# ──────────────────────────────────────────────────────────────────────────────
def sha256_file(path,chunk=8*1024*1024):
    h=hashlib.sha256()
    with Path(path).open("rb") as f:
        for b in iter(lambda:f.read(chunk),b""):
            h.update(b)
    return h.hexdigest()

def atomic_json(path,obj):
    path=Path(path)
    path.parent.mkdir(parents=True,exist_ok=True)
    tmp=path.with_suffix(path.suffix+".tmp")
    tmp.write_text(json.dumps(obj,ensure_ascii=False,indent=2,default=str),encoding="utf-8")
    tmp.replace(path)

def resolve_dataset(handle):
    owner,slug=str(handle).split("/",1)
    candidates=[
        Path("/kaggle/input")/slug,
        Path("/kaggle/input")/"datasets"/owner/slug,
        Path("/kaggle/input")/"notebooks"/owner/slug,
    ]
    valid=[p.resolve() for p in candidates if p.is_dir()]
    if valid:
        return valid[0]
    raise RuntimeError(("DATASET_NOT_MOUNTED",handle,[str(x) for x in candidates]))

def unique_under(root,name):
    hits=sorted({p.resolve() for p in Path(root).rglob(name) if p.is_file()},key=str)
    if len(hits)!=1:
        raise RuntimeError(("EXPECTED_ONE_FILE",name,str(root),[str(x) for x in hits[:20]],len(hits)))
    return hits[0]

def stage_root_from_handoff(hp):
    return Path(hp).resolve().parent.parent

def rebase(root,stored,name):
    root=Path(root).resolve()
    p=Path(str(stored))
    if p.exists():
        rp=p.resolve()
        if root==rp or root in rp.parents:
            return rp
    return unique_under(root,name)

def verify_artifact_manifest(stage_root,manifest_path,subset=None):
    root=Path(stage_root).resolve()
    m=json.loads(Path(manifest_path).read_text(encoding="utf-8"))
    mmap={str(x["relative_path"]):x for x in m.get("artifacts",[])}
    if not mmap:
        raise AssertionError("empty artifact manifest")
    if subset is None:
        rels=list(mmap)
    else:
        rels=[]
        for p in subset:
            p=Path(p).resolve()
            rels.append(p.relative_to(root).as_posix())
    checked=0
    for rel in rels:
        if rel not in mmap:
            raise AssertionError(("manifest missing",rel))
        x=mmap[rel]
        p=root/rel
        if not p.is_file():
            raise FileNotFoundError(p)
        if int(p.stat().st_size)!=int(x["size_bytes"]):
            raise AssertionError(("size mismatch",rel))
        if sha256_file(p)!=str(x["sha256"]):
            raise AssertionError(("sha mismatch",rel))
        checked+=1
    return m,checked

def tree_fingerprint(root):
    rows=[]; total=0
    for p in sorted(Path(root).rglob("*")):
        if p.is_file():
            s=int(p.stat().st_size); total+=s
            rows.append((str(p.relative_to(root)),s))
    blob=json.dumps(rows,sort_keys=True,separators=(",",":")).encode()
    return {
        "file_count":len(rows),
        "total_bytes":total,
        "path_size_sha256":hashlib.sha256(blob).hexdigest(),
    }

def read_json(path):
    return json.loads(Path(path).read_text(encoding="utf-8"))

print("[PASS] resolver helpers")

O2_REQUIRED_DATASET_HANDLES=(
    "nguyenminhtric/final24",
    "nguyenminhtric/20-note",
    "selaninguyen/03notre",
    "selaninguyen/gggggg",
    "nguyenminhtric/data-batch-1-aic",
    "selaninguyen/06note",
    "selaninguyen/11ntoe",
    "selaninguyen/fixed-v16",
)

def preflight_exact_serving_inputs(handles=O2_REQUIRED_DATASET_HANDLES):
    """
    Kaggle papermill/non-interactive runs cannot reliably attach a new dataset
    after execution starts. Resolve all exact serving handles up front and
    report the complete missing set in one fail-closed error.
    """
    report={}
    missing=[]
    for handle in handles:
        owner,slug=str(handle).split("/",1)
        candidates=[
            Path("/kaggle/input")/slug,
            Path("/kaggle/input")/"datasets"/owner/slug,
            Path("/kaggle/input")/"notebooks"/owner/slug,
        ]
        valid=[p.resolve() for p in candidates if p.is_dir()]
        if valid:
            report[handle]={
                "status":"MOUNTED_EXACT",
                "root":str(valid[0]),
                "candidates":[str(p) for p in candidates],
            }
        else:
            missing.append(handle)
            report[handle]={
                "status":"MISSING_ADD_INPUT_REQUIRED",
                "root":None,
                "candidates":[str(p) for p in candidates],
            }
    if missing:
        raise RuntimeError({
            "message":"ONLINE_O2_INPUT_PREFLIGHT_FAILED",
            "missing_exact_datasets":missing,
            "required_exact_datasets":list(handles),
            "action":"Kaggle Add Input: attach every missing exact dataset before Run All/papermill.",
            "late_dataset_attach_supported":False,
            "report":report,
        })
    return report

O2_INPUT_PREFLIGHT=preflight_exact_serving_inputs()
print("[PASS] O2 all-input preflight",{
    "required":len(O2_INPUT_PREFLIGHT),
    "mounted_exact":sum(v["status"]=="MOUNTED_EXACT" for v in O2_INPUT_PREFLIGHT.values()),
    "missing":[],
    "late_dataset_attach_supported":False,
})
````

````python
# ── F1 / CELL 3: VERIFY CANONICAL FINAL24 ─────────────────────────────────────
# Mục tiêu: pin final offline closure before serving.
# INPUT: nguyenminhtric/final24.
# OUTPUT: FINAL24_*.
# Pre-check: exact v1.0.3 master status.
# Post-check: full manifest verification + expected active state.
# History v0.2.0: final24 is closure authority, not a general search index.
# ──────────────────────────────────────────────────────────────────────────────
FINAL24_DATASET_ROOT=resolve_dataset("nguyenminhtric/final24")
FINAL24_HANDOFF_PATH=unique_under(FINAL24_DATASET_ROOT,"stage24_handoff.json")
FINAL24_ROOT=stage_root_from_handoff(FINAL24_HANDOFF_PATH)
FINAL24_DECISION_PATH=unique_under(FINAL24_ROOT,"stage24_master_decision.json")
FINAL24_MANIFEST_PATH=unique_under(FINAL24_ROOT,"stage24_artifact_manifest.json")

FINAL24_HANDOFF=read_json(FINAL24_HANDOFF_PATH)
FINAL24_DECISION=read_json(FINAL24_DECISION_PATH)

if FINAL24_HANDOFF.get("build_version")!=EXPECTED_FINAL24_BUILD:
    raise AssertionError(("final24 build",FINAL24_HANDOFF.get("build_version")))
if FINAL24_DECISION.get("status")!=EXPECTED_FINAL24_STATUS:
    raise AssertionError(("final24 status",FINAL24_DECISION.get("status")))
if FINAL24_DECISION.get("active_state")!={"KIS":"v0","TRAKE":"v3","QA":"v0_PENDING_FINAL_API"}:
    raise AssertionError(("final24 active state",FINAL24_DECISION.get("active_state")))
if FINAL24_DECISION.get("probability_calibration_status")!="NOT_CALIBRATED":
    raise AssertionError("calibration state drift")

FINAL24_MANIFEST,FINAL24_MANIFEST_CHECKED=verify_artifact_manifest(
    FINAL24_ROOT,FINAL24_MANIFEST_PATH
)

print("[PASS] canonical final24",{
    "root":str(FINAL24_ROOT),
    "manifest_entries":FINAL24_MANIFEST_CHECKED,
    "active_state":FINAL24_DECISION["active_state"],
})
````

````python
# ── F1 / CELL 4: VERIFY + LOAD STAGE20 COMPACT EVIDENCE ───────────────────────
# Mục tiêu: use Stage20 as compact full-corpus serving evidence.
# INPUT: nguyenminhtric/20-note.
# OUTPUT: PACK / FRAME_PAYLOAD / ASR_PAYLOAD / TEMPORAL_PAYLOAD / BOUNDARY_PAYLOAD.
# Pre-check: exact build/config/SHA pinned by final24 lineage.
# Post-check: core schemas exact; no ShotRef canonicalization.
# History v0.2.0: replaces broad historical parent attachment with compact EvidencePack.
# ──────────────────────────────────────────────────────────────────────────────
STAGE20_DATASET_ROOT=resolve_dataset("nguyenminhtric/20-note")
H20_PATH=unique_under(STAGE20_DATASET_ROOT,"stage20_handoff.json")
if sha256_file(H20_PATH)!=EXPECTED_STAGE20_HANDOFF_SHA256:
    raise AssertionError(("Stage20 handoff SHA",sha256_file(H20_PATH)))
H20=read_json(H20_PATH)
if H20.get("stage_id")!=EXPECTED_STAGE20_ID or H20.get("build_version")!=EXPECTED_STAGE20_BUILD:
    raise AssertionError(("Stage20 identity",H20.get("stage_id"),H20.get("build_version")))
if H20.get("config_fingerprint")!=EXPECTED_STAGE20_CONFIG_FINGERPRINT:
    raise AssertionError("Stage20 config fingerprint")

S20=stage_root_from_handoff(H20_PATH)
M20_PATH=rebase(S20,H20["artifact_manifest"],"stage20_artifact_manifest.json")

PACK_PATH=rebase(S20,H20["evidence_pack_registry_parquet"],"shadow_evidence_pack_registry.parquet")
FRAME_PATH=rebase(S20,H20["frame_payload_parquet"],"shadow_evidence_pack_frame_payload.parquet")
ASR_PATH=rebase(S20,H20["asr_payload_parquet"],"shadow_evidence_pack_asr_payload.parquet")
TEMPORAL_PATH=rebase(S20,H20["temporal_payload_parquet"],"shadow_evidence_pack_temporal_payload.parquet")
BOUNDARY_PATH=rebase(S20,H20["boundary_payload_parquet"],"shadow_evidence_pack_boundary_payload.parquet")

M20,verified20=verify_artifact_manifest(
    S20,M20_PATH,[H20_PATH,PACK_PATH,FRAME_PATH,ASR_PATH,TEMPORAL_PATH,BOUNDARY_PATH]
)

PACK=pd.read_parquet(PACK_PATH)
FRAME_PAYLOAD=pd.read_parquet(FRAME_PATH)
ASR_PAYLOAD=pd.read_parquet(ASR_PATH)
TEMPORAL_PAYLOAD=pd.read_parquet(TEMPORAL_PATH)
BOUNDARY_PAYLOAD=pd.read_parquet(BOUNDARY_PATH)

required_pack={
    "evidence_pack_id","variant","shadow_shot_interval_id","video_id",
    "start_pts_time_technical","end_pts_time_technical","duration_sec",
    "ocr_text_compact","asr_text_compact","availability_mask_json","available_modality_count",
}
required_frame={"frame_ref_id","variant","shadow_shot_interval_id","video_id","membership_status"}
if not required_pack.issubset(PACK.columns):
    raise RuntimeError(("PACK schema missing",sorted(required_pack-set(PACK.columns))))
if not required_frame.issubset(FRAME_PAYLOAD.columns):
    raise RuntimeError(("FRAME schema missing",sorted(required_frame-set(FRAME_PAYLOAD.columns))))

PACK["shadow_shot_interval_id"]=PACK.shadow_shot_interval_id.astype(str)
FRAME_PAYLOAD["shadow_shot_interval_id"]=FRAME_PAYLOAD.shadow_shot_interval_id.astype(str)
FRAME_PAYLOAD["frame_ref_id"]=FRAME_PAYLOAD.frame_ref_id.astype(str)

print("[PASS] Stage20 compact evidence",{
    "packs":len(PACK),
    "frame_payload":len(FRAME_PAYLOAD),
    "asr_payload":len(ASR_PAYLOAD),
    "temporal_payload":len(TEMPORAL_PAYLOAD),
    "boundary_payload":len(BOUNDARY_PAYLOAD),
    "verified_artifacts":verified20,
})
````

````python
# ── F1 / CELL 5: VERIFY STAGE03 + STAGE00 SERVING ARTIFACTS ───────────────────
# Mục tiêu: exact official FrameRef visual index + frozen local SigLIP2.
# INPUT: selaninguyen/03notre + selaninguyen/gggggg.
# OUTPUT: VISUAL_META, EMBEDDINGS, optional FAISS, VISUAL_MODEL_DIR.
# Pre-check: Stage03 exact reference index; Stage00 strict 2-GPU model contract.
# Post-check: model tree/index hashes exact.
# History v0.2.0: production KIS serving source.
# History v0.2.1: FAISS remains optional; primary runtime backend is initialized after GPU workers.
# History v0.2.2: assert frozen Stage03 storage dtype FP16; serving matrix will widen to FP32.
# ──────────────────────────────────────────────────────────────────────────────
STAGE00_DATASET_ROOT=resolve_dataset("selaninguyen/gggggg")
H00_PATH=unique_under(STAGE00_DATASET_ROOT,"stage00_handoff.json")
H00=read_json(H00_PATH)
if H00.get("stage_id")!=EXPECTED_STAGE00_ID:
    raise AssertionError(("Stage00 id",H00.get("stage_id")))
if int(H00.get("required_gpu_count",-1))!=2:
    raise AssertionError("Stage00 dual-GPU count")
if H00.get("allow_single_gpu_fallback") is not False or H00.get("allow_cpu_model_fallback") is not False:
    raise AssertionError("Stage00 fallback policy")

MODEL_MANIFEST_PATH=rebase(
    stage_root_from_handoff(H00_PATH),H00["visual_model_manifest"],"visual_model_manifest.json"
)
MODEL_MANIFEST=read_json(MODEL_MANIFEST_PATH)
if MODEL_MANIFEST.get("model_id")!=VISUAL_MODEL_ID:
    raise AssertionError("Stage00 visual model id")

model_candidates=[]
for p in STAGE00_DATASET_ROOT.rglob("config.json"):
    parent=p.parent
    if "siglip2" in str(parent).lower() and (
        any(parent.glob("*.safetensors")) or any(parent.glob("pytorch_model*.bin"))
    ):
        model_candidates.append(parent.resolve())
model_candidates=sorted(set(model_candidates),key=str)
if len(model_candidates)!=1:
    raise RuntimeError(("SigLIP2 model candidates",[str(x) for x in model_candidates]))
VISUAL_MODEL_DIR=model_candidates[0]
MODEL_TREE_FINGERPRINT=tree_fingerprint(VISUAL_MODEL_DIR)
if MODEL_TREE_FINGERPRINT!=MODEL_MANIFEST.get("fingerprint"):
    raise AssertionError("Stage00 model tree fingerprint")

STAGE03_DATASET_ROOT=resolve_dataset("selaninguyen/03notre")
H03_PATH=unique_under(STAGE03_DATASET_ROOT,"stage03_handoff.json")
H03=read_json(H03_PATH)
if H03.get("stage_id")!=EXPECTED_STAGE03_ID or H03.get("build_version")!=EXPECTED_STAGE03_BUILD:
    raise AssertionError(("Stage03 identity",H03.get("stage_id"),H03.get("build_version")))
S03=stage_root_from_handoff(H03_PATH)
M03_PATH=rebase(S03,H03["artifact_manifest"],"stage03_artifact_manifest.json")
INDEX_MANIFEST_PATH=rebase(S03,H03["visual_index_manifest"],"visual_index_manifest.json")
GPU03_PATH=rebase(S03,H03["dual_gpu_usage"],"dual_gpu_encoding_usage.json")
EMBEDDINGS_PATH=rebase(S03,H03["visual_embeddings"],"visual_embeddings.npy")
VISUAL_META_PATH=rebase(S03,H03["visual_metadata"],"visual_metadata.parquet")
FAISS_PATH=rebase(S03,H03["visual_faiss"],"visual_exact.faiss")

M03,verified03=verify_artifact_manifest(
    S03,M03_PATH,[H03_PATH,INDEX_MANIFEST_PATH,GPU03_PATH,EMBEDDINGS_PATH,VISUAL_META_PATH,FAISS_PATH]
)
INDEX_MANIFEST=read_json(INDEX_MANIFEST_PATH)
GPU03=read_json(GPU03_PATH)

if INDEX_MANIFEST.get("scope")!="KIS_SLICE1_OFFICIAL_ONLY":
    raise AssertionError("Stage03 index scope")
if INDEX_MANIFEST.get("index_type")!="IndexFlatIP_REFERENCE":
    raise AssertionError("Stage03 index type")
if INDEX_MANIFEST.get("model_id")!=VISUAL_MODEL_ID:
    raise AssertionError("Stage03 model id")
if INDEX_MANIFEST.get("model_tree_fingerprint")!=MODEL_TREE_FINGERPRINT:
    raise AssertionError("Stage03↔Stage00 model tree")
if INDEX_MANIFEST.get("model_manifest_sha256")!=sha256_file(MODEL_MANIFEST_PATH):
    raise AssertionError("Stage03↔Stage00 model manifest")
if GPU03.get("status")!="PASS" or GPU03.get("execution_policy")!="STRICT_DUAL_GPU":
    raise AssertionError("Stage03 build GPU evidence")

for p,key in (
    (EMBEDDINGS_PATH,"embeddings_sha256"),
    (VISUAL_META_PATH,"metadata_sha256"),
    (FAISS_PATH,"faiss_sha256"),
):
    if sha256_file(p)!=str(INDEX_MANIFEST[key]):
        raise AssertionError(("Stage03 serving artifact SHA",str(p),key))

VISUAL_META=pd.read_parquet(VISUAL_META_PATH)
FEATURE_DIM=int(INDEX_MANIFEST["feature_dim"])
VISUAL_ROWS=int(INDEX_MANIFEST["rows"])
if len(VISUAL_META)!=VISUAL_ROWS:
    raise AssertionError((len(VISUAL_META),VISUAL_ROWS))
if VISUAL_META.frame_ref_id.astype(str).duplicated().any():
    raise AssertionError("duplicate Stage03 frame_ref_id")
if not np.array_equal(
    VISUAL_META.index_row.to_numpy(dtype=np.int64),
    np.arange(len(VISUAL_META),dtype=np.int64),
):
    raise AssertionError("Stage03 index_row drift")
VISUAL_META["frame_ref_id"]=VISUAL_META.frame_ref_id.astype(str)

EMBEDDINGS=np.load(EMBEDDINGS_PATH,mmap_mode="r")
if EMBEDDINGS.shape!=(VISUAL_ROWS,FEATURE_DIM):
    raise AssertionError(("embedding shape",EMBEDDINGS.shape))
# Frozen Stage03 v0.2.2 intentionally persists merged embeddings as float16.
# Its exact FAISS reference index is built from these stored values widened to float32.
STAGE03_STORAGE_DTYPE=str(EMBEDDINGS.dtype)
if STAGE03_STORAGE_DTYPE!="float16":
    raise RuntimeError(("STAGE03_STORAGE_DTYPE_DRIFT",STAGE03_STORAGE_DTYPE,"expected float16"))

# Keep the exact Stage03 FAISS artifact verified, but do not depend on faiss at runtime.
# v0.2.1 uses the same Stage03 float32 embeddings directly on both CUDA devices.
VISUAL_INDEX=None
VISUAL_SEARCH_BACKEND="PENDING_DUAL_GPU_TORCH_EXACT_INIT"

print("[PASS] Stage03 + Stage00 serving",{
    "visual_rows":VISUAL_ROWS,
    "feature_dim":FEATURE_DIM,
    "backend":VISUAL_SEARCH_BACKEND,
    "stage03_verified":verified03,
    "model_dir":str(VISUAL_MODEL_DIR),
})
````

````python
# ── O2 / CELL: VERIFY + LOAD STAGE06 / STAGE11 / STAGE16 ──────────────────────
# Mục tiêu: load every additional O2 retrieval parent with exact lineage/hash gates.
# INPUT: selaninguyen/06note + selaninguyen/11ntoe + selaninguyen/fixed-v16.
# OUTPUT: ADAPTIVE_META/EMB, OBJ_MATRIX/ALIAS, MOTION_TYPED.
# Pre-check: Stage03 corpus/snapshot is the canonical cross-parent lineage anchor.
# Post-check: adaptive remains shadow, object threshold remains None, motion score remains normalized-only.
# History v0.5.0: first Online O2 retrieval-parent integration.
# History v0.5.2: fix Stage11 parent-lineage field source; field lives in retrieval index manifest.
# ──────────────────────────────────────────────────────────────────────────────

CORPUS_ID=str(H03["corpus_id"])
DATASET_SNAPSHOT_ID=str(H03["dataset_snapshot_id"])
OFFICIAL_FRAME_REGISTRY_SHA256=str(H03["frame_registry_sha256"])
STAGE03_HANDOFF_SHA256=sha256_file(H03_PATH)

# ---- Stage06 adaptive semantic feature bank -----------------------------------
STAGE06_DATASET_ROOT=resolve_dataset("selaninguyen/06note")
H06_PATH=unique_under(STAGE06_DATASET_ROOT,"stage06_handoff.json")
H06=read_json(H06_PATH)
if H06.get("stage_id")!=EXPECTED_STAGE06_ID or H06.get("build_version")!=EXPECTED_STAGE06_BUILD:
    raise AssertionError(("Stage06 identity",H06.get("stage_id"),H06.get("build_version")))
if str(H06.get("corpus_id"))!=CORPUS_ID or str(H06.get("dataset_snapshot_id"))!=DATASET_SNAPSHOT_ID:
    raise AssertionError("Stage06 corpus/snapshot drift")
if str(H06.get("official_frame_registry_sha256"))!=OFFICIAL_FRAME_REGISTRY_SHA256:
    raise AssertionError("Stage06 official FrameRef lineage drift")
if H06.get("canonical_membership_status")!="NOT_CREATED_PROPOSALS_ONLY":
    raise AssertionError("Stage06 adaptive canonical membership unexpectedly promoted")

S06=stage_root_from_handoff(H06_PATH)
M06_PATH=rebase(S06,H06["artifact_manifest"],"stage06_artifact_manifest.json")
ADAPTIVE_EMBEDDINGS_PATH=rebase(S06,H06["adaptive_embeddings"],"adaptive_visual_embeddings.npy")
ADAPTIVE_META_PATH=rebase(S06,H06["adaptive_metadata"],"adaptive_visual_metadata.parquet")
ADAPTIVE_METRICS_PATH=rebase(S06,H06["semantic_metrics"],"adaptive_semantic_metrics.parquet")
ADAPTIVE_PROPOSALS_PATH=rebase(S06,H06["selection_proposals"],"adaptive_selection_proposals.parquet")
M06,verified06=verify_artifact_manifest(
    S06,M06_PATH,[H06_PATH,ADAPTIVE_EMBEDDINGS_PATH,ADAPTIVE_META_PATH,ADAPTIVE_METRICS_PATH,ADAPTIVE_PROPOSALS_PATH]
)

ADAPTIVE_META=pd.read_parquet(ADAPTIVE_META_PATH)
ADAPTIVE_EMB=np.load(ADAPTIVE_EMBEDDINGS_PATH,mmap_mode="r")
if ADAPTIVE_EMB.shape!=(48656,FEATURE_DIM) or str(ADAPTIVE_EMB.dtype)!="float16":
    raise AssertionError(("Stage06 adaptive matrix",ADAPTIVE_EMB.shape,str(ADAPTIVE_EMB.dtype)))
req_adapt={"adaptive_index_row","frame_ref_id","video_id","pts_time","actual_pts_time"}
if not req_adapt.issubset(ADAPTIVE_META.columns):
    raise RuntimeError(("Stage06 adaptive metadata schema",sorted(req_adapt-set(ADAPTIVE_META.columns))))
if not np.array_equal(
    ADAPTIVE_META.adaptive_index_row.to_numpy(dtype=np.int64),
    np.arange(len(ADAPTIVE_META),dtype=np.int64),
):
    raise AssertionError("Stage06 adaptive_index_row drift")
if ADAPTIVE_META.frame_ref_id.astype(str).duplicated().any():
    raise AssertionError("Stage06 duplicate adaptive FrameRef")
if "raw_decode_verified" in ADAPTIVE_META.columns and not ADAPTIVE_META.raw_decode_verified.astype(bool).all():
    raise AssertionError("Stage06 adaptive raw decode verification drift")
ADAPTIVE_META["frame_ref_id"]=ADAPTIVE_META.frame_ref_id.astype(str)

# ---- Stage11 object searchable index ------------------------------------------
STAGE11_DATASET_ROOT=resolve_dataset("selaninguyen/11ntoe")
H11_PATH=unique_under(STAGE11_DATASET_ROOT,"stage11_handoff.json")
H11=read_json(H11_PATH)
if H11.get("stage_id")!=EXPECTED_STAGE11_ID or H11.get("build_version")!=EXPECTED_STAGE11_BUILD:
    raise AssertionError(("Stage11 identity",H11.get("stage_id"),H11.get("build_version")))
if str(H11.get("corpus_id"))!=CORPUS_ID or str(H11.get("dataset_snapshot_id"))!=DATASET_SNAPSHOT_ID:
    raise AssertionError("Stage11 corpus/snapshot drift")
# Stage11 handoff intentionally points to the retrieval index manifest; the
# parent Stage03 SHA is recorded inside that manifest, not duplicated in handoff.
if "parent_stage03_handoff_sha256" in H11:
    raise AssertionError("STAGE11_HANDOFF_UNEXPECTED_PARENT_SHA_FIELD")
if H11.get("canonical_multimodal_membership_status")!="NOT_CREATED":
    raise AssertionError("Stage11 canonical multimodal membership unexpectedly created")
if H11.get("promotion_decision_status")!="DEFER_NO_GT_OR_PROXY":
    raise AssertionError("Stage11 promotion decision drift")

S11=stage_root_from_handoff(H11_PATH)
M11_PATH=rebase(S11,H11["artifact_manifest"],"stage11_artifact_manifest.json")
OBJ_MATRIX_PATH=rebase(S11,H11["object_frame_class_max_scores"],"object_frame_class_max_scores.npy")
OBJ_ALIAS_PATH=rebase(S11,H11["object_alias_registry"],"object_alias_registry.parquet")
OBJ_BUILD_PATH=rebase(S11,H11["object_index_build_manifest"],"object_index_build_manifest.json")
INDEX11_PATH=rebase(S11,H11["retrieval_index_manifest"],"stage11_retrieval_index_manifest.json")
M11,verified11=verify_artifact_manifest(
    S11,M11_PATH,[H11_PATH,OBJ_MATRIX_PATH,OBJ_ALIAS_PATH,OBJ_BUILD_PATH,INDEX11_PATH]
)

OBJ_MATRIX=np.load(OBJ_MATRIX_PATH,mmap_mode="r")
OBJ_ALIAS=pd.read_parquet(OBJ_ALIAS_PATH)
OBJ_BUILD=read_json(OBJ_BUILD_PATH)
INDEX11=read_json(INDEX11_PATH)
if INDEX11.get("stage_id")!=EXPECTED_STAGE11_ID or INDEX11.get("build_version")!=EXPECTED_STAGE11_BUILD:
    raise AssertionError(("Stage11 retrieval index identity",INDEX11.get("stage_id"),INDEX11.get("build_version")))
if str(INDEX11.get("corpus_id"))!=CORPUS_ID or str(INDEX11.get("dataset_snapshot_id"))!=DATASET_SNAPSHOT_ID:
    raise AssertionError("Stage11 retrieval index corpus/snapshot drift")
if str(INDEX11.get("parent_stage03_handoff_sha256"))!=STAGE03_HANDOFF_SHA256:
    raise AssertionError({
        "message":"STAGE11_TO_STAGE03_LINEAGE_DRIFT",
        "stage11_recorded_parent_sha256":INDEX11.get("parent_stage03_handoff_sha256"),
        "mounted_stage03_handoff_sha256":STAGE03_HANDOFF_SHA256,
        "provenance_source":"stage11_retrieval_index_manifest.json",
    })
if OBJ_MATRIX.shape!=(VISUAL_ROWS,584) or str(OBJ_MATRIX.dtype)!="float32":
    raise AssertionError(("Stage11 object matrix",OBJ_MATRIX.shape,str(OBJ_MATRIX.dtype)))
if not {"alias_norm","class_idx"}.issubset(OBJ_ALIAS.columns):
    raise RuntimeError(("Stage11 alias schema",sorted({"alias_norm","class_idx"}-set(OBJ_ALIAS.columns))))
if OBJ_BUILD.get("status")!="PASS" or OBJ_BUILD.get("matrix_shape")!=[VISUAL_ROWS,584]:
    raise AssertionError("Stage11 object build manifest")
if int(OBJ_BUILD.get("detections_consumed",-1))!=17732100:
    raise AssertionError(("Stage11 detections",OBJ_BUILD.get("detections_consumed")))
if OBJ_BUILD.get("canonical_score_threshold") is not None:
    raise AssertionError("Stage11 object threshold must remain None")

print("[PASS] Stage11 lineage",{
    "provenance_source":"stage11_retrieval_index_manifest.json",
    "parent_stage03_handoff_sha256":INDEX11["parent_stage03_handoff_sha256"],
    "mounted_stage03_handoff_sha256":STAGE03_HANDOFF_SHA256,
})

# ---- Stage16 typed motion source ----------------------------------------------
STAGE16_DATASET_ROOT=resolve_dataset("selaninguyen/fixed-v16")
H16_PATH=unique_under(STAGE16_DATASET_ROOT,"stage16_handoff.json")
H16=read_json(H16_PATH)
if H16.get("stage_id")!=EXPECTED_STAGE16_ID or H16.get("build_version")!=EXPECTED_STAGE16_BUILD:
    raise AssertionError(("Stage16 identity",H16.get("stage_id"),H16.get("build_version")))
if str(H16.get("corpus_id"))!=CORPUS_ID or str(H16.get("dataset_snapshot_id"))!=DATASET_SNAPSHOT_ID:
    raise AssertionError("Stage16 corpus/snapshot drift")
if H16.get("pair_time_status")!="SCHEDULED_FILTER_GRID_NOT_ACTUAL_DECODE_PTS":
    raise AssertionError(("Stage16 pair time semantics",H16.get("pair_time_status")))
if H16.get("shot_ref_status")!="NOT_CREATED":
    raise AssertionError(("Stage16 ShotRef drift",H16.get("shot_ref_status")))

S16=stage_root_from_handoff(H16_PATH)
M16_PATH=rebase(S16,H16["artifact_manifest"],"stage16_artifact_manifest.json")
MOTION_EDGE_PATH=rebase(S16,H16["raw_motion_edge_evidence"],"raw_motion_edge_evidence.parquet")
MOTION_REPORT_PATH=rebase(S16,H16["raw_motion_report"],"stage16_raw_motion_report.json")
M16,verified16=verify_artifact_manifest(
    S16,M16_PATH,[H16_PATH,MOTION_EDGE_PATH,MOTION_REPORT_PATH]
)
MOTION_EDGE=pd.read_parquet(MOTION_EDGE_PATH)
MOTION_REPORT=read_json(MOTION_REPORT_PATH)
if len(MOTION_EDGE)!=9275:
    raise AssertionError(("Stage16 edge rows",len(MOTION_EDGE)))
if float(MOTION_REPORT.get("localized_motion_fraction",0.0))<0.98:
    raise AssertionError(("Stage16 localized fraction",MOTION_REPORT.get("localized_motion_fraction")))

required_motion={
    "video_id","edge_index_in_video","left_pts_time","right_pts_time","gap_sec",
    "decode_status","localized_motion_available","coverage_ratio",
}
if not required_motion.issubset(MOTION_EDGE.columns):
    raise RuntimeError(("Stage16 motion schema",sorted(required_motion-set(MOTION_EDGE.columns))))

MOTION_FEATURES=[
    "gray_mae_pair_mean","gray_mae_pair_max","gray_mae_pair_mean_per_sec",
    "flow_mag_pair_mean","flow_mag_pair_p95_max",
    "flow_mag_pair_mean_per_sec","flow_mag_pair_p95_max_per_sec",
]
MOTION_FEATURES=[c for c in MOTION_FEATURES if c in MOTION_EDGE.columns]
if len(MOTION_FEATURES)<4:
    raise RuntimeError(("Stage16 insufficient motion metrics",MOTION_FEATURES))

MOTION_TYPED=MOTION_EDGE.copy()
MOTION_TYPED["motion_available"]=MOTION_TYPED.localized_motion_available.fillna(False).astype(bool)
_loc=MOTION_TYPED.motion_available
for _c in MOTION_FEATURES:
    _v=pd.to_numeric(MOTION_TYPED[_c],errors="coerce")
    _pct=pd.Series(np.nan,index=MOTION_TYPED.index,dtype=float)
    _pct.loc[_loc]=_v.loc[_loc].rank(method="average",pct=True)
    MOTION_TYPED[_c+"__localized_pct"]=_pct
MOTION_TYPED["score_semantics"]="NORMALIZED_ONLY_NOT_PROBABILITY"

print("[PASS] O2 retrieval parents",{
    "stage06":{"rows":len(ADAPTIVE_META),"verified_artifacts":verified06,"promotion":ADAPTIVE_PROMOTION_STATUS},
    "stage11":{"matrix":list(OBJ_MATRIX.shape),"aliases":len(OBJ_ALIAS),"verified_artifacts":verified11},
    "stage16":{"edges":len(MOTION_TYPED),"localized":int(MOTION_TYPED.motion_available.sum()),"verified_artifacts":verified16},
})
````

````python
# ── F3 / CELL 8: STRICT DUAL-GPU SIGLIP2 QUERY SERVICE ────────────────────────
# Mục tiêu: create two real query encoder replicas.
# INPUT: verified Stage00 model.
# OUTPUT: QUERY_WORKERS, QUERY_USAGE.
# Pre-check: >=2 CUDA GPUs; no fallback.
# Post-check: full floating model normalized FP16 on each GPU + contract call each.
# History v0.2.0: includes Stage24 v1.0.3 dtype hotfix permanently.
# History v0.2.1: add exact float32 serving replica on each GPU for low-latency search.
# History v0.2.2: source artifact is FP16 by design; widen stored FP16 values to FP32 exactly as Stage03 FAISS build did.
# ──────────────────────────────────────────────────────────────────────────────
import threading, torch, transformers
from transformers import AutoModel, AutoProcessor

if ALLOW_SINGLE_GPU_FALLBACK or ALLOW_CPU_MODEL_FALLBACK or ALLOW_MODEL_DOWNLOAD:
    raise RuntimeError("serving fallback/download policy drift")
if not torch.cuda.is_available() or torch.cuda.device_count()<REQUIRED_GPU_COUNT:
    raise RuntimeError(("STRICT_DUAL_GPU_REQUIRED",torch.cuda.device_count() if torch.cuda.is_available() else 0))

QUERY_LOCK=threading.Lock()
QUERY_USAGE={
    d:{"contract_calls":0,"query_views":0,"long_query_contract_calls":0}
    for d in MODEL_DEVICES
}

def unwrap(value):
    if isinstance(value,torch.Tensor): t=value
    elif hasattr(value,"pooler_output") and value.pooler_output is not None: t=value.pooler_output
    elif hasattr(value,"last_hidden_state"): t=value.last_hidden_state[:,0]
    elif isinstance(value,(tuple,list)) and value and isinstance(value[0],torch.Tensor): t=value[0]
    else: raise TypeError(type(value))
    if t.ndim!=2: raise ValueError(t.shape)
    return t

def floating_dtype_summary(model):
    out={}
    for p in model.parameters():
        if torch.is_floating_point(p):
            out[str(p.dtype)]=out.get(str(p.dtype),0)+int(p.numel())
    return dict(sorted(out.items()))

def normalize_runtime_model(model,device,target_dtype=torch.float16):
    model=model.to(device=device,dtype=target_dtype).eval()
    bad=[]
    for name,p in model.named_parameters():
        if torch.is_floating_point(p) and (p.dtype!=target_dtype or str(p.device)!=str(device)):
            bad.append(("param",name,str(p.dtype),str(p.device)))
            if len(bad)>=20: break
    for name,b in model.named_buffers():
        if torch.is_floating_point(b) and (b.dtype!=target_dtype or str(b.device)!=str(device)):
            bad.append(("buffer",name,str(b.dtype),str(b.device)))
            if len(bad)>=20: break
    if bad:
        raise RuntimeError(("SIGLIP2_RUNTIME_DTYPE_DEVICE_CONTRACT_FAILED",device,bad))
    summary=floating_dtype_summary(model)
    if set(summary)!={"torch.float16"}:
        raise RuntimeError(("FULL_FP16_RUNTIME_REQUIRED",device,summary))
    return model,summary

def resolve_text_limit(model,processor):
    vals=[]
    tc=getattr(getattr(model,"config",None),"text_config",None)
    v=getattr(tc,"max_position_embeddings",None)
    if isinstance(v,int) and v>0: vals.append(int(v))
    tok=getattr(processor,"tokenizer",None)
    v=getattr(tok,"model_max_length",None)
    if isinstance(v,int) and 1<=v<=100000: vals.append(int(v))
    if not vals: raise RuntimeError("finite text token limit unavailable")
    return min(vals)

def bounded_text_batch(processor,text,max_tokens):
    tok=processor.tokenizer
    raw=tok(str(text),add_special_tokens=False,truncation=False)
    ids=raw["input_ids"]
    if ids and isinstance(ids[0],list): ids=ids[0]
    try: special=int(tok.num_special_tokens_to_add(pair=False))
    except Exception: special=2
    body=max(1,int(max_tokens)-special)
    truncated=len(ids)>body
    model_text=str(text)
    if truncated:
        head=(body+1)//2; tail=body-head
        kept=ids[:head]+(ids[-tail:] if tail else [])
        model_text=tok.decode(kept,skip_special_tokens=True,clean_up_tokenization_spaces=False)
    batch=processor(
        text=[model_text],padding="max_length",truncation=True,
        max_length=int(max_tokens),return_tensors="pt"
    )
    return batch,{"truncated":bool(truncated),"raw_body_tokens":len(ids),"body_budget":body}

def load_query_worker(gpu):
    device=f"cuda:{gpu}"
    torch.cuda.set_device(gpu)
    proc=AutoProcessor.from_pretrained(str(VISUAL_MODEL_DIR),local_files_only=True,use_fast=True)
    model=AutoModel.from_pretrained(
        str(VISUAL_MODEL_DIR),local_files_only=True,
        dtype=torch.float16,low_cpu_mem_usage=True
    )
    model,dtypes=normalize_runtime_model(model,device,torch.float16)
    limit=resolve_text_limit(model,proc)
    batch,_=bounded_text_batch(proc,"final runtime dual gpu contract",limit)
    inputs={k:v.to(device) for k,v in batch.items()}
    with torch.inference_mode():
        vec=torch.nn.functional.normalize(unwrap(model.get_text_features(**inputs)).float(),dim=-1)
    torch.cuda.synchronize(gpu)
    if tuple(vec.shape)!=(1,FEATURE_DIM) or not torch.isfinite(vec).all():
        raise AssertionError(("contract vector",device,tuple(vec.shape)))
    with QUERY_LOCK: QUERY_USAGE[device]["contract_calls"]+=1
    return {
        "gpu":gpu,"device":device,"processor":proc,"model":model,
        "text_limit":limit,"floating_dtype_summary":dtypes,
        "gpu_name":torch.cuda.get_device_name(gpu),
    }

with ThreadPoolExecutor(max_workers=2) as ex:
    QUERY_WORKERS=list(ex.map(load_query_worker,GPU_INDICES))
if {w["device"] for w in QUERY_WORKERS}!=set(MODEL_DEVICES):
    raise AssertionError("dual GPU worker set")

# Long-query regression on both GPUs.
_LONG=" ".join([f"final-long-token-{i:03d}" for i in range(180)])
def long_contract(w):
    torch.cuda.set_device(w["gpu"])
    batch,fit=bounded_text_batch(w["processor"],_LONG,w["text_limit"])
    if not fit["truncated"]:
        raise AssertionError(("long query did not truncate",fit))
    inputs={k:v.to(w["device"]) for k,v in batch.items()}
    with torch.inference_mode():
        vec=torch.nn.functional.normalize(unwrap(w["model"].get_text_features(**inputs)).float(),dim=-1)
    torch.cuda.synchronize(w["gpu"])
    if tuple(vec.shape)!=(1,FEATURE_DIM): raise AssertionError(vec.shape)
    with QUERY_LOCK: QUERY_USAGE[w["device"]]["long_query_contract_calls"]+=1
    return {"device":w["device"],**fit}

with ThreadPoolExecutor(max_workers=2) as ex:
    LONG_QUERY_CONTRACT=list(ex.map(long_contract,QUERY_WORKERS))

print("[PASS] strict dual-GPU query service",{
    "workers":[{"device":w["device"],"gpu_name":w["gpu_name"],"dtypes":w["floating_dtype_summary"]} for w in QUERY_WORKERS],
    "long_query_contract":LONG_QUERY_CONTRACT,
})

# v0.2.2 exact serving matrix:
# Frozen Stage03 stores visual_embeddings.npy as float16, then constructs its
# reference FAISS IndexFlatIP from np.asarray(chunk, dtype=float32).
# Mirror that contract exactly: preserve the stored FP16 values and widen them
# to FP32 on each GPU before any dot product.
if STAGE03_STORAGE_DTYPE!="float16" or str(EMBEDDINGS.dtype)!="float16":
    raise RuntimeError((
        "STAGE03_STORAGE_DTYPE_CONTRACT_FAILED",
        STAGE03_STORAGE_DTYPE,
        str(EMBEDDINGS.dtype),
        "expected float16",
    ))

GPU_VISUAL_EMBEDDINGS={}
GPU_VISUAL_INDEX_INFO={}
for w in QUERY_WORKERS:
    dev=w["device"]
    gpu=w["gpu"]
    torch.cuda.set_device(gpu)
    before=int(torch.cuda.memory_allocated(gpu))
    # Materialize the same widened values used by Stage03 FAISS construction.
    # np.asarray(..., dtype=np.float32) explicitly performs FP16 -> FP32 widening
    # on CPU; torch then transfers those exact float32 values to the target GPU.
    cpu_f32=np.asarray(EMBEDDINGS,dtype=np.float32)
    cpu_view=torch.from_numpy(cpu_f32)
    mat=cpu_view.to(device=dev,dtype=torch.float32,non_blocking=False).contiguous()
    torch.cuda.synchronize(gpu)
    if tuple(mat.shape)!=(VISUAL_ROWS,FEATURE_DIM) or mat.dtype!=torch.float32 or str(mat.device)!=dev:
        raise AssertionError(("GPU_VISUAL_MATRIX_CONTRACT",dev,tuple(mat.shape),str(mat.dtype),str(mat.device)))
    if not torch.isfinite(mat[:min(1024,VISUAL_ROWS)]).all():
        raise AssertionError(("GPU_VISUAL_MATRIX_NONFINITE_SAMPLE",dev))
    GPU_VISUAL_EMBEDDINGS[dev]=mat
    after=int(torch.cuda.memory_allocated(gpu))
    GPU_VISUAL_INDEX_INFO[dev]={
        "rows":VISUAL_ROWS,
        "feature_dim":FEATURE_DIM,
        "source_storage_dtype":STAGE03_STORAGE_DTYPE,
        "serving_dtype":"torch.float32",
        "reference_semantics":"STAGE03_FP16_STORED_VALUES_WIDENED_TO_FP32",
        "memory_delta_bytes":after-before,
    }

if set(GPU_VISUAL_EMBEDDINGS)!=set(MODEL_DEVICES):
    raise AssertionError(("GPU_VISUAL_REPLICA_SET",GPU_VISUAL_EMBEDDINGS.keys()))

VISUAL_SEARCH_BACKEND="DUAL_GPU_TORCH_EXACT_FLOAT32"

# Actual-score regression:
# compare one deterministic full-corpus GPU top-k against NumPy reference scores.
_ref_rows=np.arange(VISUAL_ROWS,dtype=np.int64)
_ref_matrix=np.asarray(EMBEDDINGS,dtype=np.float32)
_ref_q=np.asarray(EMBEDDINGS[12345],dtype=np.float32)
_ref_q=_ref_q/max(float(np.linalg.norm(_ref_q)),1e-12)
_ref_scores=_ref_matrix@_ref_q
_ref_k=20
_ref_idx=np.argpartition(_ref_scores,-_ref_k)[-_ref_k:]
_ref_idx=_ref_idx[np.lexsort((_ref_idx,-_ref_scores[_ref_idx]))]

GPU_VISUAL_REGRESSION=[]
for w in QUERY_WORKERS:
    dev=w["device"]; gpu=w["gpu"]
    torch.cuda.set_device(gpu)
    q=torch.from_numpy(_ref_q).to(dev,dtype=torch.float32)
    with torch.inference_mode():
        scores=torch.mv(GPU_VISUAL_EMBEDDINGS[dev],q)
        vals,idx=torch.topk(scores,k=_ref_k,largest=True,sorted=False)
    torch.cuda.synchronize(gpu)
    vals=vals.cpu().numpy().astype(np.float32)
    idx=idx.cpu().numpy().astype(np.int64)
    order=np.lexsort((idx,-vals))
    idx=idx[order]; vals=vals[order]
    if not np.array_equal(idx,_ref_idx):
        raise AssertionError(("GPU_EXACT_TOPK_MISMATCH",dev,idx.tolist(),_ref_idx.tolist()))
    max_err=float(np.max(np.abs(vals-_ref_scores[idx])))
    if max_err>2e-5:
        raise AssertionError(("GPU_EXACT_SCORE_ERROR",dev,max_err))
    GPU_VISUAL_REGRESSION.append({"device":dev,"topk_exact":True,"max_abs_score_error":max_err})

del _ref_scores, _ref_idx, _ref_q, _ref_rows, _ref_matrix, cpu_view, cpu_f32
print("[PASS] dual-GPU exact Stage03 FP16→FP32 visual replicas",{
    "backend":VISUAL_SEARCH_BACKEND,
    "replicas":GPU_VISUAL_INDEX_INFO,
    "regression":GPU_VISUAL_REGRESSION,
})
````

````python
# ── F3 / CELL 9: DUAL-VIEW QUERY ENCODING + DUAL-GPU EXACT VISUAL SEARCH ─────
# Mục tiêu: low-latency exact KIS serving kernel.
# INPUT: arbitrary KIS query.
# OUTPUT: exact RAW baseline + prompt-view candidate support.
# Pre-check: both GPUs participate in encoding and exact full-corpus search.
# Post-check: active ranking remains RAW-query exact cosine from Stage03 float32 values.
# History v0.2.0: CPU NumPy exact fallback was correct but too slow.
# History v0.2.1: each view searches a full FP32 serving replica on its own GPU in parallel.
# History v0.2.2: replica values come from Stage03 stored FP16 vectors widened to FP32, matching FAISS build semantics.
# ──────────────────────────────────────────────────────────────────────────────
def canonical_query_text(query):
    if not isinstance(query,str): raise TypeError("query must be str")
    q=" ".join(query.strip().split())
    if not q: raise ValueError("empty query")
    if len(q)>8000: raise ValueError("query too long")
    return q

def query_views(query):
    return [
        {"view_id":"RAW","text":query,"worker_index":0},
        {"view_id":"VISUAL_PROMPT","text":f"A photo that best matches this search description: {query}","worker_index":1},
    ]

def encode_view(view):
    w=QUERY_WORKERS[int(view["worker_index"])]
    torch.cuda.set_device(w["gpu"])
    batch,fit=bounded_text_batch(w["processor"],view["text"],w["text_limit"])
    inputs={k:v.to(w["device"]) for k,v in batch.items()}
    with torch.inference_mode():
        vec=torch.nn.functional.normalize(
            unwrap(w["model"].get_text_features(**inputs)).float(),dim=-1
        )
    torch.cuda.synchronize(w["gpu"])
    if tuple(vec.shape)!=(1,FEATURE_DIM) or not torch.isfinite(vec).all():
        raise AssertionError((view["view_id"],tuple(vec.shape)))
    with QUERY_LOCK:
        QUERY_USAGE[w["device"]]["query_views"]+=1
    return {
        **view,
        "device":w["device"],
        "gpu":w["gpu"],
        "vector":vec.cpu().numpy().astype(np.float32),
        "fit":fit,
    }

def encode_dualgpu(query):
    q=canonical_query_text(query)
    t0=time.perf_counter()
    with ThreadPoolExecutor(max_workers=2) as ex:
        out=list(ex.map(encode_view,query_views(q)))
    out=sorted(out,key=lambda x:x["worker_index"])
    if {x["device"] for x in out}!=set(MODEL_DEVICES):
        raise AssertionError("both GPUs must participate")
    return q,out,(time.perf_counter()-t0)*1000.0

def exact_visual_search_gpu(view,k):
    k=min(int(k),VISUAL_ROWS)
    dev=view["device"]; gpu=int(view["gpu"])
    torch.cuda.set_device(gpu)
    q=torch.from_numpy(view["vector"][0]).to(device=dev,dtype=torch.float32,non_blocking=True)
    with torch.inference_mode():
        scores=torch.mv(GPU_VISUAL_EMBEDDINGS[dev],q)
        vals,idx=torch.topk(scores,k=k,largest=True,sorted=False)
    torch.cuda.synchronize(gpu)
    vals=vals.cpu().numpy().astype(np.float32)
    idx=idx.cpu().numpy().astype(np.int64)
    # deterministic descending score, then ascending index tie-break.
    order=np.lexsort((idx,-vals))
    return vals[order].astype(float),idx[order]

def rrf_union(search_outputs,k=RRF_K):
    scores={}; ranks={}
    for vid,(D,I) in search_outputs.items():
        rr={}
        for r0,row in enumerate(I.tolist()):
            row=int(row)
            if row<0: continue
            r=r0+1; rr[row]=r
            scores[row]=scores.get(row,0.0)+1.0/(k+r)
        ranks[vid]=rr
    ordered=sorted(scores,key=lambda x:(-scores[x],x))
    return ordered,scores,ranks

def search_kis_pool(encoded,search_k=K_SEED):
    t0=time.perf_counter()
    with ThreadPoolExecutor(max_workers=2) as ex:
        results=list(ex.map(lambda v:exact_visual_search_gpu(v,search_k),encoded))
    outputs={v["view_id"]:res for v,res in zip(encoded,results)}
    search_ms=(time.perf_counter()-t0)*1000.0

    rows,rrf,ranks=rrf_union(outputs,RRF_K)
    rows=rows[:min(len(rows),K_MAX)]

    raw_vec=encoded[0]["vector"][0]
    prompt_vec=encoded[1]["vector"][0]
    mat=np.asarray(EMBEDDINGS[rows],dtype=np.float32)
    raw=mat@raw_vec
    prompt=mat@prompt_vec
    fused=0.5*raw+0.5*prompt

    rec=[]
    for j,row in enumerate(rows):
        rec.append({
            "index_row":int(row),
            "raw_cosine":float(raw[j]),
            "prompt_cosine":float(prompt[j]),
            "dual_view_fused_cosine":float(fused[j]),
            "rrf_pool_score":float(rrf[row]),
            "raw_rank_in_search":ranks["RAW"].get(row),
            "prompt_rank_in_search":ranks["VISUAL_PROMPT"].get(row),
        })
    base=pd.DataFrame(rec)
    if base.empty: raise RuntimeError("EMPTY_VISUAL_POOL")

    raw_ranked=base.sort_values(
        ["raw_cosine","index_row"],ascending=[False,True],kind="mergesort"
    ).reset_index(drop=True)
    raw_ranked["raw_final_rank"]=np.arange(1,len(raw_ranked)+1)

    fused_ranked=base.sort_values(
        ["dual_view_fused_cosine","raw_cosine","index_row"],
        ascending=[False,False,True],kind="mergesort"
    ).reset_index(drop=True)
    fused_ranked["fused_final_rank"]=np.arange(1,len(fused_ranked)+1)
    fmap=dict(zip(fused_ranked.index_row.astype(int),fused_ranked.fused_final_rank.astype(int)))
    raw_ranked["fused_final_rank"]=raw_ranked.index_row.map(fmap).astype(int)

    out=raw_ranked.merge(VISUAL_META,on="index_row",how="left",validate="one_to_one")
    if out.frame_ref_id.isna().any():
        raise AssertionError("visual metadata join")
    return out,{
        "search_ms":search_ms,
        "pool_rows":len(out),
        "visual_search_backend":VISUAL_SEARCH_BACKEND,
    }

print("[PASS] dual-GPU exact visual serving kernels ready",VISUAL_SEARCH_BACKEND,{"stage03_storage_dtype":STAGE03_STORAGE_DTYPE,"serving_dtype":"float32"})
````

````python
# ── O2 / CELL: DUAL-GPU EXACT ADAPTIVE SHADOW REPLICAS ────────────────────────
# Mục tiêu: exact Stage06 adaptive search using the same RAW/PROMPT query vectors.
# INPUT: Stage06 FP16 stored embeddings.
# OUTPUT: full FP32 adaptive matrix replica on cuda:0 and cuda:1.
# Pre-check: Stage06 is proposal-only; no canonical promotion.
# Post-check: both GPUs hold exact widened stored values; sample dot regression passes.
# History v0.5.0: first online adaptive visual shadow serving path.
# ──────────────────────────────────────────────────────────────────────────────

GPU_ADAPTIVE_EMBEDDINGS={}
GPU_ADAPTIVE_INDEX_INFO={}
O2_GPU_USAGE={d:{"adaptive_search_calls":0,"adaptive_query_views":0} for d in MODEL_DEVICES}
O2_GPU_LOCK=threading.Lock()

for _gpu,_dev in zip(GPU_INDICES,MODEL_DEVICES):
    torch.cuda.set_device(_gpu)
    _before=torch.cuda.memory_allocated(_gpu)
    _dst=torch.empty((len(ADAPTIVE_META),FEATURE_DIM),device=_dev,dtype=torch.float32)
    for _s in range(0,len(ADAPTIVE_META),12000):
        _e=min(len(ADAPTIVE_META),_s+12000)
        _chunk=np.asarray(ADAPTIVE_EMB[_s:_e],dtype=np.float32)
        _dst[_s:_e].copy_(torch.from_numpy(_chunk).to(_dev),non_blocking=False)
    torch.cuda.synchronize(_gpu)
    if tuple(_dst.shape)!=(48656,FEATURE_DIM) or not torch.isfinite(_dst).all():
        raise AssertionError(("adaptive GPU replica",_dev,tuple(_dst.shape)))
    GPU_ADAPTIVE_EMBEDDINGS[_dev]=_dst
    GPU_ADAPTIVE_INDEX_INFO[_dev]={
        "rows":int(_dst.shape[0]),
        "feature_dim":int(_dst.shape[1]),
        "source_storage_dtype":"float16",
        "serving_dtype":"torch.float32",
        "reference_semantics":"STAGE06_FP16_STORED_VALUES_WIDENED_TO_FP32",
        "memory_delta_bytes":int(torch.cuda.memory_allocated(_gpu)-_before),
    }

# Sample exact-dot regression on real stored vectors, no GT semantics.
_q=np.asarray(ADAPTIVE_EMB[0],dtype=np.float32)
_ref=np.asarray(ADAPTIVE_EMB[:2048],dtype=np.float32)@_q
O2_ADAPTIVE_GPU_REGRESSION=[]
for _gpu,_dev in zip(GPU_INDICES,MODEL_DEVICES):
    torch.cuda.set_device(_gpu)
    _qt=torch.from_numpy(_q).to(_dev,dtype=torch.float32)
    with torch.inference_mode():
        _got=torch.mv(GPU_ADAPTIVE_EMBEDDINGS[_dev][:_ref.shape[0]],_qt)
    torch.cuda.synchronize(_gpu)
    _err=float(np.max(np.abs(_got.cpu().numpy()-_ref)))
    if _err>2e-5:
        raise AssertionError(("adaptive dot regression",_dev,_err))
    O2_ADAPTIVE_GPU_REGRESSION.append({"device":_dev,"max_abs_score_error":_err})

print("[PASS] O2 dual-GPU adaptive exact replicas",{
    "replicas":GPU_ADAPTIVE_INDEX_INFO,
    "regression":O2_ADAPTIVE_GPU_REGRESSION,
    "promotion":ADAPTIVE_PROMOTION_STATUS,
})
````

````python
# ── F4 / CELL 10: CACHED STAGE20 OCR/ASR COMPACT RETRIEVAL ───────────────────
# Mục tiêu: low-overhead query-conditioned compact textual evidence branch.
# INPUT: query + Stage20 pack.
# OUTPUT: interval lexical scores; no probability.
# Pre-check: compact OCR/ASR strings only; no fake object text.
# Post-check: interval authority remains technical/shadow until raw PTS verification.
# History v0.2.0: per-query Python normalization/tokenization across 18,660 rows.
# History v0.2.1: pretokenize once; query path only performs set intersections + vectorized ranking.
# ──────────────────────────────────────────────────────────────────────────────
def norm_text(s):
    s=unicodedata.normalize("NFKC",str(s)).casefold()
    s=re.sub(r"[^\w]+"," ",s,flags=re.UNICODE)
    return " ".join(s.split())

def token_set(s):
    return frozenset(norm_text(s).split())

PACK_TEXT_BASE=PACK[[
    "evidence_pack_id","variant","shadow_shot_interval_id","video_id",
    "start_pts_time_technical","end_pts_time_technical","duration_sec",
    "ocr_text_compact","asr_text_compact","available_modality_count",
]].copy()
PACK_TEXT_BASE["shadow_shot_interval_id"]=PACK_TEXT_BASE.shadow_shot_interval_id.astype(str)

_t0=time.perf_counter()
OCR_TOKEN_CACHE=tuple(token_set(x) for x in PACK_TEXT_BASE.ocr_text_compact.fillna("").astype(str))
ASR_TOKEN_CACHE=tuple(token_set(x) for x in PACK_TEXT_BASE.asr_text_compact.fillna("").astype(str))
TEXT_CACHE_BUILD_MS=(time.perf_counter()-_t0)*1000.0

if len(OCR_TOKEN_CACHE)!=len(PACK_TEXT_BASE) or len(ASR_TOKEN_CACHE)!=len(PACK_TEXT_BASE):
    raise AssertionError("text cache size")

def coverage_from_cache(qtokens,cache):
    den=max(1,len(qtokens))
    return np.fromiter(
        (len(qtokens & toks)/den if toks else 0.0 for toks in cache),
        dtype=np.float32,count=len(cache)
    )

def rank_pct_nonzero(values):
    x=pd.to_numeric(pd.Series(values),errors="coerce").astype(float)
    x=x.where(x>0,np.nan)
    out=pd.Series(np.nan,index=x.index,dtype=float)
    m=x.notna()
    if not m.any(): return out
    if x.loc[m].nunique()==1: out.loc[m]=0.5
    else: out.loc[m]=x.loc[m].rank(method="average",pct=True)
    return out

def retrieve_compact_text_intervals(query,top_k=LEXICAL_POOL_K):
    q=canonical_query_text(query)
    qtokens=token_set(q)
    x=PACK_TEXT_BASE.copy()
    x["ocr_score_raw"]=coverage_from_cache(qtokens,OCR_TOKEN_CACHE)
    x["asr_score_raw"]=coverage_from_cache(qtokens,ASR_TOKEN_CACHE)
    x["ocr_query_pct"]=rank_pct_nonzero(x.ocr_score_raw)
    x["asr_query_pct"]=rank_pct_nonzero(x.asr_score_raw)
    x["text_available_count"]=x[["ocr_query_pct","asr_query_pct"]].notna().sum(axis=1)
    x["text_semantic_normalized_only"]=x[["ocr_query_pct","asr_query_pct"]].mean(axis=1,skipna=True)
    x.loc[x.text_available_count.eq(0),"text_semantic_normalized_only"]=np.nan
    x=x.sort_values(
        ["text_semantic_normalized_only","ocr_score_raw","asr_score_raw","shadow_shot_interval_id"],
        ascending=[False,False,False,True],na_position="last",kind="mergesort"
    ).head(int(top_k)).reset_index(drop=True)
    x["text_rank"]=np.arange(1,len(x)+1)
    x["probability_semantics"]="NONE"
    x["temporal_status"]="SHADOW_TECHNICAL_INTERVAL_REQUIRES_RAW_PTS_VERIFIER_FOR_FINAL_TRAKE"
    return x

print("[PASS] cached OCR/ASR compact interval retriever",{
    "packs":len(PACK_TEXT_BASE),
    "cache_build_ms":TEXT_CACHE_BUILD_MS,
})
````

````python
# ── F4 / CELL 11: INDEXED FRAME↔INTERVAL BRIDGE + MULTIMODAL POOL ─────────────
# Mục tiêu: bridge Stage03 official visual hits into Stage20 shadow intervals.
# INPUT: visual KIS pool + compact text scores.
# OUTPUT: evidence pool preserving source provenance.
# Pre-check: no 177k-group Python dictionary build.
# Post-check: RRF/cross-modal normalized score is candidate support, not probability.
# History v0.2.0: Python groupby loop built ~177k FrameRef dictionary and cost ~206 s startup.
# History v0.2.1: use indexed Stage20 frame payload and vectorized per-query joins.
# ──────────────────────────────────────────────────────────────────────────────
FRAME_INTERVAL_INDEX=FRAME_PAYLOAD[[
    "frame_ref_id","variant","shadow_shot_interval_id","video_id"
]].drop_duplicates().copy()
FRAME_INTERVAL_INDEX["frame_ref_id"]=FRAME_INTERVAL_INDEX.frame_ref_id.astype(str)
FRAME_INTERVAL_INDEX["shadow_shot_interval_id"]=FRAME_INTERVAL_INDEX.shadow_shot_interval_id.astype(str)
FRAME_INTERVAL_INDEX=FRAME_INTERVAL_INDEX.set_index("frame_ref_id",drop=False).sort_index()

PACK_INTERVAL_BASE=PACK[[
    "evidence_pack_id","variant","shadow_shot_interval_id","video_id",
    "start_pts_time_technical","end_pts_time_technical",
    "ocr_text_compact","asr_text_compact",
]].rename(columns={"shadow_shot_interval_id":"interval_id"}).copy()
PACK_INTERVAL_BASE["interval_id"]=PACK_INTERVAL_BASE.interval_id.astype(str)

def interval_pool_from_visual(visual_pool,top_frames=VISUAL_SEARCH_K):
    v=visual_pool.head(int(top_frames))[[
        "frame_ref_id","video_id","raw_cosine","prompt_cosine","raw_final_rank"
    ]].copy()
    v["frame_ref_id"]=v.frame_ref_id.astype(str)
    ids=v.frame_ref_id.tolist()
    present=[x for x in ids if x in FRAME_INTERVAL_INDEX.index]
    if not present:
        return pd.DataFrame(columns=[
            "interval_id","variant","video_id","visual_raw_score",
            "visual_prompt_score","raw_frame_rank","best_frame_ref_id"
        ])
    rel=FRAME_INTERVAL_INDEX.loc[present].reset_index(drop=True)
    rel=rel.merge(v,on=["frame_ref_id","video_id"],how="inner",validate="many_to_one")
    rel=rel.rename(columns={
        "shadow_shot_interval_id":"interval_id",
        "raw_cosine":"visual_raw_score",
        "prompt_cosine":"visual_prompt_score",
        "raw_final_rank":"raw_frame_rank",
    })
    rel=rel.sort_values(
        ["visual_raw_score","raw_frame_rank","frame_ref_id"],
        ascending=[False,True,True],kind="mergesort"
    )
    return rel.groupby(["interval_id","variant","video_id"],as_index=False).agg(
        visual_raw_score=("visual_raw_score","max"),
        visual_prompt_score=("visual_prompt_score","max"),
        raw_frame_rank=("raw_frame_rank","min"),
        best_frame_ref_id=("frame_ref_id","first"),
    )

def build_multimodal_interval_pool(query,visual_pool):
    vis=interval_pool_from_visual(visual_pool)
    txt=retrieve_compact_text_intervals(
        query,top_k=max(LEXICAL_POOL_K,INTERVAL_POOL_K)
    ).rename(columns={"shadow_shot_interval_id":"interval_id"})
    txt["interval_id"]=txt.interval_id.astype(str)

    ids=set(vis.interval_id.astype(str)) | set(txt.interval_id.astype(str))
    pool=PACK_INTERVAL_BASE[PACK_INTERVAL_BASE.interval_id.isin(ids)].copy()

    pool=pool.merge(
        vis[[
            "interval_id","variant","video_id","visual_raw_score",
            "visual_prompt_score","raw_frame_rank","best_frame_ref_id"
        ]],
        on=["interval_id","variant","video_id"],how="left",validate="one_to_one"
    )
    pool=pool.merge(
        txt[[
            "interval_id","variant","video_id","ocr_score_raw","asr_score_raw",
            "ocr_query_pct","asr_query_pct","text_semantic_normalized_only"
        ]],
        on=["interval_id","variant","video_id"],how="left",validate="one_to_one"
    )

    pool["visual_query_pct"]=rank_pct_nonzero(pool.visual_raw_score)
    sem=["visual_query_pct","ocr_query_pct","asr_query_pct"]
    pool["semantic_available_count"]=pool[sem].notna().sum(axis=1)
    pool["semantic_score_normalized_only"]=pool[sem].mean(axis=1,skipna=True)
    pool.loc[pool.semantic_available_count.eq(0),"semantic_score_normalized_only"]=np.nan
    pool=pool.sort_values(
        ["semantic_score_normalized_only","visual_raw_score","interval_id"],
        ascending=[False,False,True],na_position="last",kind="mergesort"
    ).head(INTERVAL_POOL_K).reset_index(drop=True)
    pool["semantic_rank"]=np.arange(1,len(pool)+1)
    pool["probability_semantics"]="NONE"
    return pool

print("[PASS] indexed multimodal interval bridge ready",{
    "frame_interval_rows":len(FRAME_INTERVAL_INDEX),
    "unique_frame_refs":int(FRAME_INTERVAL_INDEX.index.nunique()),
})
````

````python
# ── O2 / CELL: PARALLEL FEDERATED RETRIEVAL + INTERVAL UNION ──────────────────
# Mục tiêu: wire adaptive visual + object + typed motion into Stage20 interval evidence.
# INPUT: query, encoded RAW/PROMPT views, frozen official visual pool.
# OUTPUT: O2 federated interval pool with source provenance and normalized-only rerank.
# Pre-check: object/motion/adaptive sources never claim calibrated correctness.
# Post-check: unavailable source != zero score; equal-source percentile mean is explicitly heuristic.
# History v0.5.0: first full Online O2 candidate federation.
# History v0.5.3: typed timed-result unpack helper fixes ThreadPool future contract.
# ──────────────────────────────────────────────────────────────────────────────

PACK_BY_VIDEO={
    str(vid):g.reset_index(drop=True).copy()
    for vid,g in PACK_INTERVAL_BASE.groupby("video_id",sort=False)
}

# ---- Stage06 exact adaptive semantic retrieval --------------------------------
def exact_adaptive_search_gpu(view,k=O2_ADAPTIVE_K):
    k=min(int(k),len(ADAPTIVE_META))
    dev=view["device"]
    gpu=int(view["gpu"])
    torch.cuda.set_device(gpu)
    q=torch.from_numpy(view["vector"][0]).to(device=dev,dtype=torch.float32,non_blocking=True)
    with torch.inference_mode():
        scores=torch.mv(GPU_ADAPTIVE_EMBEDDINGS[dev],q)
        vals,idx=torch.topk(scores,k=k,largest=True,sorted=False)
    torch.cuda.synchronize(gpu)
    vals=vals.cpu().numpy().astype(np.float32)
    idx=idx.cpu().numpy().astype(np.int64)
    order=np.lexsort((idx,-vals))
    with O2_GPU_LOCK:
        O2_GPU_USAGE[dev]["adaptive_search_calls"]+=1
        O2_GPU_USAGE[dev]["adaptive_query_views"]+=1
    return vals[order].astype(float),idx[order]

def search_adaptive_pool(encoded,k=O2_ADAPTIVE_K):
    t0=time.perf_counter()
    with ThreadPoolExecutor(max_workers=2) as ex:
        outs=list(ex.map(lambda v:exact_adaptive_search_gpu(v,k),encoded))
    outputs={v["view_id"]:o for v,o in zip(encoded,outs)}
    union,rrf,ranks=rrf_union(outputs,RRF_K)
    union=union[:min(len(union),max(int(k),RETURN_TOPK))]

    raw_vec=encoded[0]["vector"][0]
    prompt_vec=encoded[1]["vector"][0]
    mat=np.asarray(ADAPTIVE_EMB[union],dtype=np.float32)
    raw=mat@raw_vec
    prompt=mat@prompt_vec

    rec=[]
    for j,row in enumerate(union):
        rec.append({
            "adaptive_index_row":int(row),
            "adaptive_raw_cosine":float(raw[j]),
            "adaptive_prompt_cosine":float(prompt[j]),
            "adaptive_rrf_pool_score":float(rrf[row]),
            "adaptive_raw_rank_in_search":ranks["RAW"].get(row),
            "adaptive_prompt_rank_in_search":ranks["VISUAL_PROMPT"].get(row),
        })
    q=pd.DataFrame(rec)
    if q.empty:
        return q,{"search_ms":(time.perf_counter()-t0)*1000.0,"rows":0}
    q=q.sort_values(
        ["adaptive_raw_cosine","adaptive_index_row"],
        ascending=[False,True],kind="mergesort",
    ).reset_index(drop=True)
    q["adaptive_rank"]=np.arange(1,len(q)+1)
    q=q.merge(ADAPTIVE_META,on="adaptive_index_row",how="left",validate="one_to_one")
    return q,{
        "search_ms":(time.perf_counter()-t0)*1000.0,
        "rows":len(q),
        "promotion_status":ADAPTIVE_PROMOTION_STATUS,
    }

def adaptive_interval_support(encoded):
    ap,diag=search_adaptive_pool(encoded,O2_ADAPTIVE_K)
    if ap.empty:
        return pd.DataFrame(),diag
    rows=[]
    for r in ap.head(O2_ADAPTIVE_K).itertuples(index=False):
        vid=str(r.video_id)
        g=PACK_BY_VIDEO.get(vid)
        if g is None or g.empty:
            continue
        t=float(r.actual_pts_time if pd.notna(r.actual_pts_time) else r.pts_time)
        m=(
            pd.to_numeric(g.start_pts_time_technical,errors="coerce").le(t+SEEK_TOLERANCE_S)
            &pd.to_numeric(g.end_pts_time_technical,errors="coerce").ge(t-SEEK_TOLERANCE_S)
        )
        for z in g.loc[m].itertuples(index=False):
            rows.append({
                "interval_id":str(z.interval_id),
                "variant":str(z.variant),
                "video_id":vid,
                "adaptive_raw_cosine":float(r.adaptive_raw_cosine),
                "adaptive_prompt_cosine":float(r.adaptive_prompt_cosine),
                "adaptive_rank":int(r.adaptive_rank),
                "best_adaptive_frame_ref_id":str(r.frame_ref_id),
                "adaptive_pts_time":t,
            })
    if not rows:
        return pd.DataFrame(),diag
    d=pd.DataFrame(rows).sort_values(
        ["adaptive_raw_cosine","adaptive_rank","best_adaptive_frame_ref_id"],
        ascending=[False,True,True],kind="mergesort",
    )
    d=d.groupby(["interval_id","variant","video_id"],as_index=False).agg(
        adaptive_raw_cosine=("adaptive_raw_cosine","max"),
        adaptive_prompt_cosine=("adaptive_prompt_cosine","max"),
        adaptive_rank=("adaptive_rank","min"),
        best_adaptive_frame_ref_id=("best_adaptive_frame_ref_id","first"),
        adaptive_pts_time=("adaptive_pts_time","first"),
    )
    d["adaptive_query_pct"]=rank_pct_nonzero(d.adaptive_raw_cosine)
    return d,diag

# ---- Stage11 object lexical matcher + exact frame-class score matrix -----------
def _query_token_tuple(s):
    return tuple(norm_text(s).split())

_OBJ_ALIAS_TO_CLASSES={}
for r in OBJ_ALIAS.itertuples(index=False):
    alias=str(r.alias_norm).strip()
    if alias:
        _OBJ_ALIAS_TO_CLASSES.setdefault(alias,set()).add(int(r.class_idx))
OBJ_ALIAS_TOKEN_ROWS=[
    (a,_query_token_tuple(a),tuple(sorted(c)))
    for a,c in _OBJ_ALIAS_TO_CLASSES.items()
]
OBJ_ALIAS_TOKEN_ROWS.sort(key=lambda z:(-len(z[1]),z[0]))

def query_contains_alias(query_tokens,alias_tokens):
    n=len(alias_tokens)
    if n==0 or n>len(query_tokens):
        return False
    return any(
        tuple(query_tokens[i:i+n])==tuple(alias_tokens)
        for i in range(len(query_tokens)-n+1)
    )

def object_matched_classes(query):
    qt=_query_token_tuple(query)
    aliases=[]
    classes=set()
    for alias,at,cis in OBJ_ALIAS_TOKEN_ROWS:
        if query_contains_alias(qt,at):
            aliases.append(alias)
            classes.update(cis)
    if len(classes)>O2_OBJECT_CLASS_HARD_CAP:
        raise RuntimeError(("O2_OBJECT_CLASS_HARD_CAP",len(classes),aliases[:20]))
    return sorted(classes),aliases

def object_search_frames(query,k=O2_OBJECT_K):
    t0=time.perf_counter()
    classes,aliases=object_matched_classes(query)
    if not classes:
        return pd.DataFrame(),{
            "search_ms":(time.perf_counter()-t0)*1000.0,
            "matched_aliases":[],
            "matched_classes":[],
            "status":"NO_EXPLICIT_OBJECT_ALIAS",
        }
    scores=np.zeros(VISUAL_ROWS,dtype=np.float32)
    for ci in classes:
        np.maximum(scores,np.asarray(OBJ_MATRIX[:,ci],dtype=np.float32),out=scores)
    positive=np.flatnonzero(scores>0)
    if not len(positive):
        return pd.DataFrame(),{
            "search_ms":(time.perf_counter()-t0)*1000.0,
            "matched_aliases":aliases,
            "matched_classes":classes,
            "status":"ALIASES_MATCHED_NO_OBJECT_HIT",
        }
    kk=min(int(k),len(positive))
    vals=scores[positive]
    if kk<len(positive):
        part=np.argpartition(-vals,kk-1)[:kk]
        rows=positive[part]
    else:
        rows=positive
    order=np.lexsort((rows,-scores[rows]))
    rows=rows[order][:kk]
    out=pd.DataFrame({
        "index_row":rows.astype(np.int64),
        "object_score_raw":scores[rows].astype(float),
        "object_rank":np.arange(1,len(rows)+1,dtype=np.int64),
    })
    out=out.merge(
        VISUAL_META[["index_row","frame_ref_id","video_id","pts_time"]],
        on="index_row",how="left",validate="one_to_one",
    )
    return out,{
        "search_ms":(time.perf_counter()-t0)*1000.0,
        "matched_aliases":aliases,
        "matched_classes":classes,
        "status":"READY",
    }

def object_interval_support(query):
    hits,diag=object_search_frames(query,O2_OBJECT_K)
    if hits.empty:
        return pd.DataFrame(),diag
    rel=hits.merge(
        FRAME_INTERVAL_INDEX.reset_index(drop=True)[
            ["frame_ref_id","variant","shadow_shot_interval_id","video_id"]
        ],
        on=["frame_ref_id","video_id"],how="inner",validate="one_to_many",
    ).rename(columns={"shadow_shot_interval_id":"interval_id"})
    if rel.empty:
        return pd.DataFrame(),diag
    rel=rel.sort_values(
        ["object_score_raw","object_rank","frame_ref_id"],
        ascending=[False,True,True],kind="mergesort",
    )
    d=rel.groupby(["interval_id","variant","video_id"],as_index=False).agg(
        object_score_raw=("object_score_raw","max"),
        object_rank=("object_rank","min"),
        best_object_frame_ref_id=("frame_ref_id","first"),
    )
    d["object_query_pct"]=rank_pct_nonzero(d.object_score_raw)
    d["matched_object_aliases_json"]=json.dumps(
        diag["matched_aliases"],ensure_ascii=False,separators=(",",":")
    )
    return d,diag

# ---- Stage16 typed motion intent + interval overlap -----------------------------
MOTION_INTENT_FEATURE={
    "ABRUPT_APPEARANCE_CHANGE":
        "gray_mae_pair_max__localized_pct"
        if "gray_mae_pair_max__localized_pct" in MOTION_TYPED.columns
        else "gray_mae_pair_mean__localized_pct",
    "SUSTAINED_MOTION":"flow_mag_pair_mean_per_sec__localized_pct",
    "STRONG_LOCAL_MOTION_PEAK":"flow_mag_pair_p95_max_per_sec__localized_pct",
}
for _intent,_feat in MOTION_INTENT_FEATURE.items():
    if _feat not in MOTION_TYPED.columns:
        raise RuntimeError(("O2_MOTION_FEATURE_MISSING",_intent,_feat))

MOTION_INTENT_PHRASES={
    "ABRUPT_APPEARANCE_CHANGE":(
        "sudden","suddenly","abrupt","appears","disappears","scene change","cut to",
        "đột ngột","bất ngờ","xuất hiện","biến mất","chuyển cảnh",
    ),
    "SUSTAINED_MOTION":(
        "moving","moves","walking","running","driving","riding","motion",
        "di chuyển","đang đi","đang chạy","lái xe","đạp xe",
    ),
    "STRONG_LOCAL_MOTION_PEAK":(
        "rapid","quickly","fast motion","jump","throw","wave","jerk",
        "nhanh","nhảy","ném","vẫy",
    ),
}

def detect_motion_intents(query):
    q=norm_text(query)
    hits=[]
    for intent,phrases in MOTION_INTENT_PHRASES.items():
        if any(norm_text(p) in q for p in phrases):
            hits.append(intent)
    return hits

def motion_interval_support(query):
    t0=time.perf_counter()
    intents=detect_motion_intents(query)
    if not intents:
        return pd.DataFrame(),{
            "search_ms":(time.perf_counter()-t0)*1000.0,
            "intents":[],
            "status":"NO_EXPLICIT_MOTION_INTENT",
        }
    rows=[]
    for intent in intents:
        feat=MOTION_INTENT_FEATURE[intent]
        g=MOTION_TYPED[MOTION_TYPED.motion_available & MOTION_TYPED[feat].notna()].copy()
        g=g.sort_values(
            [feat,"video_id","edge_index_in_video"],
            ascending=[False,True,True],kind="mergesort",
        ).head(O2_MOTION_K)
        g["motion_rank"]=np.arange(1,len(g)+1)
        for r in g.itertuples(index=False):
            vid=str(r.video_id)
            packs=PACK_BY_VIDEO.get(vid)
            if packs is None or packs.empty:
                continue
            a=float(r.left_pts_time)
            b=float(r.right_pts_time)
            m=(
                pd.to_numeric(packs.start_pts_time_technical,errors="coerce").le(b+SEEK_TOLERANCE_S)
                &pd.to_numeric(packs.end_pts_time_technical,errors="coerce").ge(a-SEEK_TOLERANCE_S)
            )
            score=float(getattr(r,feat))
            for z in packs.loc[m].itertuples(index=False):
                rows.append({
                    "interval_id":str(z.interval_id),
                    "variant":str(z.variant),
                    "video_id":vid,
                    "motion_intent":intent,
                    "motion_score_normalized_only":score,
                    "motion_rank":int(r.motion_rank),
                    "motion_edge_index_in_video":int(r.edge_index_in_video),
                    "motion_edge_left_pts":a,
                    "motion_edge_right_pts":b,
                })
    if not rows:
        return pd.DataFrame(),{
            "search_ms":(time.perf_counter()-t0)*1000.0,
            "intents":intents,
            "status":"INTENT_MATCHED_NO_INTERVAL_OVERLAP",
        }
    d=pd.DataFrame(rows).sort_values(
        ["motion_score_normalized_only","motion_rank","motion_intent"],
        ascending=[False,True,True],kind="mergesort",
    )
    d=d.groupby(["interval_id","variant","video_id"],as_index=False).agg(
        motion_score_normalized_only=("motion_score_normalized_only","max"),
        motion_rank=("motion_rank","min"),
        motion_intent=("motion_intent","first"),
        motion_edge_index_in_video=("motion_edge_index_in_video","first"),
        motion_edge_left_pts=("motion_edge_left_pts","first"),
        motion_edge_right_pts=("motion_edge_right_pts","first"),
    )
    d["motion_query_pct"]=d.motion_score_normalized_only
    return d,{
        "search_ms":(time.perf_counter()-t0)*1000.0,
        "intents":intents,
        "status":"READY",
    }

def _timed_call(fn,*args):
    t0=time.perf_counter()
    out=fn(*args)
    return out,(time.perf_counter()-t0)*1000.0

def _unpack_timed_result(result, *, pair_payload, label):
    """
    Contract:
      _timed_call(single_result_fn) -> (payload, elapsed_ms)
      _timed_call(pair_result_fn)   -> ((payload, diagnostics), elapsed_ms)
    """
    if not isinstance(result,(tuple,list)) or len(result)!=2:
        raise RuntimeError(("O2_TIMED_RESULT_SHAPE",label,type(result).__name__))
    payload,elapsed_ms=result
    if not isinstance(elapsed_ms,(int,float,np.integer,np.floating)):
        raise RuntimeError(("O2_TIMED_RESULT_ELAPSED_TYPE",label,type(elapsed_ms).__name__))
    elapsed_ms=float(elapsed_ms)
    if not math.isfinite(elapsed_ms) or elapsed_ms<0:
        raise RuntimeError(("O2_TIMED_RESULT_ELAPSED_VALUE",label,elapsed_ms))

    if not pair_payload:
        return payload,elapsed_ms

    if not isinstance(payload,(tuple,list)) or len(payload)!=2:
        raise RuntimeError(("O2_TIMED_PAIR_PAYLOAD_SHAPE",label,type(payload).__name__))
    data,diag=payload
    if not isinstance(diag,dict):
        raise RuntimeError(("O2_TIMED_DIAGNOSTICS_TYPE",label,type(diag).__name__))
    return data,diag,elapsed_ms

def build_federated_interval_pool(query,visual_pool,encoded):
    """
    O2 interval federation.

    All source scores are query/global percentiles or source-normalized signals.
    The equal-available-source mean is an auditable lightweight rerank only.
    It is NOT a correctness probability.
    """
    t0=time.perf_counter()
    with ThreadPoolExecutor(max_workers=4) as ex:
        f_base=ex.submit(_timed_call,build_multimodal_interval_pool,query,visual_pool)
        f_ad=ex.submit(_timed_call,adaptive_interval_support,encoded)
        f_obj=ex.submit(_timed_call,object_interval_support,query)
        f_motion=ex.submit(_timed_call,motion_interval_support,query)

        base,base_ms=_unpack_timed_result(
            f_base.result(),pair_payload=False,label="base_multimodal"
        )
        ad,ad_diag,ad_ms=_unpack_timed_result(
            f_ad.result(),pair_payload=True,label="adaptive"
        )
        obj,obj_diag,obj_ms=_unpack_timed_result(
            f_obj.result(),pair_payload=True,label="object"
        )
        mot,mot_diag,mot_ms=_unpack_timed_result(
            f_motion.result(),pair_payload=True,label="motion"
        )

    base=base.copy()
    if not base.empty:
        base=base.rename(columns={
            "semantic_score_normalized_only":"base_semantic_score_normalized_only"
        })

    ids=set(base.interval_id.astype(str)) if not base.empty else set()
    for d in (ad,obj,mot):
        if isinstance(d,pd.DataFrame) and not d.empty:
            ids |= set(d.interval_id.astype(str))
    if not ids:
        raise RuntimeError("O2_FEDERATED_POOL_EMPTY")

    pool=PACK_INTERVAL_BASE[PACK_INTERVAL_BASE.interval_id.isin(ids)].copy()

    base_cols=[
        "interval_id","variant","video_id",
        "visual_raw_score","visual_prompt_score","raw_frame_rank","best_frame_ref_id",
        "ocr_score_raw","asr_score_raw","ocr_query_pct","asr_query_pct",
        "text_semantic_normalized_only","visual_query_pct",
        "base_semantic_score_normalized_only",
    ]
    if not base.empty:
        pool=pool.merge(
            base[[c for c in base_cols if c in base.columns]],
            on=["interval_id","variant","video_id"],how="left",validate="one_to_one",
        )

    for d in (ad,obj,mot):
        if isinstance(d,pd.DataFrame) and not d.empty:
            pool=pool.merge(
                d,on=["interval_id","variant","video_id"],how="left",validate="one_to_one",
            )

    # Deterministic schema even when a supporting source is absent for this query.
    for _col in (
        "base_semantic_score_normalized_only","visual_raw_score","visual_prompt_score",
        "raw_frame_rank","best_frame_ref_id","ocr_score_raw","asr_score_raw",
        "text_semantic_normalized_only",
    ):
        if _col not in pool.columns:
            pool[_col]=np.nan

    pct_cols=[
        "visual_query_pct","ocr_query_pct","asr_query_pct",
        "adaptive_query_pct","object_query_pct","motion_query_pct",
    ]
    for c in pct_cols:
        if c not in pool.columns:
            pool[c]=np.nan

    pool["o2_source_available_count"]=pool[pct_cols].notna().sum(axis=1)
    pool["federated_score_normalized_only"]=pool[pct_cols].mean(axis=1,skipna=True)
    pool.loc[pool.o2_source_available_count.eq(0),"federated_score_normalized_only"]=np.nan

    labels=[
        ("visual_query_pct","VISUAL_OFFICIAL"),
        ("ocr_query_pct","OCR"),
        ("asr_query_pct","ASR"),
        ("adaptive_query_pct","VISUAL_ADAPTIVE_SHADOW"),
        ("object_query_pct","OBJECT"),
        ("motion_query_pct","MOTION"),
    ]
    pool["o2_sources"]=[
        "|".join(label for col,label in labels if pd.notna(row[col]))
        for _,row in pool.iterrows()
    ]
    # Downstream TRAKE/QA expects this compatibility field.
    pool["semantic_score_normalized_only"]=pool.federated_score_normalized_only
    pool["federation_policy"]=O2_FEDERATION_POLICY
    pool["adaptive_promotion_status"]=ADAPTIVE_PROMOTION_STATUS
    pool["probability_semantics"]="NONE"

    pool=pool.sort_values(
        [
            "federated_score_normalized_only",
            "o2_source_available_count",
            "base_semantic_score_normalized_only",
            "visual_raw_score",
            "interval_id",
        ],
        ascending=[False,False,False,False,True],
        na_position="last",
        kind="mergesort",
    ).head(O2_FEDERATED_K).reset_index(drop=True)
    pool["federated_rank"]=np.arange(1,len(pool)+1)

    diag={
        "policy":O2_FEDERATION_POLICY,
        "total_ms":(time.perf_counter()-t0)*1000.0,
        "base_ms":float(base_ms),
        "adaptive_ms":float(ad_ms),
        "object_ms":float(obj_ms),
        "motion_ms":float(mot_ms),
        "adaptive":ad_diag,
        "object":obj_diag,
        "motion":mot_diag,
        "pool_rows":len(pool),
        "probability_semantics":"NONE",
    }
    return pool,diag

print("[PASS] O2 federated retrieval interface ready",{
    "adaptive_rows":len(ADAPTIVE_META),
    "object_shape":list(OBJ_MATRIX.shape),
    "motion_rows":len(MOTION_TYPED),
    "policy":O2_FEDERATION_POLICY,
})
````

````python
# ── F5 / CELL 12: KIS POST-VALIDATION + DIAGNOSTIC RISK ───────────────────────
# Mục tiêu: preserve strong KIS fast baseline; risk remains diagnostic.
# INPUT: raw exact visual candidates.
# OUTPUT: hard validation + diagnostic risk features.
# Pre-check: exact FrameRef/video/PTS provenance required.
# Post-check: no heuristic threshold may auto-block or auto-replace baseline.
# History v0.2.0: KIS remains stable; refinement is selective and evidence-driven later.
# ──────────────────────────────────────────────────────────────────────────────
def validate_kis_candidates(candidates):
    req={"index_row","frame_ref_id","video_id","pts_time","raw_cosine","image_artifact_ref","frame_idx_candidate"}
    miss=sorted(req-set(candidates.columns))
    if miss: raise RuntimeError(("KIS schema missing",miss))
    if candidates.empty: raise RuntimeError("EMPTY_KIS_RESULT")
    if candidates.frame_ref_id.astype(str).duplicated().any():
        raise RuntimeError("duplicate frame_ref_id")
    for c in ("raw_cosine","pts_time"):
        if not np.isfinite(pd.to_numeric(candidates[c],errors="coerce").to_numpy(float)).all():
            raise RuntimeError(("nonfinite",c))
    if (pd.to_numeric(candidates.pts_time)<0).any():
        raise RuntimeError("negative PTS")
    return True

def kis_risk_diagnostics(candidates):
    top=candidates.head(max(20,RETURN_TOPK))
    s=top.raw_cosine.to_numpy(float)
    top1=float(s[0])
    m12=float(s[0]-s[1]) if len(s)>=2 else math.inf
    m110=float(s[0]-s[min(9,len(s)-1)]) if len(s)>=2 else math.inf
    flags=[]
    if m12<0.005: flags.append("LOW_TOP1_TOP2_MARGIN_SEED")
    if m110<0.015: flags.append("FLAT_TOP10_SCORE_SEED")
    return {
        "decision_authority":"DIAGNOSTIC_ONLY_UNTIL_CALIBRATED",
        "normalization_status":"NORMALIZED_ONLY",
        "probability_calibration_status":"NOT_CALIBRATED",
        "seed_flags":flags,
        "raw_top1_cosine":top1,
        "raw_margin_top1_top2":m12,
        "raw_margin_top1_top10":m110,
    }

print("[PASS] KIS validation/risk diagnostics")
````

````python
# ── F6 / CELL 13: THREAD-SAFE VERSIONED RESULT STORE ──────────────────────────
# Mục tiêu: append-only active/alternative/replacement provenance.
# INPUT: task result payloads.
# OUTPUT: RESULT_STORE.
# Pre-check: active initial result hard-valid.
# Post-check: uncalibrated alternatives never auto-replace active baseline.
# History v0.2.0: unified serving store for KIS/TRAKE/QA.
# History v0.5.0: RLock + deep-copy reads/snapshots for asynchronous KIS O2 shadow.
# ──────────────────────────────────────────────────────────────────────────────
import copy, threading

class VersionedResultStore:
    def __init__(self):
        self.store={}
        self._lock=threading.RLock()

    def put_initial(self,qid,payload):
        with self._lock:
            if qid in self.store:
                raise KeyError(qid)
            v={
                "version":0,"active":True,"reason":"INITIAL_FAST_RESULT",
                "payload":copy.deepcopy(payload),"created_at_utc":utc_now(),
            }
            self.store[qid]={
                "active_result":v,
                "alternatives":[],
                "history":[v],
                "replacements":[],
            }
            return copy.deepcopy(v)

    def add_alternative(self,qid,payload,reason):
        with self._lock:
            st=self.store[qid]
            v={
                "version":len(st["history"]),"active":False,"reason":str(reason),
                "payload":copy.deepcopy(payload),"created_at_utc":utc_now(),
            }
            st["alternatives"].append(v)
            st["history"].append(v)
            return copy.deepcopy(v)

    def replace_verified(self,qid,payload,reason):
        if payload.get("hard_valid") is not True or payload.get("verifier_pass") is not True:
            return False
        with self._lock:
            st=self.store[qid]
            old=st["active_result"]
            v={
                "version":len(st["history"]),"active":True,"reason":str(reason),
                "payload":copy.deepcopy(payload),"created_at_utc":utc_now(),
            }
            old["active"]=False
            st["active_result"]=v
            st["history"].append(v)
            st["replacements"].append({
                "from_version":old["version"],
                "to_version":v["version"],
                "reason":str(reason),
                "created_at_utc":utc_now(),
            })
            return True

    def get(self,qid):
        with self._lock:
            return copy.deepcopy(self.store[qid])

    def snapshot(self):
        with self._lock:
            return {
                "schema_version":"versioned-result-store-v1",
                "build_version":BUILD_VERSION,
                "queries":copy.deepcopy(self.store),
            }

RESULT_STORE=VersionedResultStore()
print("[PASS] thread-safe versioned result store")
````

````python
# ── F6 / CELL 14: REAL KIS FAST PATH + LAZY MULTIMODAL SUPPORT ────────────────
# Mục tiêu: arbitrary KIS query → exact official FrameRef result with minimum blocking work.
# INPUT: query string.
# OUTPUT: active RAW visual result + dual-view alternative; multimodal support is lazy.
# Pre-check: strict 2-GPU encode + exact 2-GPU Stage03 search.
# Post-check: active top result is RAW exact baseline; OCR/ASR cannot block fast return.
# History v0.2.0: KIS synchronously built multimodal interval support (~seconds/query).
# History v0.2.1: return after visual fast path; attach multimodal evidence only on demand.
# ──────────────────────────────────────────────────────────────────────────────
def _kis_rows(df,rank_col,limit=RETURN_TOPK):
    x=df.sort_values(rank_col).head(int(limit))
    rows=[]
    for r in x.itertuples(index=False):
        rows.append({
            "rank":int(getattr(r,rank_col)),
            "frame_ref_id":str(r.frame_ref_id),
            "video_id":str(r.video_id),
            "pts_time":float(r.pts_time),
            "frame_idx_candidate":int(r.frame_idx_candidate),
            "image_artifact_ref":str(r.image_artifact_ref),
            "raw_video_ref":str(r.raw_video_ref),
            "raw_cosine":float(r.raw_cosine),
            "prompt_cosine":float(r.prompt_cosine),
            "dual_view_fused_cosine":float(r.dual_view_fused_cosine),
            "rrf_pool_score_diagnostic":float(r.rrf_pool_score),
        })
    return rows

KIS_RUNTIME_CONTEXT={}

def run_kis(query,query_id=None,include_multimodal_support=False):
    q=canonical_query_text(query)
    if query_id is None:
        query_id="kis_"+hashlib.sha256(q.encode()).hexdigest()[:16]

    t0=time.perf_counter()
    q,encoded,encode_ms=encode_dualgpu(q)
    candidates,diag=search_kis_pool(encoded,K_SEED)
    validate_kis_candidates(candidates)
    risk=kis_risk_diagnostics(candidates)

    raw=candidates.sort_values(["raw_final_rank","index_row"]).reset_index(drop=True)
    fused=candidates.sort_values(["fused_final_rank","index_row"]).reset_index(drop=True)
    raw_rows=_kis_rows(raw,"raw_final_rank")
    fused_rows=_kis_rows(fused,"fused_final_rank")

    fast_total_ms=(time.perf_counter()-t0)*1000.0
    payload={
        "schema_version":"final-kis-fast-v1",
        "build_version":BUILD_VERSION,
        "query_id":query_id,
        "task_type":"KIS",
        "query_text":q,
        "active_policy":"RAW_QUERY_EXACT_OFFICIAL_VISUAL",
        "hard_valid":True,
        "normalization_status":"NORMALIZED_ONLY",
        "probability_calibration_status":"NOT_CALIBRATED",
        "results":raw_rows,
        "risk_diagnostics":risk,
        "multimodal_support_status":"O2_FEDERATED_SHADOW_NOT_BLOCKING_FAST_RESULT",
        "timing_ms":{
            "dual_gpu_encode":encode_ms,
            **diag,
            "fast_total":fast_total_ms,
            "total":fast_total_ms,
        },
    }
    RESULT_STORE.put_initial(query_id,payload)
    RESULT_STORE.add_alternative(query_id,{
        **payload,
        "active_policy":"DUAL_VIEW_FUSED_ABLATION_NOT_ACTIVE",
        "results":fused_rows,
        "hard_valid":True,
    },"UNCALIBRATED_DUAL_VIEW_ALTERNATIVE")

    # Keep only the candidate pool needed for optional selective evidence support.
    KIS_RUNTIME_CONTEXT[query_id]={"query":q,"visual_pool":candidates,"encoded":encoded}
    if include_multimodal_support:
        attach_kis_multimodal_support(query_id)
    return RESULT_STORE.get(query_id)

def attach_kis_multimodal_support(query_id):
    st=RESULT_STORE.get(query_id)
    for a in st["alternatives"]:
        if a["reason"]=="MULTIMODAL_RECALL_SUPPORT_NOT_ACTIVE_RANKING":
            return a
    ctx=KIS_RUNTIME_CONTEXT.get(query_id)
    if ctx is None:
        raise RuntimeError(("KIS_RUNTIME_CONTEXT_MISSING",query_id))
    t0=time.perf_counter()
    pool,diag=build_federated_interval_pool(
        ctx["query"],ctx["visual_pool"],ctx["encoded"]
    )
    evidence_ms=(time.perf_counter()-t0)*1000.0
    return RESULT_STORE.add_alternative(query_id,{
        "schema_version":"final-o2-federated-evidence-pool-v1",
        "build_version":BUILD_VERSION,
        "query_id":query_id,
        "task_type":"KIS",
        "query_text":ctx["query"],
        "active_policy":"O2_FEDERATED_INTERVAL_EVIDENCE_SHADOW_ONLY",
        "hard_valid":True,
        "interval_evidence_top20":pool.head(20).to_dict("records"),
        "federation_diagnostics":diag,
        "evidence_timing_ms":evidence_ms,
        "normalization_status":"NORMALIZED_ONLY",
        "probability_calibration_status":"NOT_CALIBRATED",
        "adaptive_promotion_status":ADAPTIVE_PROMOTION_STATUS,
    },"MULTIMODAL_RECALL_SUPPORT_NOT_ACTIVE_RANKING")

O2_SHADOW_EXECUTOR=ThreadPoolExecutor(max_workers=1)
O2_SHADOW_FUTURES={}
O2_SHADOW_STATUS={}
O2_SHADOW_LOCK=threading.Lock()

def _o2_shadow_done(query_id,fut):
    try:
        alt=fut.result()
        status={"status":"PASS","version":alt.get("version"),"completed_at_utc":utc_now()}
    except Exception as e:
        status={
            "status":"FAIL",
            "error_type":type(e).__name__,
            "error":str(e),
            "completed_at_utc":utc_now(),
        }
    with O2_SHADOW_LOCK:
        O2_SHADOW_STATUS[query_id]=status

def schedule_kis_o2_shadow(query_id):
    with O2_SHADOW_LOCK:
        old=O2_SHADOW_FUTURES.get(query_id)
        if old is not None and not old.done():
            return old
        fut=O2_SHADOW_EXECUTOR.submit(attach_kis_multimodal_support,query_id)
        O2_SHADOW_FUTURES[query_id]=fut
        O2_SHADOW_STATUS[query_id]={"status":"RUNNING","started_at_utc":utc_now()}
        fut.add_done_callback(lambda f,q=query_id:_o2_shadow_done(q,f))
        return fut

def wait_kis_o2_shadow(query_id,timeout=None):
    with O2_SHADOW_LOCK:
        fut=O2_SHADOW_FUTURES.get(query_id)
    if fut is None:
        return None
    return fut.result(timeout=timeout)

print("[READY] run_kis(query, query_id=None, include_multimodal_support=False)")
print("[READY] attach_kis_multimodal_support(query_id)")
print("[READY] schedule_kis_o2_shadow(query_id)")
````

````python
# ── F7 / CELL 15: DANTE TECHNICAL ENGINE + TRAKE EVIDENCE INTERFACE ───────────
# Mục tiêu: preserve frozen DANTE temporal DP without pretending final PTS.
# INPUT: explicit event list + Stage20 interval candidates.
# OUTPUT: provisional temporal path requiring raw-video verifier.
# Pre-check: events explicit; same-video temporal order.
# Post-check: result cannot pass final submission before actual-PTS refinement.
# History v0.2.0: port of frozen Stage22/23 DANTE engine.
# History v0.4.2: restore frozen Stage23 relation-aware span beam for supplied event specs; strict point-DP remains baseline.
# ──────────────────────────────────────────────────────────────────────────────
EPS=1e-12
DANTE_GAP_LAMBDA=0.05
DANTE_TOP_M=3

def dante_point_dp(event_tables,gap_lambda=DANTE_GAP_LAMBDA):
    if not event_tables: raise ValueError("empty event list")
    tables=[]
    for t in event_tables:
        req={"candidate_id","video_id","pts_time","semantic_score"}
        if not req.issubset(t.columns): raise ValueError(sorted(req-set(t.columns)))
        q=t.copy()
        q["video_id"]=q.video_id.astype(str)
        q["pts_time"]=pd.to_numeric(q.pts_time,errors="raise").astype(float)
        q["semantic_score"]=pd.to_numeric(q.semantic_score,errors="raise").astype(float)
        tables.append(q)
    common=sorted(set.intersection(*[set(t.video_id) for t in tables]))
    best=None
    for vid in common:
        tv=[t[t.video_id.eq(vid)].sort_values(["pts_time","candidate_id"]).reset_index(drop=True) for t in tables]
        dp=tv[0].semantic_score.to_numpy(float)
        backs=[np.full(len(tv[0]),-1,dtype=np.int64)]
        for i in range(1,len(tv)):
            prev,cur=tv[i-1],tv[i]
            ndp=np.full(len(cur),-np.inf,float)
            nb=np.full(len(cur),-1,dtype=np.int64)
            pt=prev.pts_time.to_numpy(float)
            for j,row in cur.iterrows():
                valid=np.where(pt<float(row.pts_time)-EPS)[0]
                if not len(valid): continue
                vals=dp[valid]-gap_lambda*(float(row.pts_time)-pt[valid])
                ai=int(np.argmax(vals)); k=int(valid[ai])
                ndp[j]=float(row.semantic_score)+float(vals[ai]); nb[j]=k
            dp=ndp; backs.append(nb)
        if not np.isfinite(dp).any(): continue
        j=int(np.argmax(dp)); path=[]
        for i in range(len(tv)-1,-1,-1):
            row=tv[i].iloc[j]
            path.append({
                "event_index":i,"candidate_id":str(row.candidate_id),"video_id":vid,
                "pts_time_technical":float(row.pts_time),"semantic_score":float(row.semantic_score),
            })
            if i>0:
                j=int(backs[i][j])
                if j<0: raise AssertionError("broken DANTE backpointer")
        item={"video_id":vid,"path_score_normalized_only":float(np.max(dp)),"path":list(reversed(path))}
        if best is None or item["path_score_normalized_only"]>best["path_score_normalized_only"]:
            best=item
    return best


def transition_ok(last,current,relation):
    """Frozen Stage23 relation transition semantics for technical intervals."""
    if last is None:
        return True,0.0
    if str(last["video_id"])!=str(current["video_id"]):
        return False,np.inf
    rel=str(relation).upper()
    if rel not in SUPPORTED_TRAKE_RELATIONS:
        raise ValueError(("unknown relation",relation))
    gap=float(current["start_pts"])-float(last["end_pts"])
    if rel=="STRICT_BEFORE":
        return bool(gap>=-EPS),max(0.0,gap)
    # BEFORE_OR_OVERLAP and REPEAT_ALLOWED preserve nondecreasing event starts.
    ok=float(current["start_pts"])>=float(last["start_pts"])-EPS
    return bool(ok),max(0.0,gap)

def dante_span_beam(event_specs,candidates_by_event,gap_lambda=DANTE_GAP_LAMBDA,top_m=DANTE_TOP_M,miss_penalty=0.2):
    """Generalized Stage23 beam: optional events + overlap/repeat relations."""
    states=[{"score":0.0,"path":[],"last":None,"video_id":None}]
    for i,spec in enumerate(event_specs):
        optional=bool(spec.get("optional",False))
        relation=str(spec.get("relation_from_prev","STRICT_BEFORE")).upper()
        nxt=[]
        for st in states:
            if optional:
                nxt.append({
                    "score":float(st["score"])-float(miss_penalty),
                    "path":st["path"]+[{
                        "event_index":i,
                        "candidate_id":None,
                        "status":"MISS_OPTIONAL",
                        "relation_from_prev":relation,
                    }],
                    "last":st["last"],
                    "video_id":st["video_id"],
                })
            for c in candidates_by_event.get(i,[]):
                cc={
                    "candidate_id":str(c["candidate_id"]),
                    "video_id":str(c["video_id"]),
                    "start_pts":float(c["start_pts"]),
                    "end_pts":float(c["end_pts"]),
                    "semantic_score":float(c["semantic_score"]),
                }
                ok,gap=transition_ok(st["last"],cc,relation)
                if not ok:
                    continue
                if st["video_id"] is not None and str(st["video_id"])!=cc["video_id"]:
                    continue
                nxt.append({
                    "score":float(st["score"])+cc["semantic_score"]-float(gap_lambda)*float(gap),
                    "path":st["path"]+[{
                        "event_index":i,
                        **cc,
                        "status":"SELECTED",
                        "relation_from_prev":relation,
                    }],
                    "last":cc,
                    "video_id":cc["video_id"] if st["video_id"] is None else st["video_id"],
                })
        if not nxt:
            return []
        nxt=sorted(
            nxt,
            key=lambda z:(
                -z["score"],
                tuple("" if p.get("candidate_id") is None else p["candidate_id"] for p in z["path"]),
            )
        )
        seen=set()
        kept=[]
        for z in nxt:
            sig=tuple(p.get("candidate_id") for p in z["path"])
            if sig in seen:
                continue
            seen.add(sig)
            kept.append(z)
            if len(kept)>=int(top_m):
                break
        states=kept
    return states

def run_trake_events(events,query_id=None):
    if not isinstance(events,(list,tuple)) or len(events)<2:
        raise ValueError("TRAKE v0.2.0 requires explicit >=2 event texts; natural-language event parser is next revision")
    ev=[canonical_query_text(x["text"] if isinstance(x,dict) else x) for x in events]
    if query_id is None:
        query_id="trake_"+hashlib.sha256(json.dumps(ev,ensure_ascii=False).encode()).hexdigest()[:16]
    tables=[]; evidence=[]
    for i,text in enumerate(ev):
        # Both GPUs perform visual event encoding; compact OCR/ASR joins are then added.
        _,encoded,_=encode_dualgpu(text)
        vp,_=search_kis_pool(encoded,VISUAL_SEARCH_K)
        pool=build_multimodal_interval_pool(text,vp).head(100).copy()
        evidence.append(pool.head(20).to_dict("records"))
        mid=(pd.to_numeric(pool.start_pts_time_technical)+pd.to_numeric(pool.end_pts_time_technical))/2.0
        tables.append(pd.DataFrame({
            "candidate_id":pool.interval_id.astype(str),
            "video_id":pool.video_id.astype(str),
            "pts_time":mid.astype(float),
            "semantic_score":pd.to_numeric(pool.semantic_score_normalized_only,errors="coerce").fillna(0).astype(float),
        }))
    path=dante_point_dp(tables)
    payload={
        "schema_version":"final-trake-fast-v1",
        "build_version":BUILD_VERSION,
        "query_id":query_id,
        "task_type":"TRAKE",
        "events":ev,
        "active_policy":"DANTE_TECHNICAL_INTERVAL_PATH",
        "path":path,
        "event_evidence_top20":evidence,
        "hard_valid":bool(path),
        "verifier_pass":False,
        "final_temporal_status":"PENDING_RAW_VIDEO_ACTUAL_PTS_VERIFIER",
        "normalization_status":"NORMALIZED_ONLY",
        "probability_calibration_status":"NOT_CALIBRATED",
    }
    RESULT_STORE.put_initial(query_id,payload)
    return RESULT_STORE.get(query_id)

print("[READY] run_trake_events([{text:...},{text:...}], query_id=None)")
````

````python
# ── F7 / CELL 16: QA O2 FEDERATED GROUNDED EVIDENCE INTERFACE ────────────────
# Mục tiêu: retrieve O2 multimodal evidence but never invent a final QA answer.
# INPUT: QA query.
# OUTPUT: federated top interval evidence, status PENDING_FINAL_API.
# Pre-check: real dual-GPU official + adaptive visual; OCR/ASR/Object/Motion support.
# Post-check: answer remains None; grounded final VLM/verifier mandatory.
# History v0.2.0: Stage20 visual/OCR/ASR evidence path.
# History v0.5.0: O2 Stage03/06/11/16/20 federation.
# ──────────────────────────────────────────────────────────────────────────────
def run_qa_evidence(query,query_id=None):
    q=canonical_query_text(query)
    if query_id is None:
        query_id="qa_"+hashlib.sha256(q.encode()).hexdigest()[:16]
    _,encoded,_=encode_dualgpu(q)
    vp,_=search_kis_pool(encoded,VISUAL_SEARCH_K)
    pool,fdiag=build_federated_interval_pool(q,vp,encoded)
    payload={
        "schema_version":"final-qa-o2-evidence-v1",
        "build_version":BUILD_VERSION,
        "query_id":query_id,
        "task_type":"QA",
        "query_text":q,
        "qa_status":"PENDING_FINAL_API",
        "answer":None,
        "hard_valid":True,
        "verifier_pass":False,
        "retrieval_policy":"O2_FEDERATED_STAGE03_06_11_16_20_NORMALIZED_ONLY",
        "evidence_top20":pool.head(20).to_dict("records"),
        "federation_diagnostics":fdiag,
        "grounded_vlm_required":True,
        "api_key_persisted":False,
        "normalization_status":"NORMALIZED_ONLY",
        "probability_calibration_status":"NOT_CALIBRATED",
        "adaptive_promotion_status":ADAPTIVE_PROMOTION_STATUS,
    }
    RESULT_STORE.put_initial(query_id,payload)
    return RESULT_STORE.get(query_id)

print("[READY] run_qa_evidence(query, query_id=None)")
````

````python
# ── F7 / CELL 17: DETERMINISTIC TRAKE EVENT PLANNER + QUEST CONTROL ───────────
# Mục tiêu: free-form TRAKE query → ordered event plan without learned task routing.
# INPUT: query text and optional caller/BTC supplied events.
# OUTPUT: explicit event list + planner provenance + QUEST control decision.
# Pre-check: original query is always preserved.
# Post-check: ambiguous query fails closed; QUEST performs no external lookup itself.
# History v0.3.0: first branch-specific natural-language TRAKE planner.
# History v0.4.2: preserve supplied optional/relation semantics instead of silently collapsing to text-only events.
# ──────────────────────────────────────────────────────────────────────────────

_FORWARD_SPLIT_RE=re.compile(
    r"\b(?:and\s+then|then|after\s+that|afterwards|subsequently|next|"
    r"sau\s+đó|sau\s+do|rồi|roi|tiếp\s+theo|tiep\s+theo)\b",
    flags=re.IGNORECASE,
)
_BEFORE_PATTERNS=[
    re.compile(r"^\s*(.+?)\s+\bbefore\b\s+(.+?)\s*$",re.I),
    re.compile(r"^\s*(.+?)\s+\btrước\s+khi\b\s+(.+?)\s*$",re.I),
    re.compile(r"^\s*(.+?)\s+\btruoc\s+khi\b\s+(.+?)\s*$",re.I),
]
_AFTER_PATTERNS=[
    re.compile(r"^\s*(.+?)\s+\bafter\b\s+(.+?)\s*$",re.I),
    re.compile(r"^\s*(.+?)\s+\bsau\s+khi\b\s+(.+?)\s*$",re.I),
]

def _clean_event_text(x):
    return " ".join(str(x).strip(" ,;:.").split())

def _strict_bool_value(v,field_name):
    if isinstance(v,(bool,np.bool_)):
        return bool(v)
    if v is None:
        return False
    if isinstance(v,(int,np.integer)) and int(v) in (0,1):
        return bool(int(v))
    if isinstance(v,str):
        z=v.strip().lower()
        if z in {"1","true","yes","y"}: return True
        if z in {"0","false","no","n",""}: return False
    raise ValueError((field_name,"must be strict boolean-like",v))

def _event_specs_from_texts(events):
    return [
        {
            "event_index":i,
            "event_id":f"e{i}",
            "text":_clean_event_text(t),
            "optional":False,
            "relation_from_prev":"STRICT_BEFORE",
        }
        for i,t in enumerate(events)
    ]

def _normalize_supplied_event_specs(supplied_events):
    if not isinstance(supplied_events,(list,tuple)):
        raise RuntimeError(("SUPPLIED_TRAKE_EVENTS_NOT_LIST",type(supplied_events).__name__))
    specs=[]
    for i,item in enumerate(supplied_events):
        if isinstance(item,dict):
            text=_clean_event_text(item.get("text",""))
            optional=_strict_bool_value(item.get("optional",False),f"events[{i}].optional")
            rel=str(
                item.get("relation_from_prev",item.get("relation","STRICT_BEFORE"))
            ).strip().upper()
            event_id=str(item.get("event_id",f"e{i}")).strip() or f"e{i}"
        else:
            text=_clean_event_text(item)
            optional=False
            rel="STRICT_BEFORE"
            event_id=f"e{i}"
        if not text:
            raise RuntimeError(("SUPPLIED_TRAKE_EVENT_BLANK",i))
        if rel not in SUPPORTED_TRAKE_RELATIONS:
            raise RuntimeError(("SUPPLIED_TRAKE_RELATION_UNSUPPORTED",i,rel))
        # First relation has no predecessor; normalize it to strict for deterministic provenance.
        if i==0:
            rel="STRICT_BEFORE"
        specs.append({
            "event_index":i,
            "event_id":event_id,
            "text":text,
            "optional":optional,
            "relation_from_prev":rel,
        })
    if not (2<=len(specs)<=TRAKE_MAX_EVENTS):
        raise RuntimeError(("SUPPLIED_TRAKE_EVENTS_INVALID",len(specs)))
    if sum(not z["optional"] for z in specs)<1:
        raise RuntimeError("TRAKE_REQUIRES_AT_LEAST_ONE_NONOPTIONAL_EVENT")
    return specs

def plan_trake_events(query, supplied_events=None):
    q=canonical_query_text(query)

    if supplied_events is not None:
        specs=_normalize_supplied_event_specs(supplied_events)
        return {
            "original_query":q,
            "events":[z["text"] for z in specs],
            "event_specs":specs,
            "planner_status":"RESOLVED_SUPPLIED_EVENTS",
            "planner_authority":"CALLER_OR_BTC_SUPPLIED",
            "original_query_preserved":True,
            "relation_semantics_preserved":True,
        }

    # A before B -> [A,B].
    for pat in _BEFORE_PATTERNS:
        m=pat.match(q)
        if m:
            ev=[_clean_event_text(m.group(1)),_clean_event_text(m.group(2))]
            if all(ev):
                specs=_event_specs_from_texts(ev)
                return {
                    "original_query":q,"events":ev,"event_specs":specs,
                    "planner_status":"RESOLVED_DETERMINISTIC_BINARY_BEFORE",
                    "planner_authority":"DETERMINISTIC_TEMPORAL_SYNTAX",
                    "original_query_preserved":True,
                    "relation_semantics_preserved":True,
                }

    # A after B -> [B,A].
    for pat in _AFTER_PATTERNS:
        m=pat.match(q)
        if m:
            ev=[_clean_event_text(m.group(2)),_clean_event_text(m.group(1))]
            if all(ev):
                specs=_event_specs_from_texts(ev)
                return {
                    "original_query":q,"events":ev,"event_specs":specs,
                    "planner_status":"RESOLVED_DETERMINISTIC_BINARY_AFTER",
                    "planner_authority":"DETERMINISTIC_TEMPORAL_SYNTAX",
                    "original_query_preserved":True,
                    "relation_semantics_preserved":True,
                }

    parts=[_clean_event_text(z) for z in _FORWARD_SPLIT_RE.split(q)]
    parts=[z for z in parts if z]
    if 2<=len(parts)<=TRAKE_MAX_EVENTS:
        specs=_event_specs_from_texts(parts)
        return {
            "original_query":q,
            "events":parts,
            "event_specs":specs,
            "planner_status":"RESOLVED_DETERMINISTIC_FORWARD_SEQUENCE",
            "planner_authority":"DETERMINISTIC_TEMPORAL_SYNTAX",
            "original_query_preserved":True,
            "relation_semantics_preserved":True,
        }

    return {
        "original_query":q,
        "events":[],
        "event_specs":[],
        "planner_status":"UNRESOLVED_REQUIRES_SUPPLIED_EVENTS_OR_EXTERNAL_PLANNER",
        "planner_authority":"FAIL_CLOSED",
        "original_query_preserved":True,
        "relation_semantics_preserved":True,
    }

def quest_control_plan(
    planner_status,
    flat_score_signal=False,
    out_of_knowledge_signal=False,
    external_allowed=False,
    exemplar_ref=None,
    time_budget_ms=QUEST_TIME_BUDGET_MS,
):
    trigger=bool(
        str(planner_status).startswith("UNRESOLVED")
        or flat_score_signal
        or out_of_knowledge_signal
    )
    exemplar_ok=bool(external_allowed and exemplar_ref)
    return {
        "triggered":trigger,
        "preserve_original_query":True,
        "shadow_parallel_branch":trigger,
        "exemplar_branch":bool(trigger and exemplar_ok),
        "external_lookup_performed":False,
        "external_allowed":bool(external_allowed),
        "exemplar_ref":str(exemplar_ref) if exemplar_ok else None,
        "time_budget_ms":int(time_budget_ms),
        "semantics":"QUEST_CONTROL_POLICY_NOT_CORRECTNESS_PROBABILITY",
    }

# Regressions: preserve order and original query.
_p=plan_trake_events("a person enters the room then picks up a bag then leaves")
assert _p["events"]==["a person enters the room","picks up a bag","leaves"]
assert all(z["relation_from_prev"]=="STRICT_BEFORE" for z in _p["event_specs"])
_p2=plan_trake_events("the person leaves after picking up the bag")
assert _p2["events"]==["picking up the bag","the person leaves"]
_p3=plan_trake_events("supplied relation contract",supplied_events=[
    {"text":"first","optional":False,"relation_from_prev":"STRICT_BEFORE"},
    {"text":"second","optional":True,"relation_from_prev":"BEFORE_OR_OVERLAP"},
])
assert _p3["event_specs"][1]["optional"] is True
assert _p3["event_specs"][1]["relation_from_prev"]=="BEFORE_OR_OVERLAP"
_qp=quest_control_plan(
    "UNRESOLVED_REQUIRES_SUPPLIED_EVENTS_OR_EXTERNAL_PLANNER",
    external_allowed=True,exemplar_ref="cached://example"
)
assert _qp["preserve_original_query"] and _qp["exemplar_branch"] and not _qp["external_lookup_performed"]

print("[PASS] deterministic TRAKE planner + QUEST control regressions")
````

````python
# ── F8 / CELL 18: RAW VIDEO INDEX + PTS-AWARE SEEK DECODER ────────────────────
# Mục tiêu: query-time local decode using frozen Stage05 temporal semantics.
# INPUT: video_id + target PTS.
# OUTPUT: first raw frame at/after target + actual PTS + provenance.
# Pre-check: raw path must remain under RAW_ROOT.
# Post-check: actual PTS finite/nonnegative; time_error >= -1e-6; explicit failure.
# History v0.3.0: port KEYFRAME_SEEK_FIRST_FRAME_AT_OR_AFTER_TARGET into Final runtime.
# ──────────────────────────────────────────────────────────────────────────────
import subprocess, sys, io
from PIL import Image

try:
    import av
except ImportError:
    subprocess.check_call([sys.executable,"-m","pip","install","-q","av"])
    import av

AV_VERSION=av.__version__
DECODE_POLICY="KEYFRAME_SEEK_FIRST_FRAME_AT_OR_AFTER_TARGET"

# Stage03 metadata is already canonical official FrameRef lineage and contains the raw-video ref.
_vmap=VISUAL_META[["video_id","raw_video_ref"]].drop_duplicates().copy()
if _vmap.video_id.astype(str).duplicated().any():
    bad=_vmap[_vmap.video_id.astype(str).duplicated(keep=False)].head(20)
    raise RuntimeError(("MULTIPLE_RAW_VIDEO_REFS_PER_VIDEO_ID",bad.to_dict("records")))
VIDEO_RAW_REF_MAP=dict(zip(_vmap.video_id.astype(str),_vmap.raw_video_ref.astype(str)))

RAW_DECODE_USAGE={
    "decode_calls":0,
    "decoded_frames":0,
    "failures":0,
    "policy":DECODE_POLICY,
    "pyav_version":AV_VERSION,
}

def resolve_raw_video_path(video_id):
    vid=str(video_id)
    raw_rel=VIDEO_RAW_REF_MAP.get(vid)
    if raw_rel is None:
        raise RuntimeError(("VIDEO_ID_NOT_IN_STAGE03_RAW_MAP",vid))
    p=(RAW_ROOT/raw_rel).resolve()
    rr=RAW_ROOT.resolve()
    if rr!=p and rr not in p.parents:
        raise RuntimeError(("RAW_VIDEO_PATH_ESCAPED_ROOT",vid,str(p),str(rr)))
    if not p.is_file():
        raise FileNotFoundError(p)
    return p,raw_rel

def decode_raw_first_at_or_after(video_id,target_pts,request_id=None,persist_jpeg=False):
    target=float(target_pts)
    if not math.isfinite(target) or target<0:
        raise ValueError(("INVALID_TARGET_PTS",target_pts))
    vid=str(video_id)
    p,raw_rel=resolve_raw_video_path(vid)
    if request_id is None:
        request_id="dec_"+hashlib.sha256(f"{vid}|{target:.9f}".encode()).hexdigest()[:20]

    RAW_DECODE_USAGE["decode_calls"]+=1
    try:
        with av.open(str(p)) as container:
            stream=container.streams.video[0]
            tb=float(stream.time_base)
            if not math.isfinite(tb) or tb<=0:
                raise RuntimeError(("INVALID_STREAM_TIME_BASE",vid,tb))
            seek_ts=max(0,int(math.floor(target/tb)))
            container.seek(seek_ts,stream=stream,any_frame=False,backward=True)

            found=None
            last_pts=None
            local_ordinal=-1
            for frame in container.decode(stream):
                if frame.pts is None:
                    continue
                local_ordinal+=1
                actual=float(frame.pts*stream.time_base)
                if not math.isfinite(actual):
                    continue
                if last_pts is not None and actual+SEEK_TOLERANCE_S<last_pts:
                    raise RuntimeError(("NON_MONOTONIC_PTS_AFTER_SEEK",vid,last_pts,actual))
                last_pts=actual
                if actual+SEEK_TOLERANCE_S>=target:
                    found=(frame,actual,local_ordinal)
                    break

            if found is None:
                raise RuntimeError(("NO_FRAME_AT_OR_AFTER_TARGET",vid,target))

            frame,actual,local_ordinal=found
            time_error=float(actual-target)
            if time_error < -SEEK_TOLERANCE_S:
                raise RuntimeError(("NEGATIVE_DECODE_TIME_ERROR",vid,target,actual,time_error))

            image=frame.to_image().convert("RGB")
            buf=io.BytesIO()
            image.save(buf,format="JPEG",quality=95)
            jpeg_bytes=buf.getvalue()
            image_sha=hashlib.sha256(jpeg_bytes).hexdigest()

            artifact_ref=None
            if persist_jpeg:
                out_dir=WORK_ROOT/"cache"/"raw_decode"/vid
                out_dir.mkdir(parents=True,exist_ok=True)
                out_path=out_dir/f"{request_id}.jpg"
                tmp=out_path.with_suffix(".tmp.jpg")
                tmp.write_bytes(jpeg_bytes)
                tmp.replace(out_path)
                artifact_ref=out_path.relative_to(WORK_ROOT).as_posix()

            RAW_DECODE_USAGE["decoded_frames"]+=1
            return {
                "request_id":request_id,
                "video_id":vid,
                "target_pts_time":target,
                "actual_pts_time":actual,
                "time_error_s":time_error,
                "seek_decode_ordinal":int(local_ordinal),
                "raw_video_ref":raw_rel,
                "raw_video_path_runtime":str(p),
                "decode_policy":DECODE_POLICY,
                "timestamp_status":"RAW_PTS_VERIFIED",
                "temporal_authority":"RAW_VIDEO_PTS",
                "raw_decode_verified":True,
                "image_sha256":image_sha,
                "image_size_bytes":len(jpeg_bytes),
                "image_artifact_ref":artifact_ref,
                "_image":image,
            }
    except Exception:
        RAW_DECODE_USAGE["failures"]+=1
        raise

def close_decode_images(rows):
    for r in rows:
        im=r.pop("_image",None)
        if im is not None:
            try: im.close()
            except Exception: pass

print("[PASS] PTS-aware raw decoder ready",{
    "policy":DECODE_POLICY,
    "seek_tolerance_s":SEEK_TOLERANCE_S,
    "video_ids":len(VIDEO_RAW_REF_MAP),
    "pyav":AV_VERSION,
})
````

````python
# ── F8 / CELL 19: LOCAL DENSE TARGETS + STRICT DUAL-GPU IMAGE SCORING ─────────
# Mục tiêu: refine a technical interval against raw frames near the candidate.
# INPUT: video_id, [start,end], event text.
# OUTPUT: actual-PTS raw frames + SigLIP2 cosine scores + best raw anchor.
# Pre-check: at least two real raw frames so both GPUs receive image work.
# Post-check: both GPUs participate; selected PTS comes only from raw decode.
# History v0.3.0: first Final local dense raw-video neural refinement.
# ──────────────────────────────────────────────────────────────────────────────

LOCAL_DENSE_GPU_USAGE={
    d:{"image_contract_batches":0,"image_frames":0}
    for d in MODEL_DEVICES
}

def dense_target_grid(start_pts,end_pts,n=DENSE_TARGETS_PER_EVENT,context_s=DENSE_CONTEXT_S):
    a=float(start_pts); b=float(end_pts)
    if not (math.isfinite(a) and math.isfinite(b) and 0<=a<=b):
        raise ValueError(("INVALID_DENSE_INTERVAL",a,b))
    lo=max(0.0,a-float(context_s))
    hi=max(lo,b+float(context_s))
    n=max(2,int(n))
    vals=np.linspace(lo,hi,n,dtype=np.float64)
    # Strict unique target registry.
    vals=np.unique(np.round(vals,9))
    if len(vals)<2:
        vals=np.asarray([lo,lo+0.04],dtype=np.float64)
    return vals.tolist()

def _batch_to_worker_image(batch,w):
    out={}
    for k,v in batch.items():
        if not isinstance(v,torch.Tensor):
            out[k]=v
        elif torch.is_floating_point(v):
            out[k]=v.to(device=w["device"],dtype=torch.float16,non_blocking=True)
        else:
            out[k]=v.to(device=w["device"],non_blocking=True)
    return out

def _encode_image_subset(args):
    w,rows,event_raw_vec=args
    torch.cuda.set_device(w["gpu"])
    images=[r["_image"] for r in rows]
    if not images:
        return []
    batch=w["processor"](images=images,return_tensors="pt")
    inputs=_batch_to_worker_image(batch,w)
    with torch.inference_mode():
        feat=torch.nn.functional.normalize(
            unwrap(w["model"].get_image_features(**inputs)).float(),dim=-1
        )
    q=torch.from_numpy(np.asarray(event_raw_vec,dtype=np.float32)).to(w["device"])
    scores=torch.mv(feat,q)
    torch.cuda.synchronize(w["gpu"])
    vals=scores.detach().cpu().numpy().astype(float).tolist()
    with QUERY_LOCK:
        LOCAL_DENSE_GPU_USAGE[w["device"]]["image_contract_batches"]+=1
        LOCAL_DENSE_GPU_USAGE[w["device"]]["image_frames"]+=len(rows)
    return [
        {"request_id":r["request_id"],"local_visual_cosine":float(sc),"device":w["device"]}
        for r,sc in zip(rows,vals)
    ]

def local_dense_refine_interval(video_id,start_pts,end_pts,event_text,persist_jpeg=False):
    targets=dense_target_grid(start_pts,end_pts)
    decoded=[]
    for j,t in enumerate(targets):
        decoded.append(decode_raw_first_at_or_after(
            video_id,t,
            request_id=f"dense_{video_id}_{hashlib.sha256((str(event_text)+'|'+str(j)+'|'+str(t)).encode()).hexdigest()[:16]}",
            persist_jpeg=persist_jpeg,
        ))

    # De-duplicate actual PTS while preserving first request.
    unique=[]
    seen=set()
    for r in decoded:
        key=round(float(r["actual_pts_time"]),9)
        if key in seen:
            im=r.get("_image")
            if im is not None: im.close()
            continue
        seen.add(key); unique.append(r)
    decoded=unique
    if len(decoded)<2:
        close_decode_images(decoded)
        raise RuntimeError(("DENSE_REFINEMENT_NEEDS_TWO_DISTINCT_RAW_FRAMES",video_id,start_pts,end_pts))

    # Both text views still execute; raw vector is the semantic selection authority here.
    _,encoded,_=encode_dualgpu(event_text)
    raw_vec=encoded[0]["vector"][0]

    # Deterministically split real image frames across the two GPUs; both nonempty.
    parts=[decoded[0::2],decoded[1::2]]
    if not parts[0] or not parts[1]:
        close_decode_images(decoded)
        raise RuntimeError("BOTH_GPUS_REQUIRE_REAL_IMAGE_FRAMES")

    with ThreadPoolExecutor(max_workers=2) as ex:
        scored=list(ex.map(
            _encode_image_subset,
            [(QUERY_WORKERS[0],parts[0],raw_vec),(QUERY_WORKERS[1],parts[1],raw_vec)]
        ))
    score_map={x["request_id"]:x for group in scored for x in group}
    clean=[]
    for r in decoded:
        sc=score_map.get(r["request_id"])
        if sc is None:
            close_decode_images(decoded)
            raise AssertionError(("MISSING_DENSE_SCORE",r["request_id"]))
        item={k:v for k,v in r.items() if k!="_image" and k!="raw_video_path_runtime"}
        item.update(sc)
        clean.append(item)

    close_decode_images(decoded)
    clean=sorted(clean,key=lambda x:(-x["local_visual_cosine"],x["actual_pts_time"],x["request_id"]))
    best=clean[0]
    return {
        "video_id":str(video_id),
        "event_text":canonical_query_text(event_text),
        "technical_interval":{
            "start_pts":float(start_pts),"end_pts":float(end_pts),
        },
        "decode_policy":DECODE_POLICY,
        "candidates":clean,
        "best_raw_anchor":best,
        "normalization_status":"NORMALIZED_ONLY",
        "probability_semantics":"NONE",
    }

print("[READY] local_dense_refine_interval(video_id,start_pts,end_pts,event_text)")
````

````python
# ── F9 / CELL 20: TRAKE FAST PATH → RAW-PROVENANCE REFINEMENT → DEEP VERIFIER ─
# Mục tiêu: versioned anytime TRAKE execution.
# INPUT: free-form TRAKE query or supplied event list.
# OUTPUT: provisional DANTE v0; verified raw-PTS replacement only after deep gates.
# Pre-check: deterministic planner must resolve >=2 ordered events.
# Post-check: final replacement uses one video, strict actual PTS order and raw decode proof.
# History v0.3.0: first Final actual-PTS TRAKE replacement path.
# History v0.4.2: relation-aware supplied-event path; optional events no longer misalign refinement; verifier checks relation semantics.
# History v0.4.3: deep verifier accepts legacy two-argument callers while preserving explicit relation semantics for Online O1.
# History v0.5.0: TRAKE event retrieval uses O2 federated Stage03/06/11/16/20 interval candidates.
# ──────────────────────────────────────────────────────────────────────────────

TRAKE_RUNTIME_CONTEXT={}

def _event_candidate_table(event_text):
    _,encoded,_=encode_dualgpu(event_text)
    visual,_=search_kis_pool(encoded,VISUAL_SEARCH_K)
    pool,fdiag=build_federated_interval_pool(event_text,visual,encoded)
    pool=pool.head(TRAKE_EVENT_POOL_K).copy()
    if pool.empty:
        raise RuntimeError(("EMPTY_TRAKE_EVENT_POOL",event_text))
    pool["event_candidate_id"]=[
        f"{r.variant}::{r.interval_id}::{i}"
        for i,r in enumerate(pool.itertuples(index=False))
    ]
    mid=(
        pd.to_numeric(pool.start_pts_time_technical,errors="raise")
        +pd.to_numeric(pool.end_pts_time_technical,errors="raise")
    )/2.0
    table=pd.DataFrame({
        "candidate_id":pool.event_candidate_id.astype(str),
        "video_id":pool.video_id.astype(str),
        "pts_time":mid.astype(float),
        "semantic_score":pd.to_numeric(
            pool.federated_score_normalized_only,errors="coerce"
        ).fillna(0.0).astype(float),
    })
    meta={
        str(r.event_candidate_id):{
            "interval_id":str(r.interval_id),
            "variant":str(r.variant),
            "video_id":str(r.video_id),
            "start_pts_time_technical":float(r.start_pts_time_technical),
            "end_pts_time_technical":float(r.end_pts_time_technical),
            "semantic_score_normalized_only":float(r.federated_score_normalized_only)
                if pd.notna(r.federated_score_normalized_only) else 0.0,
            "base_semantic_score_normalized_only":float(r.base_semantic_score_normalized_only)
                if pd.notna(getattr(r,"base_semantic_score_normalized_only",None)) else None,
            "o2_source_available_count":int(r.o2_source_available_count),
            "o2_sources":str(r.o2_sources),
            "best_frame_ref_id":str(r.best_frame_ref_id)
                if pd.notna(getattr(r,"best_frame_ref_id",None)) else None,
            "best_adaptive_frame_ref_id":str(r.best_adaptive_frame_ref_id)
                if pd.notna(getattr(r,"best_adaptive_frame_ref_id",None)) else None,
            "best_object_frame_ref_id":str(r.best_object_frame_ref_id)
                if pd.notna(getattr(r,"best_object_frame_ref_id",None)) else None,
            "ocr_text_compact":str(r.ocr_text_compact or ""),
            "asr_text_compact":str(r.asr_text_compact or ""),
            "federation_policy":str(r.federation_policy),
        }
        for r in pool.itertuples(index=False)
    }
    evidence=pool.head(20).to_dict("records")
    for row in evidence:
        row["federation_diagnostics_summary"]={
            "policy":fdiag["policy"],
            "adaptive_status":fdiag["adaptive"].get("promotion_status"),
            "object_status":fdiag["object"].get("status"),
            "motion_status":fdiag["motion"].get("status"),
        }
    return table,meta,evidence

def run_trake_fast(query,query_id=None,supplied_events=None):
    q=canonical_query_text(query)
    if query_id is None:
        query_id="trake_"+hashlib.sha256(q.encode()).hexdigest()[:16]
    plan=plan_trake_events(q,supplied_events=supplied_events)
    specs=plan.get("event_specs",[])
    if len(specs)<2:
        raise RuntimeError(("TRAKE_EVENT_PLANNER_UNRESOLVED",plan))

    tables=[]; meta={}; evidence=[]; candidates_by_event={}
    for i,spec in enumerate(specs):
        t,m,e=_event_candidate_table(spec["text"])
        tables.append(t)
        meta.update(m)
        evidence.append(e)
        candidates_by_event[i]=[
            {
                "candidate_id":cid,
                "video_id":mm["video_id"],
                "start_pts":mm["start_pts_time_technical"],
                "end_pts":mm["end_pts_time_technical"],
                "semantic_score":mm["semantic_score_normalized_only"],
            }
            for cid,mm in m.items()
        ]

    generalized=any(
        z.get("optional",False)
        or z.get("relation_from_prev","STRICT_BEFORE")!="STRICT_BEFORE"
        for z in specs
    )

    optional_misses=[]
    if generalized:
        states=dante_span_beam(specs,candidates_by_event)
        if not states:
            raise RuntimeError(("DANTE_GENERALIZED_NO_COMMON_PATH",query_id))
        best=states[0]
        selected=[z for z in best["path"] if z.get("status")=="SELECTED"]
        optional_misses=[z["event_index"] for z in best["path"] if z.get("status")=="MISS_OPTIONAL"]
        if len(selected)<2:
            raise RuntimeError(("DANTE_GENERALIZED_PATH_TOO_FEW_SELECTED_EVENTS",query_id,len(selected)))
        path_rows=[]
        for z in selected:
            cid=str(z["candidate_id"])
            mm=meta[cid]
            path_rows.append({
                "event_index":int(z["event_index"]),
                "relation_from_prev":str(z.get("relation_from_prev","STRICT_BEFORE")),
                "candidate_id":cid,
                "video_id":str(mm["video_id"]),
                "pts_time_technical":0.5*(
                    float(mm["start_pts_time_technical"])+float(mm["end_pts_time_technical"])
                ),
                "semantic_score":float(z["semantic_score"]),
                **mm,
            })
        path={
            "video_id":str(best["video_id"]),
            "path_score_normalized_only":float(best["score"]),
            "path":[
                {
                    "event_index":r["event_index"],
                    "candidate_id":r["candidate_id"],
                    "video_id":r["video_id"],
                    "pts_time_technical":r["pts_time_technical"],
                    "semantic_score":r["semantic_score"],
                    "relation_from_prev":r["relation_from_prev"],
                }
                for r in path_rows
            ],
            "engine":"DANTE_SPAN_BEAM_RELATION_AWARE",
        }
    else:
        path=dante_point_dp(tables)
        if path is None:
            raise RuntimeError(("DANTE_NO_COMMON_ORDERED_VIDEO_PATH",query_id))
        path_rows=[]
        for z in path["path"]:
            cid=str(z["candidate_id"])
            mm=meta[cid]
            path_rows.append({
                **z,
                **mm,
                "event_index":int(z["event_index"]),
                "relation_from_prev":"STRICT_BEFORE",
                "pts_time_technical":float(z["pts_time_technical"]),
            })
        path["engine"]="DANTE_POINT_DP_STRICT_BASELINE"

    payload={
        "schema_version":"final-trake-provisional-v2",
        "build_version":BUILD_VERSION,
        "query_id":query_id,
        "task_type":"TRAKE",
        "query_text":q,
        "planner":plan,
        "events":plan["events"],
        "event_specs":specs,
        "active_policy":"DANTE_O2_FEDERATED_TECHNICAL_INTERVAL_PROVISIONAL",
        "retrieval_policy":"O2_FEDERATED_STAGE03_06_11_16_20_NORMALIZED_ONLY",
        "dante_engine":path["engine"],
        "video_id":path["video_id"],
        "path_score_normalized_only":path["path_score_normalized_only"],
        "path":path_rows,
        "optional_missed_event_indices":optional_misses,
        "event_evidence_top20":evidence,
        "hard_valid":True,
        "verifier_pass":False,
        "final_temporal_status":"PENDING_RAW_VIDEO_ACTUAL_PTS_VERIFIER",
        "normalization_status":"NORMALIZED_ONLY",
        "probability_calibration_status":"NOT_CALIBRATED",
    }
    RESULT_STORE.put_initial(query_id,payload)
    TRAKE_RUNTIME_CONTEXT[query_id]={
        "query":q,"plan":plan,"path":path_rows,"meta":meta
    }
    return RESULT_STORE.get(query_id)

def _actual_pts_relation_ok(prev_pts,current_pts,relation):
    rel=str(relation).upper()
    if rel=="STRICT_BEFORE":
        return bool(float(current_pts)>float(prev_pts))
    if rel in {"BEFORE_OR_OVERLAP","REPEAT_ALLOWED"}:
        return bool(float(current_pts)>=float(prev_pts)-SEEK_TOLERANCE_S)
    raise ValueError(("unknown actual-PTS relation",relation))

def deep_verify_trake_refinement(query_id,refined_events,event_specs=None):
    """
    Hard verifier over raw actual PTS.

    v0.4.3 compatibility:
    - Online O1 passes explicit event_specs.
    - Historical v0.3 smoke callers may omit event_specs; those rows are interpreted
      as a strict-before sequence and the fallback authority is reported explicitly.
    """
    if len(refined_events)<2:
        return {"pass":False,"reason":"TOO_FEW_REFINED_EVENTS"}

    spec_by_index={}
    if event_specs is not None:
        if not isinstance(event_specs,(list,tuple)):
            return {"pass":False,"reason":"EVENT_SPECS_NOT_SEQUENCE"}
        for i,spec in enumerate(event_specs):
            if not isinstance(spec,dict):
                return {"pass":False,"reason":"EVENT_SPEC_NOT_DICT","event_spec_index":i}
            idx=int(spec.get("event_index",i))
            rel=str(spec.get("relation_from_prev","STRICT_BEFORE")).upper()
            if idx==0:
                rel="STRICT_BEFORE"
            if rel not in SUPPORTED_TRAKE_RELATIONS:
                return {
                    "pass":False,
                    "reason":"EVENT_SPEC_RELATION_UNSUPPORTED",
                    "event_spec_index":idx,
                    "relation":rel,
                }
            spec_by_index[idx]={**spec,"event_index":idx,"relation_from_prev":rel}
        relation_authority="EXPLICIT_EVENT_SPECS"
    else:
        relation_authority="LEGACY_STRICT_BEFORE_FALLBACK"

    videos={str(z["video_id"]) for z in refined_events}
    if len(videos)!=1:
        return {"pass":False,"reason":"MULTI_VIDEO_REFINED_PATH"}

    pts=[float(z["actual_pts_time"]) for z in refined_events]
    if not all(math.isfinite(v) and v>=0 for v in pts):
        return {"pass":False,"reason":"NONFINITE_ACTUAL_PTS"}
    if not all(z.get("raw_decode_verified") is True for z in refined_events):
        return {"pass":False,"reason":"RAW_DECODE_NOT_VERIFIED"}
    if not all(z.get("decode_policy")==DECODE_POLICY for z in refined_events):
        return {"pass":False,"reason":"DECODE_POLICY_DRIFT"}

    normalized=[]
    for j,z in enumerate(refined_events):
        event_index=int(z.get("event_index",j))
        spec=spec_by_index.get(event_index,{})
        rel=str(
            z.get(
                "relation_from_prev",
                spec.get("relation_from_prev","STRICT_BEFORE"),
            )
        ).upper()
        if j==0:
            rel="STRICT_BEFORE"
        if rel not in SUPPORTED_TRAKE_RELATIONS:
            return {
                "pass":False,
                "reason":"REFINED_EVENT_RELATION_UNSUPPORTED",
                "event_index":event_index,
                "relation":rel,
            }
        normalized.append({
            "event_index":event_index,
            "relation_from_prev":rel,
            "actual_pts_time":pts[j],
        })

    checks=[]
    for j in range(1,len(normalized)):
        prev=normalized[j-1]
        cur=normalized[j]
        ok=_actual_pts_relation_ok(
            prev["actual_pts_time"],
            cur["actual_pts_time"],
            cur["relation_from_prev"],
        )
        checks.append({
            "from_event_index":prev["event_index"],
            "to_event_index":cur["event_index"],
            "relation_from_prev":cur["relation_from_prev"],
            "prev_actual_pts":prev["actual_pts_time"],
            "current_actual_pts":cur["actual_pts_time"],
            "pass":bool(ok),
        })
        if not ok:
            return {
                "pass":False,
                "reason":"ACTUAL_PTS_RELATION_VIOLATION",
                "actual_pts":pts,
                "relation_checks":checks,
                "relation_authority":relation_authority,
            }

    return {
        "pass":True,
        "reason":"TRAKE_RAW_ACTUAL_PTS_RELATION_HARD_GATES_PASS",
        "video_id":next(iter(videos)),
        "actual_pts":pts,
        "relation_checks":checks,
        "relation_authority":relation_authority,
        "temporal_authority":"RAW_VIDEO_PTS",
    }

def refine_trake_result(query_id,persist_jpeg=False):
    st=RESULT_STORE.get(query_id)
    active=st["active_result"]["payload"]
    if active["task_type"]!="TRAKE":
        raise RuntimeError(("NOT_TRAKE_RESULT",query_id))
    if active.get("verifier_pass") is True:
        return st

    ctx=TRAKE_RUNTIME_CONTEXT.get(query_id)
    if ctx is None:
        raise RuntimeError(("TRAKE_RUNTIME_CONTEXT_MISSING",query_id))

    quest=quest_control_plan(
        ctx["plan"]["planner_status"],
        flat_score_signal=False,
        out_of_knowledge_signal=False,
        external_allowed=False,
        exemplar_ref=None,
    )

    specs=ctx["plan"]["event_specs"]
    refined=[]
    for p in ctx["path"]:
        event_index=int(p["event_index"])
        spec=specs[event_index]
        dense=local_dense_refine_interval(
            p["video_id"],
            p["start_pts_time_technical"],
            p["end_pts_time_technical"],
            spec["text"],
            persist_jpeg=persist_jpeg,
        )
        best=dense["best_raw_anchor"]
        refined.append({
            "event_index":event_index,
            "event_id":spec.get("event_id",f"e{event_index}"),
            "event_text":spec["text"],
            "optional":bool(spec.get("optional",False)),
            "relation_from_prev":str(spec.get("relation_from_prev","STRICT_BEFORE")),
            "video_id":str(best["video_id"]),
            "target_pts_time":float(best["target_pts_time"]),
            "actual_pts_time":float(best["actual_pts_time"]),
            "time_error_s":float(best["time_error_s"]),
            "request_id":best["request_id"],
            "raw_video_ref":best["raw_video_ref"],
            "image_sha256":best["image_sha256"],
            "image_artifact_ref":best["image_artifact_ref"],
            "local_visual_cosine":float(best["local_visual_cosine"]),
            "raw_decode_verified":bool(best["raw_decode_verified"]),
            "decode_policy":best["decode_policy"],
            "temporal_authority":best["temporal_authority"],
        })

    verdict=deep_verify_trake_refinement(query_id,refined,specs)
    replacement={
        "schema_version":"final-trake-verified-v2",
        "build_version":BUILD_VERSION,
        "query_id":query_id,
        "task_type":"TRAKE",
        "query_text":ctx["query"],
        "events":ctx["plan"]["events"],
        "event_specs":specs,
        "video_id":verdict.get("video_id",active["video_id"]),
        "actual_pts":verdict.get("actual_pts",[]),
        "refined_events":refined,
        "quest_control":quest,
        "hard_valid":bool(verdict.get("pass")),
        "verifier_pass":bool(verdict.get("pass")),
        "verifier":verdict,
        "final_temporal_status":"RAW_PTS_VERIFIED" if verdict.get("pass") else "VERIFIER_REJECTED",
        "normalization_status":"NORMALIZED_ONLY",
        "probability_calibration_status":"NOT_CALIBRATED",
    }
    if verdict.get("pass"):
        ok=RESULT_STORE.replace_verified(
            query_id,replacement,"TRAKE_RAW_ACTUAL_PTS_DEEP_VERIFIER_PASS"
        )
        if not ok:
            raise AssertionError("verified TRAKE replacement unexpectedly rejected")
    else:
        RESULT_STORE.add_alternative(
            query_id,replacement,"TRAKE_REFINEMENT_VERIFIER_REJECTED"
        )
    return RESULT_STORE.get(query_id)

print("[READY] run_trake_fast(query, supplied_events=None)")
print("[READY] refine_trake_result(query_id)")
````

````python
# ── F9 / CELL 21: GROUNDED VLM ADAPTER + QA DEEP VERIFIER ─────────────────────
# Mục tiêu: provider-agnostic external VLM boundary with evidence-citation hard gates.
# INPUT: QA evidence result + runtime-only API key + registered callable.
# OUTPUT: verified QA replacement or abstain/reject; secret never serialized.
# Pre-check: provider/model registered explicitly; no default provider lock.
# Post-check: every cited evidence ID must come from retrieved EvidencePacks.
# History v0.3.0: first grounded external-answer adapter in Final runtime.
# ──────────────────────────────────────────────────────────────────────────────

GROUNDED_VLM_REGISTRY={}

def register_grounded_vlm(provider_name,model_name,callable_fn):
    provider=str(provider_name).strip()
    model=str(model_name).strip()
    if not provider or not model or not callable(callable_fn):
        raise ValueError("provider/model/callable required")
    GROUNDED_VLM_REGISTRY[provider]={
        "provider":provider,
        "model":model,
        "callable":callable_fn,
        "api_key_persisted":False,
    }

def _qa_evidence_ids(payload):
    ids=[]
    for x in payload.get("evidence_top20",[]):
        eid=str(x.get("evidence_pack_id","")).strip()
        if eid: ids.append(eid)
    return ids

def deep_verify_qa_grounding(base_payload,response):
    answer=str(response.get("answer","")).strip()
    cited=[str(x).strip() for x in response.get("cited_evidence_ids",[]) if str(x).strip()]
    abstain=bool(response.get("abstain",False))
    allowed=set(_qa_evidence_ids(base_payload))

    if abstain:
        return {"pass":False,"abstain":True,"reason":"PROVIDER_ABSTAINED"}
    if not answer:
        return {"pass":False,"abstain":False,"reason":"EMPTY_ANSWER"}
    if not cited:
        return {"pass":False,"abstain":False,"reason":"NO_EVIDENCE_CITATIONS"}
    unknown=sorted(set(cited)-allowed)
    if unknown:
        return {
            "pass":False,"abstain":False,
            "reason":"CITED_EVIDENCE_OUTSIDE_RETRIEVED_PACK",
            "unknown_evidence_ids":unknown,
        }
    return {
        "pass":True,
        "abstain":False,
        "reason":"QA_GROUNDED_EVIDENCE_HARD_GATES_PASS",
        "answer":answer,
        "cited_evidence_ids":cited,
    }

def call_grounded_vlm_for_qa(query_id,provider_name,api_key):
    provider=str(provider_name).strip()
    key=str(api_key or "")
    if not key:
        raise RuntimeError("RUNTIME_API_KEY_REQUIRED")
    spec=GROUNDED_VLM_REGISTRY.get(provider)
    if spec is None:
        raise RuntimeError(("GROUNDED_VLM_PROVIDER_NOT_REGISTERED",provider))

    st=RESULT_STORE.get(query_id)
    base=st["active_result"]["payload"]
    if base.get("task_type")!="QA":
        raise RuntimeError(("NOT_QA_RESULT",query_id))
    if base.get("verifier_pass") is True:
        return st

    # Send only bounded grounded evidence. The API key is passed directly to the
    # callable and is never inserted into request/result/store objects.
    evidence=[]
    for x in base.get("evidence_top20",[])[:20]:
        evidence.append({
            "evidence_pack_id":str(x.get("evidence_pack_id","")),
            "video_id":str(x.get("video_id","")),
            "start_pts_time_technical":x.get("start_pts_time_technical"),
            "end_pts_time_technical":x.get("end_pts_time_technical"),
            "ocr_text_compact":str(x.get("ocr_text_compact","")),
            "asr_text_compact":str(x.get("asr_text_compact","")),
        })
    request={
        "purpose":"GROUNDED_VIDEO_QA",
        "query_id":query_id,
        "query_text":base["query_text"],
        "evidence":evidence,
        "required_response_schema":{
            "answer":"string",
            "cited_evidence_ids":["evidence_pack_id"],
            "abstain":"boolean",
        },
        "instruction":"Answer only from supplied evidence; cite evidence_pack_id values; abstain if insufficient.",
    }

    response=spec["callable"](
        api_key=key,
        model=spec["model"],
        request=request,
    )
    if not isinstance(response,dict):
        raise TypeError(("VLM_RESPONSE_NOT_DICT",type(response)))

    # Never store/echo a secret even if a provider callable accidentally returns it.
    forbidden={"api_key","authorization","token","secret"}
    clean_response={k:v for k,v in response.items() if str(k).lower() not in forbidden}
    verdict=deep_verify_qa_grounding(base,clean_response)

    candidate={
        "schema_version":"final-qa-verified-v1",
        "build_version":BUILD_VERSION,
        "query_id":query_id,
        "task_type":"QA",
        "query_text":base["query_text"],
        "answer":verdict.get("answer"),
        "grounded":bool(verdict.get("pass")),
        "cited_evidence_ids":verdict.get("cited_evidence_ids",[]),
        "provider":spec["provider"],
        "model":spec["model"],
        "api_key_persisted":False,
        "hard_valid":bool(verdict.get("pass")),
        "verifier_pass":bool(verdict.get("pass")),
        "verifier":verdict,
        "normalization_status":"NORMALIZED_ONLY",
        "probability_calibration_status":"NOT_CALIBRATED",
    }
    if verdict.get("pass"):
        ok=RESULT_STORE.replace_verified(
            query_id,candidate,"QA_GROUNDED_VLM_DEEP_VERIFIER_PASS"
        )
        if not ok:
            raise AssertionError("verified QA replacement unexpectedly rejected")
    else:
        RESULT_STORE.add_alternative(
            query_id,candidate,"QA_VLM_RESPONSE_REJECTED_OR_ABSTAINED"
        )
    return RESULT_STORE.get(query_id)

# Interface regression with in-memory fake provider: contract only, no external neural call.
def _contract_provider(*,api_key,model,request):
    assert api_key=="runtime-secret-not-persisted"
    eid=request["evidence"][0]["evidence_pack_id"]
    return {
        "answer":"contract answer",
        "cited_evidence_ids":[eid],
        "abstain":False,
    }

print("[PASS] grounded VLM adapter + QA verifier contracts ready")
````

````python
# ── F10 / CELL 22: ONLINE O2 CONTROL PLANE / ANYTIME ENTRYPOINT ───────────────
# Mục tiêu: deterministic pre-validation → branch plan → dispatcher → post-validation → risk/evidence gate.
# INPUT: BTC task_type + query payload.
# OUTPUT: branch result-store state + runtime-only ONLINE_RUNTIME_TRACE.
# Pre-check: no learned task router; BTC task_type is authoritative.
# Post-check: uncalibrated risk signals cannot auto-replace frozen KIS baseline.
# History v0.3.0: free-form TRAKE planner + optional synchronous refinement.
# History v0.4.2: unified Online O1 control plane and deterministic decision trace.
# History v0.5.0: branch plans declare O2 federation; KIS schedules nonblocking O2 shadow.
# ──────────────────────────────────────────────────────────────────────────────

ONLINE_RUNTIME_TRACE={}

def _optional_bool(v,default=False):
    if v is None:
        return bool(default)
    if isinstance(v,(bool,np.bool_)):
        return bool(v)
    if isinstance(v,(int,np.integer)) and int(v) in (0,1):
        return bool(int(v))
    if isinstance(v,str):
        z=v.strip().lower()
        if z in {"1","true","yes","y"}: return True
        if z in {"0","false","no","n",""}: return False
    raise ValueError(("BOOLEAN_FIELD_INVALID",v))

def validate_online_request(req):
    if not isinstance(req,dict):
        raise TypeError("request must be dict")
    task=str(req.get("task_type","")).strip().upper()
    if task not in {"KIS","TRAKE","QA"}:
        raise ValueError(("INVALID_BTC_TASK_TYPE",task))
    q=canonical_query_text(str(req.get("query_text",req.get("query",""))))
    qid=req.get("query_id")
    qid=None if qid is None else str(qid).strip()
    if qid=="":
        qid=None

    clean={
        "query_id":qid,
        "task_type":task,
        "query_text":q,
    }
    if task=="KIS":
        clean["include_multimodal_support"]=_optional_bool(
            req.get("include_multimodal_support",False)
        )
    elif task=="TRAKE":
        events=req.get("events")
        if events is not None and not isinstance(events,(list,tuple)):
            raise ValueError("TRAKE events must be a list/tuple or None")
        clean["events"]=events
        clean["refine"]=_optional_bool(req.get("refine",False))
        clean["persist_refinement_jpeg"]=_optional_bool(
            req.get("persist_refinement_jpeg",False)
        )
    return clean

def branch_execution_plan(clean_req):
    task=clean_req["task_type"]
    if task=="KIS":
        return {
            "task_type":"KIS",
            "primary":"RAW_QUERY_EXACT_OFFICIAL_VISUAL",
            "secondary":"ASYNC_O2_FEDERATED_STAGE06_11_16_20_SHADOW_SUPPORT",
            "replacement_authority":"NONE_UNCALIBRATED",
        }
    if task=="TRAKE":
        return {
            "task_type":"TRAKE",
            "planner":"DETERMINISTIC_OR_CALLER_SUPPLIED_EVENT_SPECS",
            "global_retrieval":"O2_FEDERATED_STAGE03_06_11_16_20",
            "fast":"DANTE_FEDERATED_TECHNICAL_INTERVAL_PATH",
            "slow":"LOCAL_DENSE_RAW_PTS_DEEP_VERIFIER",
            "refine_requested":bool(clean_req.get("refine",False)),
        }
    return {
        "task_type":"QA",
        "retrieval":"O2_FEDERATED_STAGE03_06_11_16_20_EVIDENCE",
        "slow":"GROUNDED_EXTERNAL_VLM_PLUS_CITATION_VERIFIER",
        "answer_without_grounding":False,
    }

def validate_online_active_result(state,expected_task):
    if not isinstance(state,dict) or "active_result" not in state:
        raise RuntimeError("ONLINE_STATE_CONTRACT_MISSING_ACTIVE_RESULT")
    p=state["active_result"].get("payload")
    if not isinstance(p,dict):
        raise RuntimeError("ONLINE_STATE_PAYLOAD_NOT_DICT")
    if p.get("task_type")!=expected_task:
        raise RuntimeError(("ONLINE_TASK_RESULT_MISMATCH",expected_task,p.get("task_type")))
    if not str(p.get("query_id","")).strip():
        raise RuntimeError("ONLINE_RESULT_QUERY_ID_MISSING")

    if expected_task=="KIS":
        if p.get("hard_valid") is not True or not p.get("results"):
            raise RuntimeError("ONLINE_KIS_HARD_CONTRACT_FAILED")
    elif expected_task=="TRAKE":
        if p.get("verifier_pass") is True:
            if p.get("final_temporal_status")!="RAW_PTS_VERIFIED":
                raise RuntimeError("ONLINE_TRAKE_VERIFIED_STATUS_DRIFT")
            if len(p.get("actual_pts",[]))<2:
                raise RuntimeError("ONLINE_TRAKE_VERIFIED_PTS_MISSING")
        else:
            if p.get("hard_valid") is not True or len(p.get("path",[]))<2:
                raise RuntimeError("ONLINE_TRAKE_PROVISIONAL_CONTRACT_FAILED")
    else:
        if p.get("verifier_pass") is True:
            if p.get("grounded") is not True or not str(p.get("answer","")).strip():
                raise RuntimeError("ONLINE_QA_VERIFIED_CONTRACT_FAILED")
        elif not p.get("evidence_top20"):
            raise RuntimeError("ONLINE_QA_EVIDENCE_EMPTY")
    return True

def online_risk_evidence_gate(payload):
    task=payload["task_type"]
    base={
        "schema_version":ONLINE_CONTROL_SCHEMA,
        "task_type":task,
        "decision_authority":"TECHNICAL_GATE_NOT_CORRECTNESS_PROBABILITY",
        "normalization_status":"NORMALIZED_ONLY",
        "probability_calibration_status":"NOT_CALIBRATED",
    }
    if task=="KIS":
        flags=list(payload.get("risk_diagnostics",{}).get("seed_flags",[]))
        return {
            **base,
            "decision":"FAST_ACCEPT_FROZEN_RAW_BASELINE",
            "risk_flags":flags,
            "lazy_multimodal_support_recommended":bool(flags),
            "active_replacement_allowed":False,
        }
    if task=="TRAKE":
        if payload.get("verifier_pass") is True:
            return {
                **base,
                "decision":"VERIFIED_FINAL_ACCEPT",
                "refinement_required_for_technical_submission":False,
            }
        return {
            **base,
            "decision":"PROVISIONAL_FAST_RESULT_REFINEMENT_REQUIRED_FOR_TECHNICAL_FINAL",
            "refinement_required_for_technical_submission":True,
            "dante_engine":payload.get("dante_engine"),
        }

    if payload.get("verifier_pass") is True and payload.get("grounded") is True:
        return {
            **base,
            "decision":"VERIFIED_GROUNDED_QA_ACCEPT",
            "external_vlm_required":False,
        }
    evidence=payload.get("evidence_top20",[])
    textual=sum(
        1 for z in evidence
        if str(z.get("ocr_text_compact","")).strip() or str(z.get("asr_text_compact","")).strip()
    )
    return {
        **base,
        "decision":"EVIDENCE_READY_GROUNDED_EXTERNAL_VLM_REQUIRED" if evidence else "INSUFFICIENT_EVIDENCE_FAIL_CLOSED",
        "evidence_count":len(evidence),
        "text_supported_evidence_count":textual,
        "external_vlm_required":True,
    }

def _record_online_trace(state,clean_req,plan,started_at,stage):
    p=state["active_result"]["payload"]
    gate=online_risk_evidence_gate(p)
    qid=str(p["query_id"])
    trace={
        "schema_version":ONLINE_CONTROL_SCHEMA,
        "build_version":BUILD_VERSION,
        "query_id":qid,
        "task_type":clean_req["task_type"],
        "stage":str(stage),
        "request_validated":True,
        "branch_plan":plan,
        "gate":gate,
        "active_version":int(state["active_result"]["version"]),
        "elapsed_ms":(time.perf_counter()-started_at)*1000.0,
        "created_at_utc":utc_now(),
    }
    if p.get("task_type")=="TRAKE":
        trace["planner"]=p.get("planner")
    ONLINE_RUNTIME_TRACE[qid]=trace
    return trace

def run_query(req):
    t0=time.perf_counter()
    clean=validate_online_request(req)
    task=clean["task_type"]
    plan=branch_execution_plan(clean)

    if task=="KIS":
        st=run_kis(
            clean["query_text"],clean["query_id"],
            include_multimodal_support=clean["include_multimodal_support"],
        )
        if O2_KIS_SHADOW_ASYNC and not clean["include_multimodal_support"]:
            _qid=st["active_result"]["payload"]["query_id"]
            schedule_kis_o2_shadow(_qid)
    elif task=="TRAKE":
        st=run_trake_fast(
            clean["query_text"],clean["query_id"],
            supplied_events=clean.get("events"),
        )
        if clean["refine"]:
            qid2=st["active_result"]["payload"]["query_id"]
            st=refine_trake_result(
                qid2,
                persist_jpeg=clean["persist_refinement_jpeg"],
            )
    else:
        st=run_qa_evidence(clean["query_text"],clean["query_id"])

    validate_online_active_result(st,task)
    _record_online_trace(st,clean,plan,t0,"POST_BRANCH")
    return st

print("[PASS] Online O2 request validation + federated branch plan + risk/evidence gate")
print("[READY] run_query({task_type: KIS|TRAKE|QA, ...})")
````

````python
# ── F10 / CELL 23: FAIL-CLOSED TECHNICAL SUBMISSION BOUNDARY ──────────────────
# Mục tiêu: provisional evidence can never escape as a final candidate.
# INPUT: active versioned result.
# OUTPUT: technical candidate; official BTC wire-format remains separate.
# Pre-check: task-specific hard gates.
# Post-check: unverified TRAKE / pending QA always blocked.
# History v0.3.0: verified TRAKE/QA may now cross the technical boundary.
# History v0.4.2: TRAKE submission gate validates preserved relation semantics, not unconditional strict-only ordering.
# ──────────────────────────────────────────────────────────────────────────────
class SubmissionBlocked(RuntimeError): pass

def submission_candidate_from_active(state):
    p=state["active_result"]["payload"]
    task=p["task_type"]

    if task=="KIS":
        if p.get("hard_valid") is not True or not p.get("results"):
            raise SubmissionBlocked("KIS_NOT_HARD_VALID")
        r=p["results"][0]
        return {
            "status":"TECHNICAL_CANDIDATE_NOT_OFFICIAL_WIRE_FORMAT",
            "query_id":p["query_id"],
            "task_type":"KIS",
            "video_id":r["video_id"],
            "pts_time":float(r["pts_time"]),
            "frame_ref_id":r["frame_ref_id"],
        }

    if task=="TRAKE":
        if p.get("verifier_pass") is not True or p.get("final_temporal_status")!="RAW_PTS_VERIFIED":
            raise SubmissionBlocked("TRAKE_REQUIRES_RAW_VIDEO_ACTUAL_PTS_DEEP_VERIFIER")
        pts=[float(x) for x in p.get("actual_pts",[])]
        refined=list(p.get("refined_events",[]))
        if len(pts)<2 or len(refined)!=len(pts):
            raise SubmissionBlocked("TRAKE_ACTUAL_PTS_RELATION_PAYLOAD_INVALID")
        for j in range(1,len(pts)):
            rel=str(refined[j].get("relation_from_prev","STRICT_BEFORE"))
            if not _actual_pts_relation_ok(pts[j-1],pts[j],rel):
                raise SubmissionBlocked(("TRAKE_ACTUAL_PTS_RELATION_INVALID",j,rel,pts[j-1],pts[j]))
        return {
            "status":"TECHNICAL_CANDIDATE_NOT_OFFICIAL_WIRE_FORMAT",
            "query_id":p["query_id"],
            "task_type":"TRAKE",
            "video_id":p["video_id"],
            "actual_pts":pts,
        }

    if task=="QA":
        if p.get("verifier_pass") is not True or p.get("grounded") is not True:
            raise SubmissionBlocked("QA_REQUIRES_GROUNDED_EXTERNAL_VLM_AND_VERIFIER")
        answer=str(p.get("answer","")).strip()
        if not answer or not p.get("cited_evidence_ids"):
            raise SubmissionBlocked("QA_VERIFIED_PAYLOAD_INCOMPLETE")
        return {
            "status":"TECHNICAL_CANDIDATE_NOT_OFFICIAL_WIRE_FORMAT",
            "query_id":p["query_id"],
            "task_type":"QA",
            "answer":answer,
            "cited_evidence_ids":list(p["cited_evidence_ids"]),
        }

    raise SubmissionBlocked("UNKNOWN_TASK")

print("[PASS] v0.3 fail-closed technical submission boundary")
````

````python
# ── TEST / CELL 19: FAST KIS SERVING SMOKE + LAZY EVIDENCE SMOKE (NO GT) ─────
# Mục tiêu: prove low-latency fast return, exact dual-GPU search and lazy evidence correctness.
# INPUT: six generic KIS probes; no GT.
# OUTPUT: fast latency + rankings + GPU proof + one selective multimodal evidence timing.
# Pre-check: this is NOT an accuracy benchmark.
# Post-check: every fast query returns hard-valid exact FrameRefs; evidence is tested separately.
# History v0.2.0: p50 ~6.55 s because CPU exact scan + synchronous evidence.
# History v0.2.1: fast path excludes multimodal support; one query validates support lazily.
# History v0.3.0: frozen KIS regression; no ranking/latency architecture change.
# History v0.5.0: lazy evidence smoke is now full O2 federation; active KIS must remain byte-for-byte policy-stable.
# ──────────────────────────────────────────────────────────────────────────────
SMOKE_QUERIES=[
    "a person walking",
    "a car on the road",
    "people sitting at a table",
    "a building exterior",
    "someone holding an object",
    "a person standing near a vehicle",
]

# Warmup excluded from measured latency.
_warm=run_kis("a person",query_id="serving_warmup")
if not _warm["active_result"]["payload"]["results"]:
    raise AssertionError("warmup empty")

SMOKE_ROWS=[]
for i,q in enumerate(SMOKE_QUERIES):
    st=run_kis(q,query_id=f"serving_smoke_{i:02d}")
    p=st["active_result"]["payload"]
    if not p["results"]: raise AssertionError(q)
    alt=st["alternatives"][0]["payload"]
    SMOKE_ROWS.append({
        "query_id":p["query_id"],
        "query":q,
        "raw_top1_frame_ref_id":p["results"][0]["frame_ref_id"],
        "raw_top1_video_id":p["results"][0]["video_id"],
        "raw_top1_pts_time":p["results"][0]["pts_time"],
        "fused_top1_frame_ref_id":alt["results"][0]["frame_ref_id"],
        "top1_changed":p["results"][0]["frame_ref_id"]!=alt["results"][0]["frame_ref_id"],
        "dual_gpu_encode_ms":p["timing_ms"]["dual_gpu_encode"],
        "search_ms":p["timing_ms"]["search_ms"],
        "fast_total_ms":p["timing_ms"]["fast_total"],
        "risk_seed_flags_json":json.dumps(p["risk_diagnostics"]["seed_flags"]),
    })

SMOKE=pd.DataFrame(SMOKE_ROWS)
if len(SMOKE)!=6:
    raise AssertionError(len(SMOKE))
if any(QUERY_USAGE[d]["query_views"]<=0 for d in MODEL_DEVICES):
    raise AssertionError(("dual GPU real query usage",QUERY_USAGE))

# Validate the compact multimodal branch once, OUTSIDE fast latency.
_before_o2=RESULT_STORE.get("serving_smoke_00")["active_result"]["payload"]
_before_top1=_before_o2["results"][0]["frame_ref_id"]
lazy_alt=attach_kis_multimodal_support("serving_smoke_00")
lazy_payload=lazy_alt["payload"]
_after_o2=RESULT_STORE.get("serving_smoke_00")["active_result"]["payload"]
if _after_o2["active_policy"]!="RAW_QUERY_EXACT_OFFICIAL_VISUAL":
    raise AssertionError("O2 KIS shadow changed active policy")
if _after_o2["results"][0]["frame_ref_id"]!=_before_top1:
    raise AssertionError("O2 KIS shadow changed active top1")
if not isinstance(lazy_payload.get("interval_evidence_top20"),list):
    raise AssertionError("lazy multimodal evidence missing")
LAZY_EVIDENCE_SMOKE={
    "status":"PASS",
    "query_id":"serving_smoke_00",
    "evidence_rows":len(lazy_payload["interval_evidence_top20"]),
    "evidence_timing_ms":float(lazy_payload["evidence_timing_ms"]),
    "blocking_fast_result":False,
    "o2_federation_policy":lazy_payload.get("federation_diagnostics",{}).get("policy"),
    "adaptive_shadow_status":lazy_payload.get("adaptive_promotion_status"),
    "active_kis_unchanged":True,
    "accuracy_claim":"NONE_NO_GT",
}

SMOKE.to_csv(ARTIFACT_ROOT/"kis_serving_smoke.csv",index=False)
atomic_json(REPORT_ROOT/"lazy_multimodal_evidence_smoke.json",LAZY_EVIDENCE_SMOKE)
atomic_json(REPORT_ROOT/"dual_gpu_query_usage.json",{
    "status":"PASS",
    "required_gpu_count":2,
    "devices":list(MODEL_DEVICES),
    "usage":QUERY_USAGE,
    "policy":"RAW_GPU0_VISUAL_PROMPT_GPU1_EVERY_QUERY_AND_EXACT_GPU_SEARCH",
    "visual_index_backend":VISUAL_SEARCH_BACKEND,
    "visual_index_replicas":GPU_VISUAL_INDEX_INFO,
    "visual_exact_regression":GPU_VISUAL_REGRESSION,
})
atomic_json(REPORT_ROOT/"kis_smoke_report.json",{
    "status":"PASS",
    "benchmark_type":"TECHNICAL_FAST_SERVING_SMOKE_NO_GT",
    "queries":6,
    "p50_fast_total_ms":float(np.percentile(SMOKE.fast_total_ms,50)),
    "p95_fast_total_ms":float(np.percentile(SMOKE.fast_total_ms,95)),
    "p50_encode_ms":float(np.percentile(SMOKE.dual_gpu_encode_ms,50)),
    "p50_search_ms":float(np.percentile(SMOKE.search_ms,50)),
    "top1_changed_queries":int(SMOKE.top1_changed.sum()),
    "lazy_evidence_smoke":LAZY_EVIDENCE_SMOKE,
    "o2_federation_policy":lazy_payload.get("federation_diagnostics",{}).get("policy"),
    "adaptive_shadow_status":lazy_payload.get("adaptive_promotion_status"),
    "active_kis_unchanged":True,
    "accuracy_claim":"NONE_NO_GT",
})

print("[PASS] frozen KIS v0.2.2 serving smoke",{
    "queries":6,
    "p50_fast_total_ms":float(np.percentile(SMOKE.fast_total_ms,50)),
    "p95_fast_total_ms":float(np.percentile(SMOKE.fast_total_ms,95)),
    "p50_search_ms":float(np.percentile(SMOKE.search_ms,50)),
    "lazy_evidence_ms":LAZY_EVIDENCE_SMOKE["evidence_timing_ms"],
    "dual_gpu_usage":QUERY_USAGE,
})
````

````python
# ── TEST / CELL 25: v0.3 PLANNER + RAW-PTS + LOCAL-DENSE + QA ADAPTER SMOKE ───
# Mục tiêu: prove new selective-refinement machinery without making accuracy claims.
# INPUT: deterministic planner strings + Stage24-proven source video L26_V255.
# OUTPUT: raw actual-PTS proof, both-GPU local image work, QA citation verifier contract.
# Pre-check: this is development source-provenance / interface smoke, NOT GT accuracy.
# Post-check: 19.64→31.68 decode order strict; both GPUs process real raw frames.
# History v0.3.0: first selective-refinement technical closure.
# History v0.4.3: keep legacy two-argument deep-verifier call as an explicit backward-compatibility regression.
# ──────────────────────────────────────────────────────────────────────────────

V03_REG={}

# 1) Planner regressions.
pp=plan_trake_events("a person enters the room then picks up a bag then leaves")
V03_REG["planner_forward"]=pp["events"]==[
    "a person enters the room","picks up a bag","leaves"
]
pa=plan_trake_events("the person leaves after picking up the bag")
V03_REG["planner_after_order"]=pa["events"]==[
    "picking up the bag","the person leaves"
]
pu=plan_trake_events("a single ambiguous event description")
V03_REG["planner_fail_closed"]=pu["planner_status"].startswith("UNRESOLVED")

# 2) PTS-aware raw decode source-provenance smoke.
# These targets are a known Stage24 development source case; this is not an accuracy label.
DEV_VIDEO_ID="L26_V255"
DEV_TARGETS=[19.64,31.68]
dev_dec=[]
for i,t in enumerate(DEV_TARGETS):
    dev_dec.append(decode_raw_first_at_or_after(
        DEV_VIDEO_ID,t,
        request_id=f"v030_dev_pts_{i}",
        persist_jpeg=False,
    ))
dev_pts=[float(x["actual_pts_time"]) for x in dev_dec]
V03_REG["raw_pts_verified"]=all(x["raw_decode_verified"] for x in dev_dec)
V03_REG["first_at_or_after"]=all(
    x["actual_pts_time"]+SEEK_TOLERANCE_S>=x["target_pts_time"]
    for x in dev_dec
)
V03_REG["strict_order"]=dev_pts[1]>dev_pts[0]

# 3) Real dual-GPU local image scoring over the two decoded raw frames.
_,enc,_=encode_dualgpu("technical raw video refinement smoke")
rawv=enc[0]["vector"][0]
parts=[[dev_dec[0]],[dev_dec[1]]]
with ThreadPoolExecutor(max_workers=2) as ex:
    score_groups=list(ex.map(
        _encode_image_subset,
        [(QUERY_WORKERS[0],parts[0],rawv),(QUERY_WORKERS[1],parts[1],rawv)]
    ))
V03_REG["dual_gpu_local_images"]=all(
    LOCAL_DENSE_GPU_USAGE[d]["image_frames"]>=1 for d in MODEL_DEVICES
)
close_decode_images(dev_dec)

# 4) Deep TRAKE verifier contract using actual decoder rows.
_trake_rows=[
    {
        "video_id":DEV_VIDEO_ID,
        "actual_pts_time":dev_pts[0],
        "raw_decode_verified":True,
        "decode_policy":DECODE_POLICY,
    },
    {
        "video_id":DEV_VIDEO_ID,
        "actual_pts_time":dev_pts[1],
        "raw_decode_verified":True,
        "decode_policy":DECODE_POLICY,
    },
]
_tv=deep_verify_trake_refinement("v030_dev",_trake_rows)
V03_REG["trake_deep_verifier"]=_tv["pass"] is True
V03_REG["trake_legacy_verifier_signature"]=(
    _tv.get("relation_authority")=="LEGACY_STRICT_BEFORE_FALLBACK"
)

# 5) Grounded QA adapter/verifier contract with a real Stage20 evidence id but fake callable.
qa_smoke=run_qa_evidence("what is shown in the scene?",query_id="v030_qa_contract")
base_qa=qa_smoke["active_result"]["payload"]
if not base_qa["evidence_top20"]:
    raise AssertionError("QA evidence smoke empty")
register_grounded_vlm("CONTRACT_ONLY","contract-model",_contract_provider)
qa_verified=call_grounded_vlm_for_qa(
    "v030_qa_contract","CONTRACT_ONLY","runtime-secret-not-persisted"
)
qa_active=qa_verified["active_result"]["payload"]
V03_REG["qa_grounded_verifier"]=bool(
    qa_active.get("verifier_pass")
    and qa_active.get("grounded")
    and qa_active.get("api_key_persisted") is False
)
V03_REG["qa_secret_not_stored"]="runtime-secret-not-persisted" not in json.dumps(
    RESULT_STORE.snapshot(),ensure_ascii=False
)

if not all(V03_REG.values()):
    raise AssertionError(V03_REG)

V03_SMOKE={
    "status":"PASS",
    "benchmark_type":"TECHNICAL_SELECTIVE_REFINEMENT_SMOKE_NO_GT",
    "planner":{k:v for k,v in V03_REG.items() if k.startswith("planner_")},
    "raw_pts":{
        "video_id":DEV_VIDEO_ID,
        "targets":DEV_TARGETS,
        "actual_pts":dev_pts,
        "decode_policy":DECODE_POLICY,
        "seek_tolerance_s":SEEK_TOLERANCE_S,
    },
    "dual_gpu_local_image_usage":LOCAL_DENSE_GPU_USAGE,
    "qa_adapter":{
        "provider":"CONTRACT_ONLY",
        "external_network_call":False,
        "api_key_persisted":False,
        "verifier_pass":True,
    },
    "accuracy_claim":"NONE_NO_GT_SOURCE_PROVENANCE_AND_INTERFACE_SMOKE_ONLY",
}
atomic_json(REPORT_ROOT/"v030_selective_refinement_smoke.json",V03_SMOKE)

print("[PASS] v0.3 selective refinement technical smoke",V03_SMOKE)
````

````python
# ── F11 / CELL 27: REAL WORKLOAD DISCOVERY + VALIDATION ───────────────────────
# Mục tiêu: load a real BTC/session workload without making it a required static parent.
# INPUT: optional CSV path/env/mounted file.
# OUTPUT: validated FINAL_WORKLOAD + discovery report.
# Pre-check: query_id unique; KIS/TRAKE/QA only; nonblank query text.
# Post-check: TRAKE supplied events_json is validated if present; otherwise planner may try at runtime.
# History v0.4.0: competition/batch query harness.
# History v0.4.1: remove recursive /kaggle/input crawl; bounded mount-pattern discovery only.
# History v0.4.2: fix pandas leading-underscore itertuples bug; preserve optional/relation_from_prev event semantics.
# History v0.5.18: self-contained strict bool parser; no cross-cell planner dependency.
# ──────────────────────────────────────────────────────────────────────────────

FINAL_WORKLOAD_REQUIRED={"query_id","task_type","query_text"}
FINAL_WORKLOAD_OPTIONAL={
    "events_json","refine","provider_name",
    "gt_video_id","gt_start_pts","gt_end_pts","gt_answer","gt_path_json",
}

def _workload_strict_bool_value(v,field_name):
    """
    Strict bool parser owned by this workload cell.
    No dependency on TRAKE planner cell/global _strict_bool_value.
    """
    if isinstance(v,(bool,np.bool_)):
        return bool(v)
    if v is None:
        return False
    if isinstance(v,(int,np.integer)) and int(v) in (0,1):
        return bool(int(v))
    if isinstance(v,str):
        z=v.strip().lower()
        if z in {"1","true","yes","y"}:
            return True
        if z in {"0","false","no","n",""}:
            return False
    raise ValueError((field_name,"must be strict boolean-like",v))

def _parse_events_json(v):
    if v is None or (isinstance(v,float) and np.isnan(v)) or str(v).strip()=="":
        return None
    if isinstance(v,list):
        data=v
    else:
        data=json.loads(str(v))
    if not isinstance(data,list) or len(data)<2:
        raise ValueError("events_json must be a JSON list with >=2 ordered events")
    out=[]
    for i,item in enumerate(data):
        if isinstance(item,dict):
            txt=str(item.get("text","")).strip()
            rel=str(
                item.get("relation_from_prev",item.get("relation","STRICT_BEFORE"))
            ).strip().upper()
            optional=_workload_strict_bool_value(item.get("optional",False),f"events_json[{i}].optional")
            event_id=str(item.get("event_id",f"e{i}")).strip() or f"e{i}"
        else:
            txt=str(item).strip()
            rel="STRICT_BEFORE"
            optional=False
            event_id=f"e{i}"
        if i==0:
            rel="STRICT_BEFORE"
        if rel not in SUPPORTED_TRAKE_RELATIONS:
            raise ValueError(("unsupported TRAKE relation",rel))
        if not txt:
            raise ValueError("blank event text")
        out.append({
            "event_id":event_id,
            "text":txt,
            "optional":optional,
            "relation_from_prev":rel,
        })
    return out

def validate_final_workload(df):
    if not isinstance(df,pd.DataFrame):
        raise TypeError("workload must be DataFrame")
    miss=sorted(FINAL_WORKLOAD_REQUIRED-set(df.columns))
    if miss:
        raise RuntimeError(("WORKLOAD_SCHEMA_MISSING",miss))
    x=df.copy()
    x["query_id"]=x.query_id.astype(str).str.strip()
    x["task_type"]=x.task_type.astype(str).str.strip().str.upper()
    x["query_text"]=x.query_text.astype(str).str.strip()
    if x.query_id.eq("").any() or x.query_text.eq("").any():
        raise RuntimeError("WORKLOAD_BLANK_ID_OR_TEXT")
    if x.query_id.duplicated().any():
        raise RuntimeError("WORKLOAD_DUPLICATE_QUERY_ID")
    if not x.task_type.isin(["KIS","TRAKE","QA"]).all():
        raise RuntimeError(("WORKLOAD_INVALID_TASK_TYPES",sorted(set(x.task_type)-{"KIS","TRAKE","QA"})))
    if "events_json" in x.columns:
        parsed=[]
        for _,row in x.iterrows():
            ev=_parse_events_json(row.get("events_json"))
            if row["task_type"]!="TRAKE" and ev is not None:
                raise RuntimeError(("EVENTS_JSON_ONLY_FOR_TRAKE",row["query_id"]))
            parsed.append(ev)
        x["events_parsed"]=parsed
    else:
        x["events_parsed"]=[None]*len(x)
    return x

def _bounded_workload_candidates(filename):
    # Critical v0.4.1 rule: never rglob /kaggle/input.
    # Raw Batch1 contains huge nested trees; recursive absence checks are unacceptable.
    base=Path("/kaggle/input")
    if not base.is_dir():
        return []

    patterns=[
        base/filename,
    ]
    # Only inspect a bounded number of directory levels used by Kaggle mounts.
    for p in sorted(base.glob(f"*/{filename}")):
        patterns.append(p)
    for p in sorted(base.glob(f"datasets/*/*/{filename}")):
        patterns.append(p)
    for p in sorted(base.glob(f"notebooks/*/*/{filename}")):
        patterns.append(p)

    return sorted({p.resolve() for p in patterns if p.is_file()},key=str)

def discover_final_workload():
    t0=time.perf_counter()
    explicit=os.environ.get("AIC2026_FINAL_QUERY_WORKLOAD","").strip()
    if explicit:
        p=Path(explicit).expanduser().resolve()
        if not p.is_file():
            raise FileNotFoundError(p)
        candidates=[p]
        discovery_mode="EXPLICIT_ENV_PATH"
    else:
        candidates=_bounded_workload_candidates(WORKLOAD_FILENAME)
        discovery_mode="BOUNDED_KAGGLE_MOUNT_PATTERNS"

    discovery_ms=(time.perf_counter()-t0)*1000.0

    if len(candidates)>1:
        raise RuntimeError(("AMBIGUOUS_FINAL_WORKLOAD",[str(x) for x in candidates]))
    if not candidates:
        return None,{
            "status":"MANUAL_UI_ONLY_NO_WORKLOAD_FILE",
            "filename":WORKLOAD_FILENAME,
            "discovery_mode":discovery_mode,
            "recursive_scan":False,
            "discovery_ms":discovery_ms,
        }

    p=candidates[0]
    df=validate_final_workload(pd.read_csv(p))
    return df,{
        "status":"READY_REAL_WORKLOAD",
        "path":str(p),
        "rows":len(df),
        "task_counts":df.task_type.value_counts().to_dict(),
        "sha256":sha256_file(p),
        "discovery_mode":discovery_mode,
        "recursive_scan":False,
        "discovery_ms":discovery_ms,
    }

FINAL_WORKLOAD,FINAL_WORKLOAD_DISCOVERY=discover_final_workload()
if FINAL_WORKLOAD_DISCOVERY.get("recursive_scan") is not False:
    raise AssertionError("recursive workload discovery is forbidden in v0.4.1")
atomic_json(REPORT_ROOT/"final_workload_discovery.json",FINAL_WORKLOAD_DISCOVERY)
print("[PASS] bounded final workload harness",FINAL_WORKLOAD_DISCOVERY)
````

````python
# ── F11 / CELL 28: BATCH / SESSION EXECUTOR ───────────────────────────────────
# Mục tiêu: execute real workload rows through the same interactive dispatcher.
# INPUT: validated workload; optional runtime provider/API key.
# OUTPUT: session records + result-store state + technical candidates.
# Pre-check: secrets are function args only.
# Post-check: per-row failures are explicit and do not fabricate results.
# History v0.4.0: batch competition executor.
# History v0.4.2: explicit workload-row request builder fixes `_events_parsed` loss and preserves Online O1 trace.
# ──────────────────────────────────────────────────────────────────────────────

SESSION_EVENTS=[]

def _safe_active_summary(st):
    p=st["active_result"]["payload"]
    return {
        "query_id":p.get("query_id"),
        "task_type":p.get("task_type"),
        "hard_valid":p.get("hard_valid"),
        "verifier_pass":p.get("verifier_pass"),
        "active_policy":p.get("active_policy"),
        "final_temporal_status":p.get("final_temporal_status"),
        "qa_status":p.get("qa_status"),
        "grounded":p.get("grounded"),
    }

def execute_session_request(
    req,
    *,
    auto_refine_trake=True,
    qa_provider_name=None,
    qa_api_key=None,
):
    t0=time.perf_counter()
    qid=str(req.get("query_id","")).strip() or None
    task=str(req.get("task_type","")).strip().upper()

    clean_req={
        "query_id":qid,
        "task_type":task,
        "query_text":str(req.get("query_text",req.get("query",""))),
    }
    if task=="TRAKE":
        clean_req["events"]=req.get("events")
        clean_req["refine"]=bool(req.get("refine",auto_refine_trake))

    event={
        "query_id":qid,
        "task_type":task,
        "started_at_utc":utc_now(),
        "status":"RUNNING",
    }
    try:
        st=run_query(clean_req)
        active=st["active_result"]["payload"]

        if task=="QA" and qa_provider_name:
            if not qa_api_key:
                raise RuntimeError("QA_RUNTIME_API_KEY_REQUIRED_FOR_SELECTED_PROVIDER")
            st=call_grounded_vlm_for_qa(
                active["query_id"],
                qa_provider_name,
                qa_api_key,
            )
            active=st["active_result"]["payload"]
            validate_online_active_result(st,"QA")
            _record_online_trace(
                st,
                {"query_id":active["query_id"],"task_type":"QA","query_text":active["query_text"]},
                branch_execution_plan({"task_type":"QA","query_id":active["query_id"],"query_text":active["query_text"]}),
                t0,
                "POST_GROUNDED_EXTERNAL",
            )

        try:
            tech=submission_candidate_from_active(st)
            tech_status="VERIFIED_TECHNICAL_CANDIDATE"
        except SubmissionBlocked as e:
            tech=None
            tech_status=f"BLOCKED::{e}"

        event.update({
            "query_id":active.get("query_id",qid),
            "status":"PASS",
            "elapsed_ms":(time.perf_counter()-t0)*1000.0,
            "active_summary":_safe_active_summary(st),
            "online_control_trace":ONLINE_RUNTIME_TRACE.get(str(active.get("query_id",qid))),
            "technical_candidate_status":tech_status,
            "technical_candidate":tech,
        })
        SESSION_EVENTS.append(event)
        return st,event
    except Exception as e:
        event.update({
            "status":"FAIL",
            "elapsed_ms":(time.perf_counter()-t0)*1000.0,
            "error_type":type(e).__name__,
            "error":str(e),
        })
        SESSION_EVENTS.append(event)
        raise

def workload_row_to_request(row,columns,auto_refine_trake=True):
    """Convert one validated pandas Series to a dispatcher request without attribute-name rewriting."""
    req={
        "query_id":str(row["query_id"]),
        "task_type":str(row["task_type"]),
        "query_text":str(row["query_text"]),
    }
    if req["task_type"]=="TRAKE":
        req["events"]=row.get("events_parsed")
        if "refine" in columns:
            rv=row.get("refine",auto_refine_trake)
            if isinstance(rv,str):
                req["refine"]=rv.strip().lower() in {"1","true","yes","y"}
            else:
                req["refine"]=bool(rv)
        else:
            req["refine"]=bool(auto_refine_trake)
    return req

def execute_workload(
    workload_df,
    *,
    auto_refine_trake=True,
    qa_provider_name=None,
    qa_api_key=None,
    continue_on_error=True,
):
    x=validate_final_workload(workload_df)
    rows=[]
    for _,r in x.iterrows():
        req=workload_row_to_request(r,x.columns,auto_refine_trake)
        try:
            _,evrec=execute_session_request(
                req,
                auto_refine_trake=auto_refine_trake,
                qa_provider_name=qa_provider_name if r["task_type"]=="QA" else None,
                qa_api_key=qa_api_key if r["task_type"]=="QA" else None,
            )
            rows.append(evrec)
        except Exception:
            rows.append(SESSION_EVENTS[-1])
            if not continue_on_error:
                raise
    return pd.DataFrame(rows)

print("[READY] execute_session_request(...)")
print("[READY] execute_workload(FINAL_WORKLOAD, ...)")
````

````python
# ── F12 / CELL 29: CONFIGURABLE OPENAI-COMPATIBLE GROUNDED HTTP ADAPTER ───────
# Mục tiêu: real network-capable adapter without provider/model lock.
# INPUT: full endpoint URL + provider label + model + runtime API key.
# OUTPUT: register_grounded_vlm-compatible callable.
# Pre-check: endpoint/model are explicit; API key never stored in registry.
# Post-check: response must contain strict JSON answer/citations/abstain.
# History v0.4.0: configurable external integration boundary.
# ──────────────────────────────────────────────────────────────────────────────
import base64
import requests

def _frame_ref_image_data_url(frame_ref_id,max_bytes=4_000_000):
    if not frame_ref_id:
        return None
    hit=VISUAL_META[VISUAL_META.frame_ref_id.astype(str).eq(str(frame_ref_id))]
    if hit.empty:
        return None
    rel=str(hit.iloc[0].image_artifact_ref)
    p=(RAW_ROOT/rel).resolve()
    rr=RAW_ROOT.resolve()
    if rr!=p and rr not in p.parents:
        raise RuntimeError("IMAGE_ARTIFACT_PATH_ESCAPED_RAW_ROOT")
    if not p.is_file() or p.stat().st_size>max_bytes:
        return None
    ext=p.suffix.lower()
    mime="image/jpeg" if ext in {".jpg",".jpeg"} else "image/png" if ext==".png" else None
    if mime is None:
        return None
    b64=base64.b64encode(p.read_bytes()).decode("ascii")
    return f"data:{mime};base64,{b64}"

def _parse_json_content(content):
    if isinstance(content,dict):
        return content
    text=str(content).strip()
    if text.startswith("```"):
        text=re.sub(r"^```(?:json)?\s*","",text,flags=re.I)
        text=re.sub(r"\s*```$","",text)
    data=json.loads(text)
    if not isinstance(data,dict):
        raise TypeError("external grounded response JSON must be object")
    return data

def register_openai_compatible_grounded_vlm(
    provider_name,
    endpoint_url,
    model_name,
    *,
    timeout_s=45,
    max_images=4,
):
    provider=str(provider_name).strip()
    endpoint=str(endpoint_url).strip()
    model=str(model_name).strip()
    if not provider or not endpoint or not model:
        raise ValueError("provider, endpoint URL and model are required")
    if not re.match(r"^https?://",endpoint,re.I):
        raise ValueError("endpoint must be an explicit http(s) URL")

    # Closure stores endpoint/model only. No secret.
    def _callable(*,api_key,model,request):
        key=str(api_key or "")
        if not key:
            raise RuntimeError("RUNTIME_API_KEY_REQUIRED")

        evidence=request["evidence"]
        user_blocks=[{
            "type":"text",
            "text":json.dumps({
                "query_text":request["query_text"],
                "evidence":evidence,
                "required_response_schema":request["required_response_schema"],
            },ensure_ascii=False),
        }]

        # Best-effort multimodal support from retrieved frame refs. Providers that
        # implement OpenAI-compatible image_url content can consume these blocks.
        image_count=0
        base=RESULT_STORE.get(request["query_id"])["active_result"]["payload"]
        for e in base.get("evidence_top20",[]):
            if image_count>=int(max_images):
                break
            fr=e.get("best_frame_ref_id")
            data_url=_frame_ref_image_data_url(fr)
            if data_url:
                user_blocks.append({
                    "type":"image_url",
                    "image_url":{"url":data_url},
                })
                image_count+=1

        payload={
            "model":model,
            "messages":[
                {
                    "role":"system",
                    "content":"Answer ONLY from supplied evidence. Return JSON only with keys answer, cited_evidence_ids, abstain. Cite only supplied evidence_pack_id values. If evidence is insufficient, set abstain=true."
                },
                {
                    "role":"user",
                    "content":user_blocks,
                },
            ],
            "temperature":0,
        }
        headers={
            "Authorization":f"Bearer {key}",
            "Content-Type":"application/json",
        }
        resp=requests.post(
            endpoint,
            headers=headers,
            json=payload,
            timeout=float(timeout_s),
        )
        resp.raise_for_status()
        body=resp.json()
        try:
            content=body["choices"][0]["message"]["content"]
        except Exception as e:
            raise RuntimeError(("OPENAI_COMPATIBLE_RESPONSE_SCHEMA_MISMATCH",body)) from e
        return _parse_json_content(content)

    register_grounded_vlm(provider,model,_callable)
    # Return nonsecret provenance only.
    return {
        "provider":provider,
        "model":model,
        "endpoint":endpoint,
        "transport":"OPENAI_COMPATIBLE_CHAT_COMPLETIONS_STYLE",
        "api_key_persisted":False,
        "max_images":int(max_images),
    }

print("[READY] register_openai_compatible_grounded_vlm(provider, endpoint, model)")
````

````python
# ── F13 / CELL 30: OFFICIAL SUBMISSION ADAPTER REGISTRY ───────────────────────
# Mục tiêu: keep official BTC wire format strictly separate from technical candidates.
# INPUT: exact organizer adapter callable + schema name + schema hash.
# OUTPUT: official export only after explicit registration.
# Pre-check: no built-in guessed schema; competition rows come from SESSION_EVENTS only.
# Post-check: unregistered export always raises; smoke/test RESULT_STORE entries cannot leak.
# History v0.4.0: final fail-closed organizer boundary.
# History v0.4.1: session-scoped candidate/store export; development smoke contamination blocked.
# ──────────────────────────────────────────────────────────────────────────────

OFFICIAL_SUBMISSION_ADAPTER=None
OFFICIAL_SUBMISSION_SCHEMA=None

class OfficialSubmissionSchemaUnavailable(RuntimeError):
    pass

def register_official_submission_adapter(schema_name,schema_sha256,adapter_fn):
    global OFFICIAL_SUBMISSION_ADAPTER,OFFICIAL_SUBMISSION_SCHEMA
    name=str(schema_name).strip()
    sha=str(schema_sha256).strip().lower()
    if not name or not re.fullmatch(r"[0-9a-f]{64}",sha):
        raise ValueError("exact schema name and 64-char SHA256 are required")
    if not callable(adapter_fn):
        raise TypeError("adapter_fn must be callable")
    OFFICIAL_SUBMISSION_ADAPTER=adapter_fn
    OFFICIAL_SUBMISSION_SCHEMA={
        "schema_name":name,
        "schema_sha256":sha,
        "registered_at_utc":utc_now(),
    }
    return dict(OFFICIAL_SUBMISSION_SCHEMA)

def competition_session_query_ids():
    # SESSION_EVENTS is the sole authority for competition/manual-session membership.
    # Internal warmups/smokes never enter this list unless explicitly executed via
    # execute_session_request(), which is intentional and auditable.
    out=[]
    seen=set()
    for ev in SESSION_EVENTS:
        qid=str(ev.get("query_id","")).strip()
        if not qid or qid in seen:
            continue
        seen.add(qid)
        out.append(qid)
    return out

def competition_result_store_snapshot():
    qids=competition_session_query_ids()
    full=RESULT_STORE.snapshot()
    subset={qid:full["queries"][qid] for qid in qids if qid in full["queries"]}
    return {
        "schema_version":"versioned-result-store-v1",
        "build_version":BUILD_VERSION,
        "scope":"COMPETITION_SESSION_EVENTS_ONLY",
        "query_ids":qids,
        "queries":subset,
    }

def verified_technical_candidates():
    # Default and only official-facing candidate source = session events.
    rows=[]
    full=RESULT_STORE.snapshot()
    for qid in competition_session_query_ids():
        st=full["queries"].get(qid)
        if st is None:
            continue
        try:
            cand=submission_candidate_from_active(st)
        except SubmissionBlocked:
            continue
        rows.append(cand)
    return rows

def official_submission_export(output_path):
    if OFFICIAL_SUBMISSION_ADAPTER is None or OFFICIAL_SUBMISSION_SCHEMA is None:
        raise OfficialSubmissionSchemaUnavailable(OFFICIAL_SUBMISSION_STATUS)

    technical=verified_technical_candidates()
    scoped_store=competition_result_store_snapshot()
    out=OFFICIAL_SUBMISSION_ADAPTER(
        technical_candidates=technical,
        result_store_snapshot=scoped_store,
        schema=dict(OFFICIAL_SUBMISSION_SCHEMA),
    )

    path=Path(output_path)
    path.parent.mkdir(parents=True,exist_ok=True)
    if isinstance(out,pd.DataFrame):
        out.to_csv(path,index=False)
    elif isinstance(out,(dict,list)):
        path.write_text(json.dumps(out,ensure_ascii=False,indent=2),encoding="utf-8")
    elif isinstance(out,(bytes,bytearray)):
        path.write_bytes(bytes(out))
    else:
        raise TypeError(("UNSUPPORTED_OFFICIAL_ADAPTER_OUTPUT",type(out)))
    return {
        "path":str(path),
        "size_bytes":int(path.stat().st_size),
        "sha256":sha256_file(path),
        "schema":dict(OFFICIAL_SUBMISSION_SCHEMA),
        "session_query_count":len(competition_session_query_ids()),
        "verified_candidate_count":len(technical),
    }

# Regression: must be blocked by default.
try:
    official_submission_export(WORK_ROOT/"should_not_exist.csv")
    raise AssertionError("official export must fail before exact schema registration")
except OfficialSubmissionSchemaUnavailable:
    pass

print("[PASS] official submission registry fail-closed + session-scoped",OFFICIAL_SUBMISSION_STATUS)
````

````python
# ── F13 / CELL 31: SESSION-SCOPED TECHNICAL EXPORT ────────────────────────────
# Mục tiêu: export only real/manual session states, never internal warmups/smokes.
# INPUT: SESSION_EVENTS + matching RESULT_STORE query states.
# OUTPUT: session JSON + verified candidate CSV + session-event CSV.
# Pre-check: SESSION_EVENTS is membership authority.
# Post-check: no development-smoke payload and no secret/API key fields.
# History v0.4.0: exported whole RESULT_STORE and could leak smoke candidates.
# History v0.4.1: session-scoped result store and candidates only.
# ──────────────────────────────────────────────────────────────────────────────

def _assert_no_secret_blob(obj):
    text=json.dumps(obj,ensure_ascii=False,default=str).lower()
    forbidden=[
        "runtime-secret-not-persisted",
        '"authorization"',
        '"api_key":',
        '"secret":',
    ]
    bad=[x for x in forbidden if x in text]
    if bad:
        raise RuntimeError(("SECRET_LIKE_CONTENT_IN_EXPORT",bad))

def export_technical_session():
    scoped_store=competition_result_store_snapshot()
    candidates=verified_technical_candidates()

    payload={
        "schema_version":SESSION_EXPORT_SCHEMA,
        "build_version":BUILD_VERSION,
        "scope":"COMPETITION_SESSION_EVENTS_ONLY",
        "official_submission_status":OFFICIAL_SUBMISSION_STATUS,
        "result_store":scoped_store,
        "session_events":SESSION_EVENTS,
        "runtime_self_test_summary":{
            "internal_result_store_query_count":len(RESULT_STORE.snapshot()["queries"]),
            "excluded_internal_query_count":max(
                0,len(RESULT_STORE.snapshot()["queries"])-len(scoped_store["queries"])
            ),
            "internal_queries_exported_as_candidates":0,
        },
        "created_at_utc":utc_now(),
    }
    _assert_no_secret_blob(payload)

    session_json=EXPORT_ROOT/"technical_session_snapshot.json"
    atomic_json(session_json,payload)

    cand_csv=EXPORT_ROOT/"verified_technical_candidates.csv"
    if candidates:
        pd.DataFrame(candidates).to_csv(cand_csv,index=False)
    else:
        # Stable zero-row schema, not a one-byte blank CSV.
        pd.DataFrame(columns=[
            "status","query_id","task_type","video_id",
            "pts_time","frame_ref_id","actual_pts","answer","cited_evidence_ids",
        ]).to_csv(cand_csv,index=False)

    events_csv=EXPORT_ROOT/"session_events.csv"
    if SESSION_EVENTS:
        pd.DataFrame(SESSION_EVENTS).to_csv(events_csv,index=False)
    else:
        pd.DataFrame(columns=[
            "query_id","task_type","started_at_utc","status","elapsed_ms",
            "technical_candidate_status",
        ]).to_csv(events_csv,index=False)

    return {
        "session_json":str(session_json),
        "candidates_csv":str(cand_csv),
        "events_csv":str(events_csv),
        "session_query_ids":competition_session_query_ids(),
        "verified_candidates":len(candidates),
        "session_events":len(SESSION_EVENTS),
        "excluded_internal_query_count":payload["runtime_self_test_summary"]["excluded_internal_query_count"],
        "official_submission_status":OFFICIAL_SUBMISSION_STATUS,
    }

print("[READY] export_technical_session()")
````

````python
# ── FINAL PATCH / CELL: FROZEN KIS + TRAKE/QA CONTRACT + BTC ADAPTER ──────────
# This cell intentionally runs AFTER the original online1 technical harness and
# BEFORE the public server cell.
#
# It changes the active KIS serving backend to the final promoted CLIP corpus,
# keeps the frozen TRAKE v3/raw-PTS engine, activates final-runtime QA provider
# selection, and enables the now-known AIC26 preliminary CSV wire format.
# ──────────────────────────────────────────────────────────────────────────────

from pathlib import Path
from concurrent.futures import ThreadPoolExecutor
import io, base64, csv, zipfile, hashlib, json, math, os, re, time
import subprocess, sys, threading, socket, shutil
import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F
import transformers
from transformers import CLIPModel, CLIPProcessor
from huggingface_hub import snapshot_download

# ---- final identity -----------------------------------------------------------
LEGACY_ONLINE_BUILD_VERSION=BUILD_VERSION
BUILD_VERSION="aic2026-final-competition-server-v1.0.21"
STAGE_ID="FINAL_competition_multitask_server"

FINAL_FROZEN_CONTRACTS={
    "KIS":{
        "source_stage":"50MEGArr",
        "handoff_sha256":"20050888c4ac7944c2e96c1effe2305b7053cc17a5471b44013eb98a32de93d2",
        "contract_sha256":"33e975018ac032e62ca7303d354f374f6e69507a20a020dd18b2d6796cc43139",
        "status":"FROZEN",
        "selected_run_name":"CLIP_ONLY",
        "selected_config_id":"CLIP_VIT_B32_ORGANIZER_SHADOW",
        "promotion_score_internal":0.388,
        "official_competition_score_claim":False,
    },
    "TRAKE":{
        "source_stage":"51TRAKEMEGArr",
        "handoff_sha256":"e225904954adac2b81f57e75c72c917bc678fc7ddd8882131339ea399cb755f1",
        "contract_sha256":"4735576b274f0e227545ea50259f4f274a7f90f17f9961fe305c36b58488cc61",
        "status":"FROZEN_TECHNICAL",
        "temporal_authority":"RAW_VIDEO_ACTUAL_PTS",
    },
    "QA":{
        "source_stage":"52QAMEGArr",
        "handoff_sha256":"dfdb18ceae08382cedc3b08be860e054fb5511487a3a5576ca98d5cea8467478",
        "contract_sha256":"8cd53b23b21920d45924198c31aefe96c3c1d7d7374023ba959a3eefa73583d6",
        "status":"READY_EXTERNAL_API_PENDING",
        "provider_lock":"NONE",
        "model_lock":"NONE",
    },
}

OFFICIAL_SUBMISSION_STATUS="READY_AIC26_PRELIMINARY_SPEC_PINNED"
BTC_SUBMISSION_ROOT=Path("/kaggle/working/submission")
BTC_SUBMISSION_ROOT.mkdir(parents=True,exist_ok=True)
BTC_ZIP_PATH=Path("/kaggle/working/submission.zip")

# KIS final runtime still requires exactly two GPUs.
if not torch.cuda.is_available() or torch.cuda.device_count()!=2:
    raise RuntimeError(("FINAL_EXACTLY_TWO_CUDA_GPUS_REQUIRED",
                        torch.cuda.device_count() if torch.cuda.is_available() else 0))

# Disable the old Stage03-active KIS asynchronous shadow, because KIS active is
# now the promoted CLIP backend.
O2_KIS_SHADOW_ASYNC=False

# ---- exact final Component06 CLIP corpus -------------------------------------
FINAL_CLIP_MODEL_ID="openai/clip-vit-base-patch32"
FINAL_CLIP_REVISION="3d74acf9a28c67741b2f4f2ea7635f0aaf6f0268"
FINAL_CLIP_WEIGHT_SHA="a63082132ba4f97a80bea76823f544493bffa8082296d62d71581a4feff1576f"
FINAL_COMP06_BUILD="aic2026-kis-rrf-shadow-v0.2.1"
FINAL_COMP06_HANDOFF_SHA="787ad4472a8ff0149a2338dcb3c0650d1f72b10446495d58126cafa1c595d5ca"
FINAL_CLIP_DIM=512
FINAL_CLIP_ROWS=177321
FINAL_KIS_OUTPUT_TOPK=100
FINAL_CLIP_PARITY_MIN=0.9999

FINAL_COMP06_ROOT=resolve_dataset("selaninguyen/6rrnote")
FINAL_H06_PATH=unique_under(FINAL_COMP06_ROOT,"component06_handoff.json")
if sha256_file(FINAL_H06_PATH)!=FINAL_COMP06_HANDOFF_SHA:
    raise RuntimeError(("FINAL_COMP06_HANDOFF_SHA",sha256_file(FINAL_H06_PATH)))
FINAL_H06=read_json(FINAL_H06_PATH)
if FINAL_H06.get("build_version")!=FINAL_COMP06_BUILD:
    raise RuntimeError(("FINAL_COMP06_BUILD",FINAL_H06.get("build_version")))

FINAL_R06_PATH=unique_under(FINAL_COMP06_ROOT,"component06_rrf_shadow_report.json")
FINAL_R06=read_json(FINAL_R06_PATH)
if FINAL_R06.get("canonical_identity")!="STAGE03_FRAME_REF_ID":
    raise RuntimeError(("FINAL_COMP06_CANONICAL_ID",FINAL_R06.get("canonical_identity")))
if FINAL_R06.get("pts_rounding_identity_used") is not False:
    raise RuntimeError("FINAL_COMP06_PTS_ROUNDING_FORBIDDEN")

FINAL_CLIP_MATRIX_PATH=unique_under(
    FINAL_COMP06_ROOT,"legacy_clip_aligned_stage03_order.fp16.npy"
)
FINAL_CLIP_NP=np.load(FINAL_CLIP_MATRIX_PATH,mmap_mode="r")
if FINAL_CLIP_NP.shape!=(FINAL_CLIP_ROWS,FINAL_CLIP_DIM):
    raise RuntimeError(("FINAL_CLIP_CORPUS_SHAPE",FINAL_CLIP_NP.shape))
if len(VISUAL_META)!=FINAL_CLIP_ROWS:
    raise RuntimeError(("FINAL_STAGE03_ROWS",len(VISUAL_META)))
if not np.array_equal(
    VISUAL_META.index_row.to_numpy(np.int64),
    np.arange(FINAL_CLIP_ROWS,dtype=np.int64),
):
    raise RuntimeError("FINAL_STAGE03_ORDER_DRIFT")

# ---- pinned CLIP model --------------------------------------------------------
if transformers.__version__!="5.0.0":
    raise RuntimeError(("FINAL_TRANSFORMERS_VERSION",
                        transformers.__version__,"5.0.0"))

clip_env=os.environ.get("AIC2026_CLIP_MODEL_DIR","").strip()
if clip_env:
    FINAL_CLIP_MODEL_DIR=Path(clip_env).resolve()
    if not (FINAL_CLIP_MODEL_DIR/"config.json").is_file():
        raise RuntimeError(("AIC2026_CLIP_MODEL_DIR_INVALID",str(FINAL_CLIP_MODEL_DIR)))
    FINAL_CLIP_MODEL_SOURCE="EXPLICIT_LOCAL_DIR"
else:
    FINAL_CLIP_MODEL_DIR=Path(snapshot_download(
        repo_id=FINAL_CLIP_MODEL_ID,
        revision=FINAL_CLIP_REVISION,
        allow_patterns=[
            "config.json","preprocessor_config.json","tokenizer.json",
            "tokenizer_config.json","vocab.json","merges.txt",
            "special_tokens_map.json","model.safetensors","pytorch_model.bin",
        ],
    )).resolve()
    FINAL_CLIP_MODEL_SOURCE="HF_PINNED_REVISION"

_weights=[
    p for p in (
        FINAL_CLIP_MODEL_DIR/"model.safetensors",
        FINAL_CLIP_MODEL_DIR/"pytorch_model.bin",
    ) if p.is_file()
]
if len(_weights)!=1:
    raise RuntimeError(("FINAL_CLIP_WEIGHT_COUNT",len(_weights),[str(x) for x in _weights]))
FINAL_CLIP_WEIGHT_PATH=_weights[0]
if sha256_file(FINAL_CLIP_WEIGHT_PATH)!=FINAL_CLIP_WEIGHT_SHA:
    raise RuntimeError(("FINAL_CLIP_WEIGHT_SHA",
                        sha256_file(FINAL_CLIP_WEIGHT_PATH),FINAL_CLIP_WEIGHT_SHA))

def _final_load_clip_worker(gpu_id):
    torch.cuda.set_device(gpu_id)
    dev=f"cuda:{gpu_id}"
    proc=CLIPProcessor.from_pretrained(
        str(FINAL_CLIP_MODEL_DIR),local_files_only=True,use_fast=True
    )
    model=CLIPModel.from_pretrained(
        str(FINAL_CLIP_MODEL_DIR),
        local_files_only=True,
        torch_dtype=torch.float16,
        low_cpu_mem_usage=True,
    ).eval().to(dev)

    # Preserve the executed promotion hotfix: CLIP text path is FP32 under
    # transformers 5.0 even though the rest of the model may be FP16.
    model.text_model.to(device=dev,dtype=torch.float32)
    model.text_projection.to(device=dev,dtype=torch.float32)

    if {p.dtype for p in model.text_model.parameters()}!={torch.float32}:
        raise RuntimeError(("FINAL_CLIP_TEXT_DTYPE",dev))
    if {p.dtype for p in model.text_projection.parameters()}!={torch.float32}:
        raise RuntimeError(("FINAL_CLIP_PROJ_DTYPE",dev))
    return {"gpu_id":gpu_id,"device":dev,"processor":proc,"model":model}

with ThreadPoolExecutor(max_workers=2) as ex:
    FINAL_CLIP_WORKERS=list(ex.map(_final_load_clip_worker,(0,1)))

_split=(FINAL_CLIP_ROWS+1)//2
_ranges=((0,_split),(_split,FINAL_CLIP_ROWS))

def _final_attach_clip_shard(worker,ab):
    a,b=ab
    torch.cuda.set_device(worker["gpu_id"])
    x=torch.from_numpy(
        np.asarray(FINAL_CLIP_NP[a:b],dtype=np.float32).copy()
    ).to(worker["device"])
    x=F.normalize(x,dim=1)
    worker["row_start"]=a
    worker["row_end"]=b
    worker["clip_shard"]=x
    return {"device":worker["device"],"rows":b-a,"shape":tuple(x.shape)}

with ThreadPoolExecutor(max_workers=2) as ex:
    FINAL_CLIP_SHARDS=list(ex.map(
        _final_attach_clip_shard,FINAL_CLIP_WORKERS,_ranges
    ))
if sum(x["rows"] for x in FINAL_CLIP_SHARDS)!=FINAL_CLIP_ROWS:
    raise RuntimeError("FINAL_CLIP_SHARD_TOTAL")

FINAL_CLIP_GPU_USAGE={
    "cuda:0":{"text_encodes":0,"searches":0},
    "cuda:1":{"text_encodes":0,"searches":0},
}

def _final_clip_encode_one(worker,text):
    torch.cuda.set_device(worker["gpu_id"])
    # OpenAI CLIP text encoder has max_position_embeddings=77.
    # Manual BTC queries can be much longer, so fail-safe truncate at the
    # tokenizer boundary instead of crashing with sequence_length > 77.
    inp=worker["processor"](
        text=[str(text)],
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=77,
    )
    inp={k:v.to(worker["device"]) for k,v in inp.items()}
    with torch.inference_mode():
        raw=worker["model"].get_text_features(**inp)
        if hasattr(raw,"pooler_output") and raw.pooler_output is not None:
            raw=raw.pooler_output
        if not torch.is_tensor(raw):
            raise TypeError(("FINAL_CLIP_TEXT_OUTPUT",type(raw)))
        v=F.normalize(raw.float(),dim=-1)[0]
        if not torch.isfinite(v).all():
            raise RuntimeError(("FINAL_CLIP_NONFINITE_QUERY",worker["device"]))
    FINAL_CLIP_GPU_USAGE[worker["device"]]["text_encodes"]+=1
    return v.detach().cpu().numpy().astype(np.float32)

def _final_clip_encode_replicated(text):
    with ThreadPoolExecutor(max_workers=2) as ex:
        vecs=list(ex.map(
            lambda w:_final_clip_encode_one(w,text),
            FINAL_CLIP_WORKERS
        ))
    a=vecs[0]/max(float(np.linalg.norm(vecs[0])),1e-12)
    b=vecs[1]/max(float(np.linalg.norm(vecs[1])),1e-12)
    cos=float(np.dot(a,b))
    if cos<FINAL_CLIP_PARITY_MIN:
        raise RuntimeError(("FINAL_CLIP_REPLICA_PARITY",cos,FINAL_CLIP_PARITY_MIN))
    return a.astype(np.float32),cos

def _final_clip_search_shard(worker,q,k):
    torch.cuda.set_device(worker["gpu_id"])
    qt=torch.from_numpy(np.asarray(q,dtype=np.float32)).to(worker["device"])
    with torch.inference_mode():
        scores=worker["clip_shard"]@qt
        kk=min(int(k),len(scores))
        val,idx=torch.topk(scores,k=kk,largest=True,sorted=True)
    FINAL_CLIP_GPU_USAGE[worker["device"]]["searches"]+=1
    return (
        val.detach().cpu().numpy().astype(np.float32),
        idx.detach().cpu().numpy().astype(np.int64)+int(worker["row_start"]),
    )

def _final_clip_search(q,k=100):
    with ThreadPoolExecutor(max_workers=2) as ex:
        got=list(ex.map(
            lambda w:_final_clip_search_shard(w,q,k),
            FINAL_CLIP_WORKERS
        ))
    vals=np.concatenate([x[0] for x in got])
    inds=np.concatenate([x[1] for x in got])
    kk=min(int(k),len(vals))
    sel=np.argpartition(-vals,kk-1)[:kk]
    order=np.lexsort((inds[sel],-vals[sel]))
    sel=sel[order]
    return vals[sel],inds[sel]

def _final_kis_rows(scores,indices):
    rows=[]
    for rank,(score,idx) in enumerate(zip(scores,indices),1):
        r=VISUAL_META.iloc[int(idx)]
        rows.append({
            "rank":rank,
            "frame_ref_id":str(r.frame_ref_id),
            "video_id":str(r.video_id),
            "pts_time":float(r.pts_time),
            "frame_idx_candidate":int(r.frame_idx_candidate),
            "image_artifact_ref":str(getattr(r,"image_artifact_ref","")),
            "raw_video_ref":str(getattr(r,"raw_video_ref","")),
            "clip_cosine":float(score),
            # Compatibility aliases consumed by the original UI walkers.
            "raw_cosine":float(score),
            "prompt_cosine":float(score),
            "dual_view_fused_cosine":float(score),
            "rrf_pool_score_diagnostic":None,
        })
    return rows

# Preserve the pre-promotion local KIS backend as an explicit emergency fallback.
# At this point `run_kis` is still the original Online1/Stage03 SigLIP2 path.
# The final CLIP override below remains the normal primary backend.
run_kis_stage03_local_fallback = run_kis

# FINAL ACTIVE KIS — override old Stage03 active backend.
def run_kis(query,query_id=None,include_multimodal_support=False):
    q=canonical_query_text(query)
    if query_id is None:
        query_id="kis_"+hashlib.sha256(q.encode()).hexdigest()[:16]
    t0=time.perf_counter()
    qvec,parity=_final_clip_encode_replicated(q)
    scores,indices=_final_clip_search(qvec,FINAL_KIS_OUTPUT_TOPK)
    rows=_final_kis_rows(scores,indices)
    if not rows:
        raise RuntimeError("FINAL_KIS_EMPTY")
    payload={
        "schema_version":"final-kis-clip-promoted-v1",
        "build_version":BUILD_VERSION,
        "query_id":query_id,
        "task_type":"KIS",
        "query_text":q,
        "active_policy":"CLIP_VIT_B32_ORGANIZER_SHADOW_PROMOTED_FINAL",
        "selected_config_id":"CLIP_VIT_B32_ORGANIZER_SHADOW",
        "hard_valid":True,
        "results":rows,
        "risk_diagnostics":{"seed_flags":[],"final_backend_frozen":True},
        "normalization_status":"L2_NORMALIZED_EXACT_COSINE",
        "probability_calibration_status":"NOT_CALIBRATED",
        "official_competition_score_claim":False,
        "timing_ms":{"total":(time.perf_counter()-t0)*1000.0},
        "dual_gpu_text_replica_cosine":parity,
    }
    RESULT_STORE.put_initial(query_id,payload)
    return RESULT_STORE.get(query_id)

# Update branch-plan semantics without changing TRAKE/QA retrieval engines.
_old_branch_execution_plan=branch_execution_plan
def branch_execution_plan(clean_req):
    if clean_req["task_type"]=="KIS":
        return {
            "task_type":"KIS",
            "primary":"FINAL_PROMOTED_CLIP_VIT_B32_STAGE03_ORDER",
            "selected_config_id":"CLIP_VIT_B32_ORGANIZER_SHADOW",
            "replacement_authority":"FROZEN_FINAL_KIS_BACKEND_CONTRACT",
        }
    return _old_branch_execution_plan(clean_req)

# ---- official Stage03 frame adapter ------------------------------------------
_FINAL_META_BY_VIDEO={
    vid:g.sort_values(["pts_time","frame_idx_candidate","frame_ref_id"]).reset_index(drop=True)
    for vid,g in VISUAL_META.groupby("video_id",sort=False)
}

def _final_same_video_nearest_frame(video_id,target_pts,after_frame=None):
    vid=str(video_id).strip()
    if vid not in _FINAL_META_BY_VIDEO:
        raise RuntimeError(("FINAL_VIDEO_NOT_IN_STAGE03",vid))
    t=float(target_pts)
    if not math.isfinite(t) or t<0:
        raise RuntimeError(("FINAL_INVALID_PTS",target_pts))
    g=_FINAL_META_BY_VIDEO[vid]
    if after_frame is not None:
        g=g[g.frame_idx_candidate.astype(np.int64)>int(after_frame)]
        if g.empty:
            raise RuntimeError(("FINAL_NO_LATER_FRAME",vid,after_frame,t))
    d=(g.pts_time.astype(float)-t).abs()
    x=g.loc[d.idxmin()]
    return {
        "video_id":vid,
        "frame_ref_id":str(x.frame_ref_id),
        "frame_idx":int(x.frame_idx_candidate),
        "stage03_pts_time":float(x.pts_time),
        "actual_pts_time":t,
        "abs_map_error_sec":abs(float(x.pts_time)-t),
    }

def _final_trake_btc(active_payload):
    if active_payload.get("task_type")!="TRAKE":
        raise RuntimeError("NOT_TRAKE")
    if active_payload.get("verifier_pass") is not True:
        raise RuntimeError("TRAKE_NOT_VERIFIED")
    if active_payload.get("final_temporal_status")!="RAW_PTS_VERIFIED":
        raise RuntimeError(("TRAKE_FINAL_TEMPORAL_STATUS",
                            active_payload.get("final_temporal_status")))
    vid=str(active_payload.get("video_id") or active_payload.get("recovered_video_id") or "").strip()
    pts=active_payload.get("actual_pts",[])
    if not isinstance(pts,list) or not (2<=len(pts)<=5):
        raise RuntimeError(("TRAKE_ACTUAL_PTS_COUNT",len(pts) if isinstance(pts,list) else None))
    mapped=[]; prev=-1
    for t in pts:
        m=_final_same_video_nearest_frame(vid,float(t),after_frame=prev)
        mapped.append(m); prev=m["frame_idx"]
    frames=[x["frame_idx"] for x in mapped]
    if not all(frames[i]<frames[i+1] for i in range(len(frames)-1)):
        raise RuntimeError(("TRAKE_BTC_FRAME_ORDER",frames))
    return [[vid,*frames]],mapped

# ---- QA agentic final-runtime helper -----------------------------------------
def _final_ev_time(x):
    fr=str(x.get("best_frame_ref_id") or "").strip()
    if fr:
        hit=VISUAL_META[VISUAL_META.frame_ref_id.astype(str).eq(fr)]
        if len(hit)==1:
            return float(hit.iloc[0].pts_time)
    for k in ("pts_time","actual_pts_time","start_pts_time_technical","end_pts_time_technical"):
        v=x.get(k)
        try:
            f=float(v)
            if math.isfinite(f):
                if k=="start_pts_time_technical":
                    try:
                        e=float(x.get("end_pts_time_technical"))
                        if math.isfinite(e): return 0.5*(f+e)
                    except Exception:
                        pass
                return f
        except Exception:
            pass
    return None

def _final_qa_evidence(result,limit=8):
    active=result.get("active_result") or {}
    rows=list(active.get("evidence_top20") or [])
    out=[]
    for i,x in enumerate(rows[:int(limit)]):
        vid=str(x.get("video_id","")).strip()
        pts=_final_ev_time(x)
        eid=str(
            x.get("evidence_pack_id")
            or x.get("interval_id")
            or x.get("candidate_id")
            or f"e{i}"
        )
        if not vid or pts is None:
            continue
        image_b64=None
        image_mime=None
        try:
            dec=decode_raw_first_at_or_after(
                vid,float(pts),
                request_id="final_qa_"+hashlib.sha256(
                    f"{vid}|{float(pts):.6f}|{i}".encode()
                ).hexdigest()[:18],
                persist_jpeg=False,
            )
            im=dec["_image"]
            buf=io.BytesIO()
            im.save(buf,format="JPEG",quality=86,optimize=True)
            image_b64=base64.b64encode(buf.getvalue()).decode("ascii")
            image_mime="image/jpeg"
            try: im.close()
            except Exception: pass
        except Exception:
            image_b64=None
            image_mime=None
        out.append({
            "evidence_index":i,
            "evidence_id":eid,
            "video_id":vid,
            "actual_pts_time":float(pts),
            "ocr_text_compact":str(x.get("ocr_text_compact",""))[:2500],
            "asr_text_compact":str(x.get("asr_text_compact",""))[:2500],
            "image_b64":image_b64,
            "image_mime":image_mime,
        })
    if not out:
        raise RuntimeError("FINAL_QA_NO_USABLE_EVIDENCE")
    if not any(x["image_b64"] for x in out):
        raise RuntimeError("FINAL_QA_NO_PREVIEW_IMAGES")
    return out



# ---- tier-2 Kaggle Qwen3-4B TEXT runtime ------------------------------------
#
# Temporary competition tier while Qwen3.5-4B is being prepared.
#
# IMPORTANT:
# - Qwen3-4B is TEXT-ONLY, not a vision-language model.
# - It is excellent for Vietnamese/English query adaptation and TRAKE event text.
# - For QA fallback it can only reason over OCR/ASR/text metadata retrieved locally.
# - It must never be described as having seen the evidence images.
#
# Exact Kaggle Model selected by the user:
# https://www.kaggle.com/models/qwen-lm/qwen-3/Transformers/4b/1

QWEN_LOCAL_MODEL_FAMILY="Qwen3-4B"
QWEN_LOCAL_VARIANT="TEXT_ONLY"
QWEN_KAGGLE_MODEL_URL=(
    "https://www.kaggle.com/models/qwen-lm/"
    "qwen-3/Transformers/4b/1"
)
QWEN_KAGGLE_CANONICAL_PATH=Path(
    "/kaggle/input/qwen-3/transformers/4b/1"
)

QWEN_LOCAL_WORKER_PORT=int(
    os.environ.get("AIC2026_QWEN_PORT","8877")
)
QWEN_LOCAL_WORKER_URL=(
    f"http://127.0.0.1:{QWEN_LOCAL_WORKER_PORT}"
)
QWEN_LOCAL_RUNTIME_DIR=Path(
    "/kaggle/working/qwen3_4b_runtime"
)
QWEN_LOCAL_RUNTIME_PKGS=(
    QWEN_LOCAL_RUNTIME_DIR/"pkgs"
)
QWEN_LOCAL_WORKER_SCRIPT=(
    QWEN_LOCAL_RUNTIME_DIR/"worker.py"
)
QWEN_LOCAL_WORKER_LOG=(
    QWEN_LOCAL_RUNTIME_DIR/"worker.log"
)
QWEN_LOCAL_RUNTIME_MARKER=(
    QWEN_LOCAL_RUNTIME_DIR/"runtime.json"
)
QWEN_LOCAL_TRANSFORMERS_VERSION="5.0.0"
QWEN_LOCAL_WORKER_PROC=globals().get(
    "QWEN_LOCAL_WORKER_PROC"
)
QWEN_LOCAL_LOCK=globals().get(
    "QWEN_LOCAL_LOCK",
    threading.RLock(),
)


def _qwen_validate_model_dir(path):
    p=Path(path)
    cfgp=p/"config.json"
    if not cfgp.is_file():
        return False,None

    try:
        cfg=json.loads(
            cfgp.read_text(encoding="utf-8")
        )
    except Exception:
        return False,None

    model_type=str(
        cfg.get("model_type","")
    ).lower()
    arch=" ".join(
        str(x)
        for x in cfg.get("architectures",[])
    ).lower()
    low=str(p).lower()

    # Official Qwen3-4B text model.
    ok=(
        model_type=="qwen3"
        and "base" not in low
        and (
            "qwen3forcausallm" in arch
            or not arch
        )
    )
    if not ok:
        return False,cfg

    has_weights=(
        any(p.glob("*.safetensors"))
        or (p/"model.safetensors.index.json").is_file()
        or (p/"pytorch_model.bin").is_file()
    )
    return bool(has_weights),cfg


def _qwen_resolve_kaggle_model_dir():
    explicit=os.environ.get(
        "AIC2026_QWEN_MODEL_DIR",""
    ).strip()

    candidates=[]
    if explicit:
        candidates.append(Path(explicit))

    root=Path("/kaggle/input")
    candidates.extend([
        QWEN_KAGGLE_CANONICAL_PATH,
        root/"qwen-3"/"Transformers"/"4b"/"1",
        root/"qwen-lm"/"qwen-3"/"transformers"/"4b"/"1",
        root/"models"/"qwen-lm"/"qwen-3"/"transformers"/"4b"/"1",
    ])

    # Robust bounded discovery for Kaggle mount-layout differences.
    if root.exists():
        patterns=[
            "**/qwen-3/**/4b/**/config.json",
            "**/qwen3/**/4b/**/config.json",
            "**/qwen*/**/config.json",
        ]
        for pat in patterns:
            try:
                for cfgp in root.glob(pat):
                    candidates.append(cfgp.parent)
            except Exception:
                pass

    uniq=[]
    seen=set()
    for p in candidates:
        try:
            rp=p.resolve()
        except Exception:
            rp=p
        key=str(rp)
        if key not in seen:
            seen.add(key)
            uniq.append(rp)

    valid=[]
    for p in uniq:
        ok,cfg=_qwen_validate_model_dir(p)
        if not ok:
            continue

        low=str(p).lower()
        score=0
        if "/qwen-3/" in low:
            score+=100
        if "/4b/" in low:
            score+=80
        if "transformers" in low:
            score+=20
        if str(p)==str(QWEN_KAGGLE_CANONICAL_PATH):
            score+=200
        try:
            score+=min(99,int(p.name))
        except Exception:
            pass

        valid.append((score,str(p),p))

    if not valid:
        return None

    valid.sort(
        key=lambda x:(x[0],x[1]),
        reverse=True,
    )
    return valid[0][2]


def _qwen_model_status():
    p=_qwen_resolve_kaggle_model_dir()
    return {
        "found":p is not None,
        "state":"READY_LOCAL" if p else "MISSING_KAGGLE_INPUT",
        "source":"KAGGLE_MODEL_INPUT" if p else None,
        "model_dir":str(p) if p else None,
        "family":QWEN_LOCAL_MODEL_FAMILY,
        "variant":QWEN_LOCAL_VARIANT,
        "kaggle_model_url":QWEN_KAGGLE_MODEL_URL,
        "canonical_path":str(QWEN_KAGGLE_CANONICAL_PATH),
        "worker_url":QWEN_LOCAL_WORKER_URL,
        "main_transformers":
            transformers.__version__,
        "worker_transformers":
            QWEN_LOCAL_TRANSFORMERS_VERSION,
        "strict_dual_gpu":True,
        "vision_capable":False,
        "qa_fallback_scope":"OCR_ASR_TEXT_ONLY",
    }


def _qwen_prepare_isolated_runtime():
    """
    Qwen3 is supported by Transformers >=4.51.0.
    The final notebook already freezes Transformers 5.0.0.

    Run Qwen in a separate subprocess, but DO NOT pip-install another
    Transformers build during the competition.
    """
    if transformers.__version__!="5.0.0":
        raise RuntimeError(
            (
                "QWEN3_EXPECTS_FROZEN_TRANSFORMERS_5_0_0",
                transformers.__version__,
            )
        )

    QWEN_LOCAL_RUNTIME_DIR.mkdir(
        parents=True,
        exist_ok=True,
    )
    atomic_json(
        QWEN_LOCAL_RUNTIME_MARKER,
        {
            "transformers":"5.0.0",
            "source":"MAIN_ENV_IN_SEPARATE_PROCESS",
            "created_at_utc":
                pd.Timestamp.utcnow().isoformat(),
            "family":QWEN_LOCAL_MODEL_FAMILY,
        },
    )


_QWEN_WORKER_SOURCE=r"""
import os, sys, json, threading
from pathlib import Path
from http.server import BaseHTTPRequestHandler, ThreadingHTTPServer

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL_DIR=Path(os.environ["AIC_QWEN_MODEL_DIR"])
PORT=int(os.environ["AIC_QWEN_PORT"])

if not torch.cuda.is_available() or torch.cuda.device_count()!=2:
    raise RuntimeError(
        (
            "QWEN_EXACTLY_TWO_CUDA_GPUS_REQUIRED",
            torch.cuda.device_count()
            if torch.cuda.is_available()
            else 0,
        )
    )

_tokenizer=None
_model=None
_load_error=None
_gen_lock=threading.RLock()


def _norm_device(x):
    if isinstance(x,int):
        return x
    s=str(x).lower()
    if s.startswith("cuda:"):
        try:
            return int(s.split(":",1)[1])
        except Exception:
            return s
    return s


def load_model():
    global _tokenizer,_model,_load_error

    if _model is not None:
        return
    if _load_error is not None:
        raise RuntimeError(_load_error)

    try:
        _tokenizer=AutoTokenizer.from_pretrained(
            str(MODEL_DIR),
            local_files_only=True,
        )

        kwargs=dict(
            local_files_only=True,
            device_map="balanced",
            low_cpu_mem_usage=True,
            # Force the 4B model to genuinely span both T4s while
            # leaving substantial VRAM headroom for retrieval.
            max_memory={
                0:"6GiB",
                1:"6GiB",
            },
        )

        try:
            _model=AutoModelForCausalLM.from_pretrained(
                str(MODEL_DIR),
                dtype=torch.float16,
                **kwargs,
            )
        except TypeError:
            _model=AutoModelForCausalLM.from_pretrained(
                str(MODEL_DIR),
                torch_dtype=torch.float16,
                **kwargs,
            )

        _model.eval()

        dmap=getattr(_model,"hf_device_map",{}) or {}
        used=set()
        bad=[]

        for d in dmap.values():
            nd=_norm_device(d)
            if nd in {0,1}:
                used.add(nd)
            else:
                bad.append(str(d))

        if bad:
            raise RuntimeError(
                (
                    "QWEN_CPU_OR_DISK_OFFLOAD_FORBIDDEN",
                    bad,
                )
            )

        if used!={0,1}:
            raise RuntimeError(
                (
                    "QWEN_MUST_USE_BOTH_GPUS",
                    sorted(used),
                )
            )

    except Exception as e:
        _load_error=(
            f"{type(e).__name__}: {e}"
        )
        raise


def generate(payload):
    load_model()

    system_prompt=str(
        payload.get("system_prompt","")
    ).strip()
    user_text=str(
        payload.get("user_text","")
    )

    max_new_tokens=max(
        32,
        min(
            700,
            int(payload.get(
                "max_new_tokens",400
            )),
        ),
    )

    messages=[]
    if system_prompt:
        messages.append({
            "role":"system",
            "content":system_prompt,
        })
    messages.append({
        "role":"user",
        "content":user_text,
    })

    rendered=_tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
        enable_thinking=False,
    )

    inputs=_tokenizer(
        rendered,
        return_tensors="pt",
    )

    first_device=next(_model.parameters()).device
    inputs={
        k:v.to(first_device)
        for k,v in inputs.items()
    }
    input_len=int(
        inputs["input_ids"].shape[-1]
    )

    with _gen_lock, torch.inference_mode():
        out=_model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            use_cache=True,
        )

    text=_tokenizer.decode(
        out[0][input_len:],
        skip_special_tokens=True,
    ).strip()

    return {
        "ok":True,
        "text":text,
        "model_dir":str(MODEL_DIR),
        "used_gpus":[0,1],
        "vision_capable":False,
    }


class H(BaseHTTPRequestHandler):
    def log_message(self,*args):
        return

    def sendj(self,code,obj):
        raw=json.dumps(
            obj,
            ensure_ascii=False,
        ).encode("utf-8")
        self.send_response(code)
        self.send_header(
            "Content-Type",
            "application/json; charset=utf-8",
        )
        self.send_header(
            "Content-Length",
            str(len(raw)),
        )
        self.end_headers()
        self.wfile.write(raw)

    def do_GET(self):
        if self.path=="/health":
            try:
                load_model()
                return self.sendj(
                    200,
                    {
                        "ok":True,
                        "model_dir":str(MODEL_DIR),
                        "cuda_count":
                            torch.cuda.device_count(),
                        "transformers":
                            __import__(
                                "transformers"
                            ).__version__,
                        "vision_capable":False,
                    },
                )
            except Exception as e:
                return self.sendj(
                    500,
                    {
                        "ok":False,
                        "error":
                            f"{type(e).__name__}: {e}",
                    },
                )

        self.sendj(404,{"ok":False})

    def do_POST(self):
        if self.path!="/generate":
            return self.sendj(
                404,
                {"ok":False},
            )

        try:
            n=int(self.headers.get(
                "Content-Length","0"
            ))
            body=json.loads(
                self.rfile.read(n).decode("utf-8")
            )
            return self.sendj(
                200,
                generate(body),
            )
        except Exception as e:
            return self.sendj(
                500,
                {
                    "ok":False,
                    "error":
                        f"{type(e).__name__}: {e}",
                },
            )


ThreadingHTTPServer(
    ("127.0.0.1",PORT),
    H,
).serve_forever()
"""


def _qwen_write_worker_script():
    QWEN_LOCAL_RUNTIME_DIR.mkdir(
        parents=True,
        exist_ok=True,
    )
    QWEN_LOCAL_WORKER_SCRIPT.write_text(
        _QWEN_WORKER_SOURCE,
        encoding="utf-8",
    )


def _qwen_worker_health(timeout=3):
    try:
        r=requests.get(
            QWEN_LOCAL_WORKER_URL+"/health",
            timeout=timeout,
        )
        if r.status_code==200:
            data=r.json()
            return bool(data.get("ok")),data

        try:
            return False,r.json()
        except Exception:
            return False,{
                "status":r.status_code
            }

    except Exception as e:
        return False,{
            "error":
                f"{type(e).__name__}: {e}"
        }


def _qwen_ensure_worker():
    global QWEN_LOCAL_WORKER_PROC

    with QWEN_LOCAL_LOCK:
        ok,info=_qwen_worker_health(
            timeout=1
        )
        if ok:
            return info

        model_dir=_qwen_resolve_kaggle_model_dir()
        if model_dir is None:
            raise RuntimeError(
                "QWEN3_4B_KAGGLE_INPUT_NOT_FOUND"
            )

        _qwen_prepare_isolated_runtime()
        _qwen_write_worker_script()

        p=QWEN_LOCAL_WORKER_PROC
        if p is not None:
            try:
                if p.poll() is None:
                    p.terminate()
                    p.wait(timeout=5)
            except Exception:
                try:
                    p.kill()
                except Exception:
                    pass

        logf=open(
            QWEN_LOCAL_WORKER_LOG,
            "a",
            encoding="utf-8",
        )

        env=dict(os.environ)
        env.update({
            "AIC_QWEN_MODEL_DIR":
                str(model_dir),
            "AIC_QWEN_PORT":
                str(QWEN_LOCAL_WORKER_PORT),
            "CUDA_VISIBLE_DEVICES":"0,1",
            "TOKENIZERS_PARALLELISM":"false",
        })

        QWEN_LOCAL_WORKER_PROC=subprocess.Popen(
            [
                sys.executable,
                str(QWEN_LOCAL_WORKER_SCRIPT),
            ],
            stdout=logf,
            stderr=subprocess.STDOUT,
            env=env,
        )

        globals()["QWEN_LOCAL_WORKER_PROC"]=(
            QWEN_LOCAL_WORKER_PROC
        )

        deadline=time.time()+240
        last={}

        while time.time()<deadline:
            if QWEN_LOCAL_WORKER_PROC.poll() is not None:
                break

            ok,last=_qwen_worker_health(
                timeout=2
            )
            if ok:
                return last

            time.sleep(2)

        tail=""
        try:
            tail=QWEN_LOCAL_WORKER_LOG.read_text(
                encoding="utf-8",
                errors="ignore",
            )[-6000:]
        except Exception:
            pass

        raise RuntimeError(
            "QWEN_WORKER_START_FAIL: "
            +str(last)
            +" LOG="+tail
        )


def _qwen_local_raw(
    system_prompt,
    user_text,
    evidence=None,
    max_new_tokens=400,
):
    """
    Text-only Qwen3-4B.
    `evidence` images are intentionally ignored.
    QA receives OCR/ASR through user_obj instead.
    """
    _qwen_ensure_worker()

    r=requests.post(
        QWEN_LOCAL_WORKER_URL+"/generate",
        json={
            "system_prompt":
                str(system_prompt),
            "user_text":
                str(user_text),
            "max_new_tokens":
                int(max_new_tokens),
        },
        timeout=360,
    )

    if r.status_code>=400:
        try:
            detail=r.json()
        except Exception:
            detail=r.text[-2000:]
        raise RuntimeError(
            (
                "QWEN_LOCAL_HTTP",
                r.status_code,
                detail,
            )
        )

    data=r.json()
    if not data.get("ok"):
        raise RuntimeError(
            ("QWEN_LOCAL_FAIL",data)
        )

    text=str(
        data.get("text","")
    ).strip()

    if not text:
        raise RuntimeError(
            "QWEN_LOCAL_EMPTY_OUTPUT"
        )

    return text


def _qwen_local_json(
    system_prompt,
    user_obj,
    evidence=None,
):
    user_text=(
        user_obj
        if isinstance(user_obj,str)
        else json.dumps(
            user_obj,
            ensure_ascii=False,
        )
    )

    raw=_qwen_local_raw(
        system_prompt,
        user_text,
        evidence=None,
        max_new_tokens=500,
    )

    return _extract_json_object(raw)


class _QwenLocalCfg:
    provider="qwen_local"
    model="Qwen3-4B@Kaggle"
    kind="qwen3_text_local"
    base_url=QWEN_LOCAL_WORKER_URL
    api_key=None
    timeout_s=360


class QwenLocalClient:
    def __init__(self):
        self.cfg=_QwenLocalCfg()

    def generate_json(
        self,
        system_prompt,
        user_text,
    ):
        return _qwen_local_json(
            system_prompt,
            user_text,
            evidence=None,
        )


def _qwen_local_client():
    return QwenLocalClient()


def _qwen_adapter_output_sane(
    task,
    original_query,
    adapted,
):
    t=str(task).upper()

    bad={
        "","...","…",".","-",
        "n/a","none","null",
    }

    if t=="KIS":
        x=str(
            adapted.get(
                "search_query_en",""
            )
        ).strip()

        words=re.findall(
            r"[A-Za-z0-9]+",
            x,
        )

        if (
            x.lower() in bad
            or not (3<=len(words)<=50)
        ):
            return False,"KIS_LENGTH_OR_DEGENERATE"

        viet_chars=set(
            "ăâđêôơưĂÂĐÊÔƠƯ"
            "áàảãạấầẩẫậắằẳẵặ"
            "éèẻẽẹếềểễệ"
            "íìỉĩịóòỏõọốồổỗộớờởỡợ"
            "úùủũụứừửữựýỳỷỹỵ"
        )

        if any(
            ch in viet_chars
            for ch in x
        ):
            return False,"KIS_NOT_ENGLISH"

        src_nums=set(
            re.findall(
                r"\d+",
                str(original_query),
            )
        )
        out_nums=set(
            re.findall(r"\d+",x)
        )

        if (
            src_nums
            and not src_nums.issubset(
                out_nums
            )
        ):
            return False,"KIS_NUMERIC_ANCHOR_LOSS"

        acronyms={
            z.lower()
            for z in re.findall(
                r"\b[A-Z][A-Z0-9_-]{2,}\b",
                str(original_query),
            )
        }

        if acronyms:
            low=x.lower()
            missing=[
                z
                for z in acronyms
                if z not in low
            ]
            if missing:
                return False,(
                    "KIS_NAMED_ANCHOR_LOSS:"
                    +",".join(missing)
                )

        return True,"PASS"

    if t=="TRAKE":
        arr=adapted.get("events_en")

        if (
            not isinstance(arr,list)
            or len(arr)<2
        ):
            return False,"TRAKE_EVENT_LIST_INVALID"

        for i,x in enumerate(arr):
            z=str(x).strip()
            wc=len(
                re.findall(
                    r"[A-Za-z0-9]+",
                    z,
                )
            )

            if (
                z.lower() in bad
                or not (2<=wc<=50)
            ):
                return False,(
                    f"TRAKE_EVENT_{i}_DEGENERATE"
                )

        return True,"PASS"

    return False,"UNSUPPORTED_TASK"


QWEN_LOCAL_DISCOVERY=_qwen_model_status()


def _final_provider_multimodal_json(client,system_prompt,user_obj,evidence):
    """
    Uses the request-only provider/model/key already parsed by the online1 server.
    No secret is persisted.
    """
    cfg=client.cfg
    user_text=json.dumps(user_obj,ensure_ascii=False)
    imgs=[x for x in evidence if x.get("image_b64")]
    if str(cfg.provider).lower()=="groq":
        imgs=imgs[:3]

    if cfg.kind=="qwen3_text_local":
        return _qwen_local_json(
            system_prompt,
            user_obj,
            evidence=None,
        )

    if cfg.kind=="openai_compatible":
        content=[{"type":"text","text":user_text}]
        for x in imgs:
            content.append({
                "type":"image_url",
                "image_url":{"url":f"data:{x['image_mime']};base64,{x['image_b64']}"},
            })
        payload={
            "model":cfg.model,
            "messages":[
                {"role":"system","content":system_prompt},
                {"role":"user","content":content},
            ],
            "temperature":0,
            "max_tokens":600,
            "response_format":{"type":"json_object"},
        }
        r=requests.post(
            cfg.base_url.rstrip("/")+"/chat/completions",
            headers=client._openai_headers(),
            json=payload,
            timeout=cfg.timeout_s,
        )
        if r.status_code>=400:
            payload.pop("response_format",None)
            r=requests.post(
                cfg.base_url.rstrip("/")+"/chat/completions",
                headers=client._openai_headers(),
                json=payload,
                timeout=cfg.timeout_s,
            )
        if r.status_code>=400:
            raise RuntimeError(
                f"{cfg.provider} HTTP {r.status_code}: {_redact_secret_text(r.text)}"
            )
        body=r.json()
        raw=body["choices"][0]["message"]["content"]
        if isinstance(raw,list):
            raw="\n".join(
                str(z.get("text","")) for z in raw if isinstance(z,dict)
            )
        return _extract_json_object(raw)

    if cfg.kind=="gemini":
        parts=[{"text":user_text}]
        for x in imgs:
            parts.append({
                "inlineData":{
                    "mimeType":x["image_mime"],
                    "data":x["image_b64"],
                }
            })
        payload={
            "systemInstruction":{"parts":[{"text":system_prompt}]},
            "contents":[{"role":"user","parts":parts}],
            "generationConfig":{
                "responseMimeType":"application/json",
                "maxOutputTokens":600,
                "temperature":0,
            },
        }
        r=requests.post(
            cfg.base_url.rstrip("/")+"/models/"+cfg.model+":generateContent",
            headers={
                "Content-Type":"application/json",
                "x-goog-api-key":cfg.api_key,
            },
            json=payload,
            timeout=cfg.timeout_s,
        )
        if r.status_code>=400:
            raise RuntimeError(
                f"gemini HTTP {r.status_code}: {_redact_secret_text(r.text)}"
            )
        data=r.json()
        raw="\n".join(
            str(z.get("text",""))
            for z in data["candidates"][0]["content"]["parts"]
            if isinstance(z,dict)
        )
        return _extract_json_object(raw)

    if cfg.kind=="anthropic":
        content=[{"type":"text","text":user_text}]
        for x in imgs:
            content.append({
                "type":"image",
                "source":{
                    "type":"base64",
                    "media_type":x["image_mime"],
                    "data":x["image_b64"],
                },
            })
        payload={
            "model":cfg.model,
            "max_tokens":600,
            "temperature":0,
            "system":system_prompt,
            "messages":[{"role":"user","content":content}],
        }
        r=requests.post(
            cfg.base_url.rstrip("/")+"/messages",
            headers={
                "Content-Type":"application/json",
                "x-api-key":cfg.api_key,
                "anthropic-version":"2023-06-01",
            },
            json=payload,
            timeout=cfg.timeout_s,
        )
        if r.status_code>=400:
            raise RuntimeError(
                f"anthropic HTTP {r.status_code}: {_redact_secret_text(r.text)}"
            )
        data=r.json()
        raw="\n".join(
            str(z.get("text",""))
            for z in data.get("content",[])
            if isinstance(z,dict) and z.get("type")=="text"
        )
        return _extract_json_object(raw)

    raise RuntimeError(("FINAL_QA_UNSUPPORTED_PROVIDER_KIND",cfg.kind))

def _final_valid_evidence_ids(evidence):
    return {str(x["evidence_id"]) for x in evidence}

def _final_qa_prepare_answer(original_query,retrieval_response,client):
    """
    Phase 1: retrieve a grounded candidate answer once.

    IMPORTANT:
    - no verifier call happens here;
    - no API key is stored in the returned phase context;
    - the evidence pack is kept in server RAM for phase 2.
    """
    evidence_limit=3 if str(client.cfg.provider).lower()=="groq" else 8
    evidence=_final_qa_evidence(
        retrieval_response,
        limit=evidence_limit,
    )

    compact=[
        {
            "evidence_index":x["evidence_index"],
            "evidence_id":x["evidence_id"],
            "video_id":x["video_id"],
            "actual_pts_time":x["actual_pts_time"],
            "ocr_text_compact":x["ocr_text_compact"],
            "asr_text_compact":x["asr_text_compact"],
        }
        for x in evidence
    ]
    allowed=_final_valid_evidence_ids(evidence)

    is_qwen3_text=(
        getattr(client.cfg,"kind","")
        =="qwen3_text_local"
    )

    if is_qwen3_text:
        text_signal=[
            (
                str(x.get("ocr_text_compact","")).strip()
                +" "
                +str(x.get("asr_text_compact","")).strip()
            ).strip()
            for x in compact
        ]
        if not any(text_signal):
            raise RuntimeError(
                "QWEN3_TEXT_QA_NO_OCR_ASR_EVIDENCE"
            )

    answer_prompt=(
        "Answer ONLY from the supplied OCR/ASR/text evidence. "
        "You are TEXT-ONLY and have NOT seen any image. "
        "Return JSON only with keys answer, cited_evidence_ids, abstain. "
        "answer must be <=100 characters. Cite only supplied evidence_id values. "
        "If the text evidence is insufficient, abstain=true."
        if is_qwen3_text
        else
        "Answer ONLY from the supplied video evidence and images. "
        "Return JSON only with keys answer, cited_evidence_ids, abstain. "
        "answer must be <=100 characters. Cite only supplied evidence_id values. "
        "If insufficient, abstain=true."
    )

    answer=_final_provider_multimodal_json(
        client,
        answer_prompt,
        {
            "task_type":"QA",
            "question":original_query,
            "evidence":compact,
            "required_response_schema":{
                "answer":"string <=100 chars",
                "cited_evidence_ids":["supplied evidence_id"],
                "abstain":"boolean",
            },
        },
        evidence,
    )

    ans=str(answer.get("answer","")).strip()
    ac=[
        str(x).strip()
        for x in answer.get("cited_evidence_ids",[])
        if str(x).strip()
    ]

    if bool(answer.get("abstain",False)):
        raise RuntimeError("FINAL_QA_PROVIDER_ABSTAIN")
    if not ans or len(ans)>100:
        raise RuntimeError(("FINAL_QA_ANSWER_LENGTH",len(ans)))
    if not ac or set(ac)-allowed:
        raise RuntimeError(
            ("FINAL_QA_BAD_ANSWER_CITATIONS",ac,sorted(allowed))
        )

    return {
        "original_query":str(original_query),
        "retrieval_response":retrieval_response,
        "evidence":evidence,
        "compact_evidence":compact,
        "allowed_evidence_ids":sorted(allowed),
        "candidate_answer":ans,
        "answer_evidence_ids":ac,
        "answer_provider":client.cfg.provider,
        "answer_model":client.cfg.model,
        "provider_api_key_persisted":False,
    }


def _final_qa_verify_prepared(prepared,client):
    """
    Phase 2: verify an already-created answer.

    This can use a DIFFERENT API key from phase 1. Therefore verifier
    rate-limit/key rotation never recomputes the answer.
    """
    original_query=str(prepared["original_query"])
    evidence=prepared["evidence"]
    compact=prepared["compact_evidence"]
    allowed=set(prepared["allowed_evidence_ids"])
    ans=str(prepared["candidate_answer"])
    ac=[str(x) for x in prepared["answer_evidence_ids"]]

    is_qwen3_text=(
        getattr(client.cfg,"kind","")
        =="qwen3_text_local"
    )
    verifier_prompt=(
        "Verify the candidate answer ONLY against the supplied OCR/ASR/text evidence. "
        "You are TEXT-ONLY and have NOT seen any image. "
        "Return JSON only with keys supported, cited_evidence_ids, abstain, reason. "
        "If the text evidence cannot prove the answer, abstain=true."
        if is_qwen3_text
        else
        "Verify the candidate answer ONLY against the same supplied evidence/images. "
        "Return JSON only with keys supported, cited_evidence_ids, abstain, reason. "
        "Cite only supplied evidence_id values."
    )

    verifier=_final_provider_multimodal_json(
        client,
        verifier_prompt,
        {
            "task_type":"QA_VERIFY",
            "question":original_query,
            "candidate_answer":ans,
            "evidence":compact,
            "required_response_schema":{
                "supported":"boolean",
                "cited_evidence_ids":["supplied evidence_id"],
                "abstain":"boolean",
                "reason":"short string",
            },
        },
        evidence,
    )

    vc=[
        str(x).strip()
        for x in verifier.get("cited_evidence_ids",[])
        if str(x).strip()
    ]

    if (
        bool(verifier.get("abstain",False))
        or verifier.get("supported") is not True
    ):
        raise RuntimeError(
            ("FINAL_QA_VERIFIER_REJECT",verifier.get("reason"))
        )

    if not vc or set(vc)-allowed:
        raise RuntimeError(
            ("FINAL_QA_BAD_VERIFIER_CITATIONS",vc)
        )

    common=[x for x in ac if x in set(vc)]
    if not common:
        raise RuntimeError("FINAL_QA_CITATION_NO_OVERLAP")

    byid={str(x["evidence_id"]):x for x in evidence}
    rep=byid[common[0]]

    return {
        "answer":ans,
        "recommended_result_index":int(rep["evidence_index"]),
        "confidence":None,
        "reason":str(verifier.get("reason",""))[:500],
        "evidence_ids":ac,
        "verifier_evidence_ids":vc,
        "abstain":False,
        "grounded":True,
        "verifier_pass":True,
        "representative_evidence_id":rep["evidence_id"],
        "representative_video_id":rep["video_id"],
        "representative_actual_pts_time":float(
            rep["actual_pts_time"]
        ),
        "answer_provider":prepared.get("answer_provider"),
        "answer_model":prepared.get("answer_model"),
        "verifier_provider":client.cfg.provider,
        "verifier_model":client.cfg.model,
        "provider_api_key_persisted":False,
    }


def _final_qa_agent_verify(original_query,retrieval_response,client):
    """
    Compatibility wrapper for legacy callers.
    New manual UI uses the two phase endpoints below instead.
    """
    prepared=_final_qa_prepare_answer(
        original_query,
        retrieval_response,
        client,
    )
    return _final_qa_verify_prepared(prepared,client)


# ---- exact AIC26 BTC row generation ------------------------------------------
def _final_clean_video(v):
    s=str(v).strip()
    if s.lower().endswith(".mp4"):
        s=s[:-4]
    if not re.fullmatch(r"[A-Za-z0-9_]+",s):
        raise RuntimeError(("BTC_VIDEO_ID",v))
    return s

def _final_int_frame(v):
    x=float(v)
    if not math.isfinite(x) or x<0 or abs(x-round(x))>1e-9:
        raise RuntimeError(("BTC_FRAME_ID",v))
    return int(round(x))

def _final_btc_from_result(task,result,verification=None):
    task=str(task).upper()
    p=result.get("active_result") or {}

    if task=="KIS":
        rows=[]
        seen=set()
        for x in p.get("results",[])[:100]:
            row=[_final_clean_video(x["video_id"]),_final_int_frame(x["frame_idx_candidate"])]
            k=tuple(row)
            if k not in seen:
                seen.add(k); rows.append(row)
        if not rows:
            return {"ready":False,"reason":"KIS_NO_ROWS","rows":[]}
        return {"ready":True,"reason":"PASS","rows":rows}

    if task=="TRAKE":
        try:
            rows,mapped=_final_trake_btc(p)
            return {"ready":True,"reason":"PASS","rows":rows,"mapped_events":mapped}
        except Exception as e:
            return {"ready":False,"reason":f"TRAKE_NOT_FINAL: {e}","rows":[]}

    if task=="QA":
        v=verification or {}
        if v.get("verifier_pass") is not True or v.get("grounded") is not True:
            return {"ready":False,"reason":"QA_EXTERNAL_GROUNDED_VERIFIER_REQUIRED","rows":[]}
        ans=str(v.get("answer","")).strip()
        if not ans or len(ans)>100:
            return {"ready":False,"reason":"QA_ANSWER_INVALID","rows":[]}
        try:
            m=_final_same_video_nearest_frame(
                v["representative_video_id"],
                float(v["representative_actual_pts_time"]),
            )
            row=[_final_clean_video(m["video_id"]),_final_int_frame(m["frame_idx"]),ans]
            return {"ready":True,"reason":"PASS","rows":[row],"mapped_evidence":m}
        except Exception as e:
            return {"ready":False,"reason":f"QA_FRAME_MAP_FAIL: {e}","rows":[]}

    return {"ready":False,"reason":"UNKNOWN_TASK","rows":[]}

def _final_btc_csv_text(rows):
    sio=io.StringIO(newline="")
    csv.writer(sio,delimiter=",",lineterminator="\n",quoting=csv.QUOTE_MINIMAL).writerows(rows)
    return sio.getvalue()

def _final_validate_query_filename(name,task):
    base=Path(str(name).strip()).name
    if base.lower().endswith(".txt"):
        base=base[:-4]
    elif base.lower().endswith(".csv"):
        base=base[:-4]
    if not base:
        raise ValueError("query filename empty")
    suffix={"KIS":"-kis","QA":"-qa","TRAKE":"-trake"}[str(task).upper()]
    if not base.lower().endswith(suffix):
        raise ValueError(f"query filename must end with {suffix}")
    if not re.fullmatch(r"[A-Za-z0-9._-]+",base):
        raise ValueError("unsafe query filename")
    return base+".csv"

def _final_validate_btc_rows(task,rows):
    task=str(task).upper()
    if not isinstance(rows,list) or not (1<=len(rows)<=100):
        raise ValueError("BTC rows must contain 1..100 rows")
    out=[]
    for row in rows:
        if not isinstance(row,list):
            raise ValueError("BTC row must be list")
        if task=="KIS":
            if len(row)!=2: raise ValueError("KIS BTC row must have 2 columns")
            out.append([_final_clean_video(row[0]),_final_int_frame(row[1])])
        elif task=="QA":
            if len(row)!=3: raise ValueError("QA BTC row must have 3 columns")
            ans=str(row[2])
            if not ans or len(ans)>100: raise ValueError("QA answer empty or >100 chars")
            out.append([_final_clean_video(row[0]),_final_int_frame(row[1]),ans])
        elif task=="TRAKE":
            if len(row)<3: raise ValueError("TRAKE needs video + >=2 frames")
            frames=[_final_int_frame(x) for x in row[1:]]
            if not all(frames[i]<frames[i+1] for i in range(len(frames)-1)):
                raise ValueError("TRAKE frame IDs must be strictly increasing")
            out.append([_final_clean_video(row[0]),*frames])
        else:
            raise ValueError("unknown task")
    return out

def _final_save_btc(query_file,task,rows):
    rows=_final_validate_btc_rows(task,rows)
    filename=_final_validate_query_filename(query_file,task)
    path=BTC_SUBMISSION_ROOT/filename
    with path.open("w",encoding="utf-8",newline="") as f:
        csv.writer(f,delimiter=",",lineterminator="\n",quoting=csv.QUOTE_MINIMAL).writerows(rows)
    # Reparse strict no-header structure.
    with path.open("r",encoding="utf-8",newline="") as f:
        parsed=list(csv.reader(f))
    _final_validate_btc_rows(task,parsed)
    return {
        "file":str(path),
        "filename":filename,
        "rows":len(rows),
        "sha256":sha256_file(path),
    }

def _final_build_submission_zip():
    files=sorted(BTC_SUBMISSION_ROOT.glob("*.csv"),key=lambda p:p.name)
    if not files:
        raise RuntimeError("NO_BTC_CSV_FILES_SAVED")
    if BTC_ZIP_PATH.exists(): BTC_ZIP_PATH.unlink()
    with zipfile.ZipFile(BTC_ZIP_PATH,"w",zipfile.ZIP_DEFLATED) as z:
        for p in files:
            z.write(p,arcname=f"submission/{p.name}")
    with zipfile.ZipFile(BTC_ZIP_PATH) as z:
        bad=z.testzip()
        if bad is not None: raise RuntimeError(("BTC_ZIP_CRC",bad))
        names=z.namelist()
        if any(not n.startswith("submission/") for n in names):
            raise RuntimeError(("BTC_ZIP_ROOT",names))
    return {
        "zip_path":str(BTC_ZIP_PATH),
        "sha256":sha256_file(BTC_ZIP_PATH),
        "file_count":len(files),
        "files":[p.name for p in files],
    }

FINAL_COMPETITION_CONTRACT={
    "build_version":BUILD_VERSION,
    "source_online1_build":LEGACY_ONLINE_BUILD_VERSION,
    "frozen_backends":FINAL_FROZEN_CONTRACTS,
    "official_submission_status":OFFICIAL_SUBMISSION_STATUS,
    "btc":{
        "encoding":"UTF-8",
        "delimiter":",",
        "header":False,
        "max_rows_per_query":100,
        "KIS":"<video_id>,<frame_id>",
        "QA":"<video_id>,<frame_id>,<answer>",
        "TRAKE":"<video_id>,<frame_1>,...,<frame_N>",
        "zip_root":"submission/",
    },
    "strict_dual_gpu":True,
    "provider_model_runtime_selected":True,
    "api_key_persisted":False,
    "groq_qa_max_images_per_request":3,
    "emergency_local_fallback":"STAGE03_SIGLIP2_FOR_KIS",
    "hybrid_fallback_order":[
        "REMOTE_API_MAGAZINE",
        "KAGGLE_QWEN3_4B_TEXT",
        "FROZEN_LOCAL_BACKEND",
    ],
    "qwen_local_model_discovery":QWEN_LOCAL_DISCOVERY,
    "qwen_hf_repo_id":QWEN_HF_REPO_ID,
    "qwen_hf_revision":QWEN_HF_REVISION,
    "qwen3_kaggle_input_only":True,
    "qwen_hf_preflight":_qwen_hf_preflight(),
    "qwen_process_isolation":True,
    "qwen_uses_existing_transformers_5_0_0":True,
}
atomic_json(
    Path("/kaggle/working/final_competition_contract.json"),
    FINAL_COMPETITION_CONTRACT,
)

print("[PASS] FINAL frozen backends + AIC26 wire adapter",{
    "build":BUILD_VERSION,
    "KIS":"CLIP_VIT_B32_ORGANIZER_SHADOW",
    "TRAKE":"RAW_VIDEO_ACTUAL_PTS",
    "QA":"RUNTIME_PROVIDER_MODEL",
    "btc_status":OFFICIAL_SUBMISSION_STATUS,
    "clip_model_source":FINAL_CLIP_MODEL_SOURCE,
    "clip_gpu_shards":FINAL_CLIP_SHARDS,
})
````

````python

# ── SERVER / FINAL CELL: ONLINE O2 PUBLIC HTTP API ─────────────────────────────
# Mục tiêu: expose Online O2 to the external GitHub frontend; no notebook UI.
# INPUT: verified in-memory Online O2 runtime + optional per-request agent credentials.
# OUTPUT: public HTTPS URL + API contract files in /kaggle/working. No server token.
# Pre-check: this executes only after technical O2 harness/readiness publication.
# Post-check: provider API keys never enter result store, artifacts, logs, or responses.
# History v0.5.18: persist one redacted JSON exchange per query; direct tester no manual event JSON.
# ──────────────────────────────────────────────────────────────────────────────

from dataclasses import dataclass
from http.server import BaseHTTPRequestHandler, ThreadingHTTPServer
from urllib.parse import urlparse
import io, queue, secrets, shutil, subprocess, urllib.request, signal

import requests

# Capture previous in-kernel runtime objects before this cell redefines server state.
# This is what makes rerunning the server cell idempotent.
_PREVIOUS_SERVER_HTTPD=globals().get("SERVER_HTTPD")
_PREVIOUS_SERVER_THREAD=globals().get("SERVER_THREAD")
_PREVIOUS_SERVER_PUBLIC_URL=globals().get("SERVER_PUBLIC_URL")
_PREVIOUS_SERVER_TUNNEL_PROCESS=globals().get("SERVER_TUNNEL_PROCESS")

SERVER_HOST="127.0.0.1"
SERVER_REQUESTED_PORT=int(os.environ.get("AIC2026_SERVER_PORT","8765"))
SERVER_PORT=SERVER_REQUESTED_PORT
SERVER_PORT_SCAN_LIMIT=int(os.environ.get("AIC2026_SERVER_PORT_SCAN_LIMIT","20"))
SERVER_REQUIRE_PUBLIC=os.environ.get("AIC2026_REQUIRE_PUBLIC_API","0")!="0"
SERVER_CORS_ORIGIN=os.environ.get("AIC2026_CORS_ORIGIN","*").strip() or "*"
SERVER_URL_PATH=Path("/kaggle/working/url.txt")
SERVER_CONTRACT_PATH=Path("/kaggle/working/server_contract.json")
SERVER_STATUS_PATH=Path("/kaggle/working/server_status.json")
SERVER_STATE_PATH=Path("/kaggle/working/server_state.json")
QUERY_JSON_DIR=Path("/kaggle/working/aic2026_query_json")
QUERY_JSON_DIR.mkdir(parents=True,exist_ok=True)
QUERY_JSON_LOCK=threading.Lock()
SERVER_TUNNEL_PID_PATH=Path("/kaggle/working/aic2026_cloudflared.pid")
SERVER_SINGLETON_POLICY="REUSE_HEALTHY_ELSE_REPLACE_ONE_TRACKED_TUNNEL"
SERVER_AUTH_MODE="NONE"
SERVER_PUBLIC_HEALTH_RETRIES=int(os.environ.get("AIC2026_PUBLIC_HEALTH_RETRIES","12"))
SERVER_PUBLIC_HEALTH_DELAY_S=float(os.environ.get("AIC2026_PUBLIC_HEALTH_DELAY_S","1.0"))
SERVER_EXEC_LOCK=threading.Lock()
SERVER_HTTPD=_PREVIOUS_SERVER_HTTPD
SERVER_THREAD=_PREVIOUS_SERVER_THREAD
SERVER_TUNNEL_PROCESS=_PREVIOUS_SERVER_TUNNEL_PROCESS
SERVER_PUBLIC_URL=_PREVIOUS_SERVER_PUBLIC_URL
SERVER_LOCAL_REUSE_STATUS="UNINITIALIZED"

# Provider API keys are request-memory-only. There is no server-side agent config registry.

# ---- request-local agent provider adapters ------------------------------------

class AgentAPIError(RuntimeError):
    pass

AGENT_PROVIDER_PRESETS={
    "local":{"kind":"local","base_url":""},
    "gemini":{"kind":"gemini","base_url":"https://generativelanguage.googleapis.com/v1beta"},
    "anthropic":{"kind":"anthropic","base_url":"https://api.anthropic.com/v1"},
    "groq":{"kind":"openai_compatible","base_url":"https://api.groq.com/openai/v1"},
    "openai":{"kind":"openai_compatible","base_url":"https://api.openai.com/v1"},
    "openrouter":{"kind":"openai_compatible","base_url":"https://openrouter.ai/api/v1"},
    "mistral":{"kind":"openai_compatible","base_url":"https://api.mistral.ai/v1"},
    "deepseek":{"kind":"openai_compatible","base_url":"https://api.deepseek.com/v1"},
    "xai":{"kind":"openai_compatible","base_url":"https://api.x.ai/v1"},
    "together":{"kind":"openai_compatible","base_url":"https://api.together.xyz/v1"},
    "fireworks":{"kind":"openai_compatible","base_url":"https://api.fireworks.ai/inference/v1"},
    "cerebras":{"kind":"openai_compatible","base_url":"https://api.cerebras.ai/v1"},
    "custom_openai_compatible":{"kind":"openai_compatible","base_url":""},
}

def _redact_secret_text(text):
    z=str(text or "")
    z=re.sub(r"sk-[A-Za-z0-9._-]{6,}","[REDACTED]",z)
    z=re.sub(r"AIza[A-Za-z0-9_-]{8,}","[REDACTED]",z)
    z=re.sub(r"(?i)(api[_ -]?key|authorization|bearer)\s*[:=]\s*\S+",
             r"\1=[REDACTED]",z)
    return z[:1500]

def _extract_json_object(text):
    raw=str(text or "").strip()
    raw=re.sub(r"^```(?:json)?\s*","",raw,flags=re.I)
    raw=re.sub(r"\s*```$","",raw)
    decoder=json.JSONDecoder()
    for i,ch in enumerate(raw):
        if ch!="{":
            continue
        try:
            value,_=decoder.raw_decode(raw[i:])
            if isinstance(value,dict):
                return value
        except json.JSONDecodeError:
            pass
    raise AgentAPIError("model did not return a JSON object")

@dataclass
class AgentRequestConfig:
    provider:str
    kind:str
    api_key:str
    model:str
    base_url:str
    timeout_s:int=90

class RequestAgentClient:
    def __init__(self,cfg):
        self.cfg=cfg

    def _openai_headers(self):
        h={"Content-Type":"application/json"}
        if self.cfg.api_key:
            h["Authorization"]="Bearer "+self.cfg.api_key
        if self.cfg.provider=="openrouter":
            h["X-Title"]="AIC2026 Online O2"
        return h

    def generate_json(self,system_prompt,user_prompt):
        if self.cfg.kind=="openai_compatible":
            if not self.cfg.base_url or not self.cfg.model:
                raise AgentAPIError("base_url/model required")
            endpoint=self.cfg.base_url.rstrip("/")+"/chat/completions"
            payload={
                "model":self.cfg.model,
                "messages":[
                    {"role":"system","content":system_prompt},
                    {"role":"user","content":user_prompt},
                ],
                "temperature":0,
                "max_tokens":1200,
                "response_format":{"type":"json_object"},
            }
            r=requests.post(
                endpoint,headers=self._openai_headers(),json=payload,
                timeout=self.cfg.timeout_s,
            )
            if r.status_code>=400:
                payload.pop("response_format",None)
                r=requests.post(
                    endpoint,headers=self._openai_headers(),json=payload,
                    timeout=self.cfg.timeout_s,
                )
            if r.status_code>=400:
                raise AgentAPIError(
                    f"{self.cfg.provider} HTTP {r.status_code}: {_redact_secret_text(r.text)}"
                )
            data=r.json()
            try:
                content=data["choices"][0]["message"]["content"]
            except Exception as e:
                raise AgentAPIError("unexpected chat-completions response") from e
            if isinstance(content,list):
                content="\n".join(
                    str(z.get("text",""))
                    for z in content if isinstance(z,dict)
                )
            return _extract_json_object(content)

        if self.cfg.kind=="gemini":
            if not self.cfg.api_key or not self.cfg.model:
                raise AgentAPIError("Gemini api_key/model required")
            endpoint=self.cfg.base_url.rstrip("/")+"/models/"+self.cfg.model+":generateContent"
            payload={
                "systemInstruction":{"parts":[{"text":system_prompt}]},
                "contents":[{"role":"user","parts":[{"text":user_prompt}]}],
                "generationConfig":{
                    "responseMimeType":"application/json",
                    "maxOutputTokens":1200,
                    "temperature":0,
                },
            }
            r=requests.post(
                endpoint,
                headers={
                    "Content-Type":"application/json",
                    "x-goog-api-key":self.cfg.api_key,
                },
                json=payload,
                timeout=self.cfg.timeout_s,
            )
            if r.status_code>=400:
                raise AgentAPIError(
                    f"gemini HTTP {r.status_code}: {_redact_secret_text(r.text)}"
                )
            data=r.json()
            try:
                parts=data["candidates"][0]["content"]["parts"]
                content="\n".join(
                    str(z.get("text","")) for z in parts if isinstance(z,dict)
                )
            except Exception as e:
                raise AgentAPIError("unexpected Gemini response") from e
            return _extract_json_object(content)

        if self.cfg.kind=="anthropic":
            if not self.cfg.api_key or not self.cfg.model:
                raise AgentAPIError("Anthropic api_key/model required")
            endpoint=self.cfg.base_url.rstrip("/")+"/messages"
            payload={
                "model":self.cfg.model,
                "max_tokens":1200,
                "temperature":0,
                "system":system_prompt,
                "messages":[{"role":"user","content":user_prompt}],
            }
            r=requests.post(
                endpoint,
                headers={
                    "Content-Type":"application/json",
                    "x-api-key":self.cfg.api_key,
                    "anthropic-version":"2023-06-01",
                },
                json=payload,
                timeout=self.cfg.timeout_s,
            )
            if r.status_code>=400:
                raise AgentAPIError(
                    f"anthropic HTTP {r.status_code}: {_redact_secret_text(r.text)}"
                )
            data=r.json()
            content="\n".join(
                str(z.get("text",""))
                for z in data.get("content",[])
                if isinstance(z,dict) and z.get("type")=="text"
            )
            return _extract_json_object(content)

        raise AgentAPIError("unsupported provider kind")

    def list_models(self):
        if self.cfg.kind=="openai_compatible":
            if not self.cfg.base_url:
                raise AgentAPIError("base_url required")
            r=requests.get(
                self.cfg.base_url.rstrip("/")+"/models",
                headers=self._openai_headers(),
                timeout=min(self.cfg.timeout_s,30),
            )
            if r.status_code>=400:
                raise AgentAPIError(
                    f"models HTTP {r.status_code}: {_redact_secret_text(r.text)}"
                )
            data=r.json()
            rows=data.get("data",data.get("models",[]))
            out=[]
            for row in rows:
                if isinstance(row,dict):
                    mid=row.get("id") or row.get("name")
                    if mid:
                        out.append(str(mid).removeprefix("models/"))
                elif row:
                    out.append(str(row))
            return sorted(set(out))

        if self.cfg.kind=="gemini":
            if not self.cfg.api_key:
                raise AgentAPIError("Gemini api_key required")
            r=requests.get(
                self.cfg.base_url.rstrip("/")+"/models",
                headers={"x-goog-api-key":self.cfg.api_key},
                timeout=min(self.cfg.timeout_s,30),
            )
            if r.status_code>=400:
                raise AgentAPIError(
                    f"Gemini models HTTP {r.status_code}: {_redact_secret_text(r.text)}"
                )
            out=[]
            for row in r.json().get("models",[]):
                if "generateContent" in row.get("supportedGenerationMethods",[]):
                    mid=row.get("baseModelId") or row.get("name")
                    if mid:
                        out.append(str(mid).removeprefix("models/"))
            return sorted(set(out))

        if self.cfg.kind=="anthropic":
            if not self.cfg.api_key:
                raise AgentAPIError("Anthropic api_key required")
            r=requests.get(
                self.cfg.base_url.rstrip("/")+"/models",
                headers={
                    "x-api-key":self.cfg.api_key,
                    "anthropic-version":"2023-06-01",
                },
                timeout=min(self.cfg.timeout_s,30),
            )
            if r.status_code>=400:
                raise AgentAPIError(
                    f"Anthropic models HTTP {r.status_code}: {_redact_secret_text(r.text)}"
                )
            out=[]
            for row in r.json().get("data",[]):
                if isinstance(row,dict):
                    mid=row.get("id")
                    if mid:
                        out.append(str(mid))
            return sorted(set(out))
        return []

def _validate_agent_mode(value):
    mode=str(value or "local").strip().lower()
    if mode not in {"local","llm_plan","llm_plan_verify"}:
        raise ValueError("agent_mode must be local, llm_plan or llm_plan_verify")
    return mode

def _agent_config_from_fields(provider,api_key,model,base_url,*,require_model=True):
    provider=str(provider or "local").strip().lower()
    if provider not in AGENT_PROVIDER_PRESETS:
        raise ValueError(("unknown provider",provider))
    preset=AGENT_PROVIDER_PRESETS[provider]
    resolved_base_url=str(base_url or preset["base_url"]).strip()
    resolved_model=str(model or "").strip()
    resolved_key=str(api_key or "")

    if preset["kind"]!="local":
        if require_model and not resolved_model:
            raise ValueError("model is required for non-local provider")
        if not resolved_base_url:
            raise ValueError("base_url is required for non-local provider")
        if not resolved_key and provider!="custom_openai_compatible":
            raise ValueError("api_key is required for this provider")

    return AgentRequestConfig(
        provider=provider,
        kind=preset["kind"],
        api_key=resolved_key,
        model=resolved_model,
        base_url=resolved_base_url,
    )

def _request_agent_config(body, *, require_model=True):
    """
    Build provider config from this request only.
    No API key/model/provider config is persisted server-side.
    """
    body=body or {}
    return _agent_config_from_fields(
        body.get("provider","local"),
        body.get("api_key",""),
        body.get("model",""),
        body.get("base_url",""),
        require_model=require_model,
    )

AGENT_PLAN_PROMPT="""
You are the query planner for an AIC video retrieval system.
The organizer task_type is authoritative; NEVER change it.
Return JSON only.

KIS or QA:
{"search_query":"string","events":null,"notes":[]}

TRAKE:
{
 "search_query":"string",
 "events":[
   {"text":"event 1","optional":false,"relation_from_prev":"STRICT_BEFORE"},
   {"text":"event 2","optional":false,"relation_from_prev":"STRICT_BEFORE"}
 ],
 "notes":[]
}

Allowed relations:
STRICT_BEFORE, BEFORE_OR_OVERLAP, REPEAT_ALLOWED.

Do not invent objects, text, brands, colors, actions or events absent from the user query.
An unchanged search_query is valid.
"""

AGENT_VERIFY_PROMPT="""
You verify evidence returned by the AIC Online O2 retrieval system.
Return JSON only:
{
 "answer":"string or null",
 "recommended_result_index":"integer or null",
 "confidence":"number 0..1",
 "reason":"short string",
 "evidence_ids":["only IDs/timestamps actually present"],
 "abstain":"boolean"
}
Never invent video IDs, timestamps, visible text, objects or frames.
For QA, answer only when evidence supports it. Otherwise abstain=true.
For KIS/TRAKE, recommendation is advisory only and cannot create a new candidate.
"""

def _normalize_agent_plan(task,query,raw):
    search_query=str(raw.get("search_query") or query).strip() or query
    notes=raw.get("notes",[])
    notes=[str(z) for z in notes[:8]] if isinstance(notes,list) else []
    events=None
    if task=="TRAKE":
        candidate=raw.get("events")
        if isinstance(candidate,list):
            clean=[]
            allowed={"STRICT_BEFORE","BEFORE_OR_OVERLAP","REPEAT_ALLOWED"}
            for i,z in enumerate(candidate):
                if not isinstance(z,dict):
                    continue
                text=str(z.get("text","")).strip()
                if not text:
                    continue
                rel=str(z.get("relation_from_prev","STRICT_BEFORE")).upper()
                if i==0:
                    rel="STRICT_BEFORE"
                if rel not in allowed:
                    rel="STRICT_BEFORE"
                clean.append({
                    "text":text,
                    "optional":bool(z.get("optional",False)),
                    "relation_from_prev":rel,
                })
            if len(clean)>=2:
                events=clean
    return {"search_query":search_query,"events":events,"notes":notes}

def _agent_plan(task,query,client):
    raw=client.generate_json(
        AGENT_PLAN_PROMPT,
        json.dumps({"task_type":task,"query":query},ensure_ascii=False),
    )
    return _normalize_agent_plan(task,query,raw)

def _agent_verify(task,original_query,retrieval_response,client):
    compact={
        "task_type":task,
        "query":original_query,
        "active_result":retrieval_response.get("active_result"),
        "alternatives":retrieval_response.get("alternatives",[])[:2],
    }
    text=json.dumps(_jsonable(compact),ensure_ascii=False)
    if len(text)>70000:
        text=text[:70000]
    return client.generate_json(AGENT_VERIFY_PROMPT,text)

# ---- server serialization/safety ----------------------------------------------

def _jsonable(v):
    if v is None or isinstance(v,(str,int,bool)):
        return v
    if isinstance(v,float):
        return v if math.isfinite(v) else None
    if isinstance(v,np.integer):
        return int(v)
    if isinstance(v,np.floating):
        z=float(v)
        return z if math.isfinite(z) else None
    if isinstance(v,np.ndarray):
        return [_jsonable(z) for z in v.tolist()]
    if isinstance(v,dict):
        return {str(k):_jsonable(val) for k,val in v.items()}
    if isinstance(v,(list,tuple,set)):
        return [_jsonable(z) for z in v]
    if isinstance(v,Path):
        return str(v)
    if hasattr(v,"item"):
        try:
            return _jsonable(v.item())
        except Exception:
            pass
    return str(v)

def _redact_query_payload(value):
    """Recursively redact provider credentials before writing query JSON artifacts."""
    if isinstance(value,dict):
        out={}
        for k,v in value.items():
            key=str(k)
            kl=key.lower()
            if kl in {"api_key","authorization","bearer","server_token"} or kl.endswith("_api_key"):
                out[key]="[REDACTED]" if v else ""
            else:
                out[key]=_redact_query_payload(v)
        return out
    if isinstance(value,list):
        return [_redact_query_payload(v) for v in value]
    if isinstance(value,tuple):
        return [_redact_query_payload(v) for v in value]
    return _jsonable(value)

def _query_id_from_response(response_payload):
    active=response_payload.get("active_result") if isinstance(response_payload,dict) else None
    if isinstance(active,dict):
        qid=active.get("query_id")
        if qid:
            return str(qid)
    qid=response_payload.get("query_id") if isinstance(response_payload,dict) else None
    return str(qid) if qid else "query"

def _safe_filename_token(value,max_len=80):
    z=re.sub(r"[^A-Za-z0-9._-]+","_",str(value or ""))
    z=z.strip("._-") or "query"
    return z[:max_len]

def _persist_query_exchange(endpoint,request_body,response_payload):
    """
    One completed /search or /agent/search request -> one JSON file in Kaggle working.
    Provider API keys are redacted. The directory is runtime-local and intentionally
    outside notebook-visible UI output.
    """
    task=_safe_filename_token((request_body or {}).get("task_type","TASK"),20).upper()
    qid=_safe_filename_token(_query_id_from_response(response_payload),100)
    stamp=str(time.time_ns())
    path=QUERY_JSON_DIR/f"{stamp}_{task}_{qid}.json"
    record={
        "created_at_utc":utc_now(),
        "endpoint":str(endpoint),
        "build_version":BUILD_VERSION,
        "request":_redact_query_payload(request_body or {}),
        "response":_redact_query_payload(response_payload or {}),
    }
    with QUERY_JSON_LOCK:
        atomic_json(path,record)
        atomic_json(QUERY_JSON_DIR/"latest.json",record)
    return str(path)

def _server_contract():
    return {
        "service":"AIC2026 FINAL Competition Server",
        "build_version":BUILD_VERSION,
        "frontend":"external GitHub application",
        "cors_origin":SERVER_CORS_ORIGIN,
        "auth_mode":"NONE",
        "server_token":"REMOVED",
        "query_json_dir":str(QUERY_JSON_DIR),
        "provider_api_key":{
            "transport":"request body on /models, /agent/test, /agent/search",
            "persistence":"NONE_REQUEST_MEMORY_ONLY",
            "returned":False,
        },
        "endpoints":{
            "GET /":{"auth":False,"purpose":"browser-friendly server status; returns HTTP 200"},
            "GET /health":{"auth":False},
            "GET /schema":{"auth":False},
            "GET /providers":{"auth":False},
            "POST /search":{
                "auth":False,
                "body":{"task_type":"KIS|TRAKE|QA","query":"string","events":"optional TRAKE events"},
            },
            "POST /agent/search":{
                "auth":False,
                "body":{
                    "task_type":"KIS|TRAKE|QA",
                    "query":"string",
                    "agent_mode":"local|llm_plan|llm_plan_verify",
                    "provider":"provider id",
                    "api_key":"request-memory-only",
                    "model":"model id",
                    "base_url":"optional override",
                },
            },
            "POST /models":{
                "auth":False,
                "body":{
                    "provider":"provider id",
                    "api_key":"request-memory-only",
                    "base_url":"optional",
                    "model":"NOT_REQUIRED_FOR_LISTING",
                },
            },
            "POST /agent/test":{
                "auth":False,
                "body":{
                    "provider":"provider id",
                    "api_key":"request-memory-only",
                    "model":"model id",
                    "base_url":"optional",
                },
            },
            "POST /preview":{
                "auth":False,
                "body":{"video_id":"string","pts_time":"seconds"},
                "response":"image/jpeg",
            },
        },
        "provider_ids":sorted(AGENT_PROVIDER_PRESETS),
        "provider_api_key_persisted":False,
        "server_auth_mode":"NONE",
        "normalization_status":"NORMALIZED_ONLY",
        "probability_calibration_status":"NOT_CALIBRATED",
    }


# ---- FINAL competition contract overlay ---------------------------------------
_original_server_contract=_server_contract
def _server_contract():
    c=_original_server_contract()
    c["build_version"]=BUILD_VERSION
    c["service"]="AIC2026 FINAL Competition Server"
    c["final_frozen_backends"]=FINAL_FROZEN_CONTRACTS
    c["official_submission_status"]=OFFICIAL_SUBMISSION_STATUS
    c["btc_wire_contract"]=FINAL_COMPETITION_CONTRACT["btc"]
    c["endpoints"]["POST /btc/save"]={
        "auth":False,
        "body":{"query_file":"query-X-kis.txt","task_type":"KIS|QA|TRAKE","rows":"btc rows"},
        "purpose":"save exact no-header UTF-8 CSV into /kaggle/working/submission",
    }
    c["endpoints"]["POST /btc/build"]={
        "auth":False,
        "body":{},
        "purpose":"build /kaggle/working/submission.zip with required submission/ root",
    }
    return c

_SERVER_REQUEST_SEQ=0
_SERVER_REQUEST_SEQ_LOCK=threading.Lock()

def _fresh_execution_query_id(task,query):
    """
    RESULT_STORE is intentionally append-only. The historical implementation used
    a deterministic query hash as query_id, so retrying the same BTC query in the
    same kernel caused RESULT_STORE.put_initial() to raise KeyError.

    Every HTTP execution now gets a unique query_id while retaining a stable
    content hash in the suffix for traceability.
    """
    global _SERVER_REQUEST_SEQ
    stable=hashlib.sha256(
        f"{str(task).upper()}|{canonical_query_text(str(query))}".encode("utf-8")
    ).hexdigest()[:12]
    with _SERVER_REQUEST_SEQ_LOCK:
        _SERVER_REQUEST_SEQ+=1
        seq=_SERVER_REQUEST_SEQ
    return (
        f"{str(task).lower()}_{stable}_"
        f"{int(time.time_ns())}_{int(seq):06d}"
    )

def _execute_retrieval(task,query,events=None):
    execution_qid=_fresh_execution_query_id(task,query)
    req={
        "query_id":execution_qid,
        "task_type":task,
        "query_text":query,
    }
    if task=="TRAKE" and events is not None:
        req["events"]=events
    st,event=execute_session_request(
        req,
        auto_refine_trake=True,
        qa_provider_name=None,
        qa_api_key=None,
    )
    p=st["active_result"]["payload"]
    return {
        "active_result":p,
        "online_control_trace":ONLINE_RUNTIME_TRACE.get(str(p.get("query_id",""))),
        "alternatives":st.get("alternatives",[]),
        "execution_query_id":execution_qid,
        "stable_query_fingerprint":hashlib.sha256(
            f"{str(task).upper()}|{canonical_query_text(str(query))}".encode("utf-8")
        ).hexdigest(),
    }


RETRIEVAL_ADAPT_PROMPT = """
You are a strict query adapter for a video-retrieval system.

Do NOT answer the user's request.
Convert the retrieval description into concise ENGLISH for an OpenAI CLIP text encoder.

Rules:
- Preserve every visually discriminative fact from the input.
- Preserve people, animals, objects, actions, colors, counts/ranges, place context,
  on-screen text, logos, relationships, and temporal order.
- Never invent facts.
- Remove filler, news-writing style, and repetition.
- Keep each output search string <= 40 English words.
- JSON only.

KIS:
{"search_query_en":"..."}

TRAKE:
{"events_en":["event 1","event 2", "..."]}

For TRAKE, output exactly the same number of events, in exactly the same order.
"""

def _retrieval_adapter(task,query,events,client):
    t=str(task).upper()

    if t=="KIS":
        raw=client.generate_json(
            RETRIEVAL_ADAPT_PROMPT,
            json.dumps(
                {"task_type":"KIS","query":str(query)},
                ensure_ascii=False,
            ),
        )
        out=str(raw.get("search_query_en","")).strip()
        bad_literals={"","...","…",".","-","n/a","none","null"}
        word_count=len(re.findall(r"[A-Za-z0-9]+",out))
        if out.lower() in bad_literals or word_count<3:
            raise RuntimeError(
                ("ADAPTER_DEGENERATE_KIS",out,word_count)
            )
        return {
            "task_type":"KIS",
            "search_query_en":out,
            "events_en":None,
        }

    if t=="TRAKE":
        src=[]
        for z in (events or []):
            if isinstance(z,dict):
                src.append(str(z.get("text","")).strip())
            else:
                src.append(str(z).strip())
        src=[x for x in src if x]
        if len(src)<2:
            raise RuntimeError("ADAPTER_TRAKE_NEEDS_EXPLICIT_EVENTS")

        raw=client.generate_json(
            RETRIEVAL_ADAPT_PROMPT,
            json.dumps(
                {
                    "task_type":"TRAKE",
                    "query":str(query),
                    "events":src,
                },
                ensure_ascii=False,
            ),
        )
        got=raw.get("events_en")
        if not isinstance(got,list):
            raise RuntimeError("ADAPTER_TRAKE_EVENTS_NOT_LIST")
        got=[str(x).strip() for x in got]
        bad_literals={"","...","…",".","-","n/a","none","null"}
        degenerate=[
            x for x in got
            if x.lower() in bad_literals
            or len(re.findall(r"[A-Za-z0-9]+",x))<2
        ]
        if len(got)!=len(src) or degenerate:
            raise RuntimeError(
                (
                    "ADAPTER_TRAKE_INVALID_OUTPUT",
                    len(src),
                    len(got),
                    degenerate,
                )
            )
        return {
            "task_type":"TRAKE",
            "search_query_en":None,
            "events_en":got,
        }

    raise ValueError(("ADAPTER_TASK_UNSUPPORTED",t))


def _execute_emergency_local_fallback(task,query,events=None):
    """
    Emergency local mode inspired by the original Online1 local planner path.

    KIS:
      use the preserved Stage03 SigLIP2 backend, not the promoted CLIP adapter path.

    TRAKE:
      use the frozen local temporal engine directly with the original event text.

    QA:
      local evidence retrieval only. No fabricated answer is generated.
    """
    t=str(task).upper().strip()
    q=canonical_query_text(str(query))
    qid=_fresh_execution_query_id(t,q)

    if t=="KIS":
        st=run_kis_stage03_local_fallback(
            q,
            query_id=qid,
            include_multimodal_support=False,
        )
        p=st["active_result"]["payload"]
        return {
            "active_result":p,
            "online_control_trace":ONLINE_RUNTIME_TRACE.get(
                str(p.get("query_id",""))
            ),
            "alternatives":st.get("alternatives",[]),
            "execution_query_id":qid,
            "local_fallback_backend":"ONLINE1_STAGE03_SIGLIP2",
        }

    if t=="TRAKE":
        return _execute_retrieval("TRAKE",q,events)

    if t=="QA":
        # Evidence only. Final BTC QA rows require a grounded answer and therefore
        # must remain not-ready rather than inventing a local answer.
        return _execute_retrieval("QA",q,None)

    raise ValueError(("LOCAL_FALLBACK_UNSUPPORTED_TASK",t))


# QA answer/verifier handoff lives in RAM only.
# It never stores provider API keys.
QA_PHASE_STORE=globals().get("QA_PHASE_STORE",{})

def _qa_phase_put(prepared):
    phase_id="qa_"+secrets.token_hex(12)
    QA_PHASE_STORE[phase_id]=prepared

    # Tiny bounded manual-session cache.
    while len(QA_PHASE_STORE)>32:
        oldest=next(iter(QA_PHASE_STORE))
        QA_PHASE_STORE.pop(oldest,None)

    return phase_id

def _qa_phase_get(phase_id):
    x=QA_PHASE_STORE.get(str(phase_id))
    if x is None:
        raise KeyError(("QA_PHASE_NOT_FOUND",phase_id))
    return x

class AIC2026ServerHandler(BaseHTTPRequestHandler):
    server_version="AIC2026OnlineO2/0.5.18"

    def log_message(self,fmt,*args):
        # Never emit request body/provider API keys.
        return

    def _cors(self):
        self.send_header("Access-Control-Allow-Origin",SERVER_CORS_ORIGIN)
        self.send_header("Access-Control-Allow-Methods","GET,POST,OPTIONS")
        self.send_header("Access-Control-Allow-Headers","Content-Type")
        self.send_header("Access-Control-Max-Age","600")

    def _send_json(self,status,payload):
        raw=json.dumps(_jsonable(payload),ensure_ascii=False,separators=(",",":")).encode("utf-8")
        self.send_response(int(status))
        self.send_header("Content-Type","application/json; charset=utf-8")
        self.send_header("Content-Length",str(len(raw)))
        self.send_header("Cache-Control","no-store")
        self._cors()
        self.end_headers()
        self.wfile.write(raw)

    def _send_jpeg(self,status,data):
        self.send_response(int(status))
        self.send_header("Content-Type","image/jpeg")
        self.send_header("Content-Length",str(len(data)))
        self.send_header("Cache-Control","no-store")
        self._cors()
        self.end_headers()
        self.wfile.write(data)

    def _read_json(self):
        n=int(self.headers.get("Content-Length","0") or 0)
        if n<=0 or n>2_000_000:
            raise ValueError("invalid Content-Length")
        obj=json.loads(self.rfile.read(n).decode("utf-8"))
        if not isinstance(obj,dict):
            raise ValueError("JSON body must be object")
        return obj

    def do_OPTIONS(self):
        self.send_response(204)
        self._cors()
        self.end_headers()

    def do_GET(self):
        path=urlparse(self.path).path
        if path in {"","/"}:
            return self._send_json(200,{
                "ok":True,
                "service":"AIC2026 FINAL Competition Server",
                "status":"ready",
                "build_version":BUILD_VERSION,
                "health":"/health",
                "schema":"/schema",
                "providers":"/providers",
                "frontend":"IN_NOTEBOOK_UI_PLUS_EXTERNAL_API",
                "auth_mode":"NONE",
                "note":"No Bearer server token. Provider API keys, when used, are request-memory-only.",
            })
        if path=="/health":
            return self._send_json(200,{
                "status":"ok",
                "build_version":BUILD_VERSION,
                "online_phase":"FINAL_KIS_TRAKE_QA_COMPETITION",
                "frontend":"IN_NOTEBOOK_UI_PLUS_EXTERNAL_API",
                "auth_mode":"NONE",
                "query_json_dir":str(QUERY_JSON_DIR),
                "strict_dual_gpu":True,
                "normalization_status":"NORMALIZED_ONLY",
                "probability_calibration_status":"NOT_CALIBRATED",
                "official_submission_status":OFFICIAL_SUBMISSION_STATUS,
                "repeat_query_safe":True,
                "result_store_policy":"APPEND_ONLY_UNIQUE_EXECUTION_QUERY_ID",
                "ui_backend_transport":"LOCALHOST_BY_DEFAULT",
                "public_tunnel_required":bool(SERVER_REQUIRE_PUBLIC),
                "external_api_scope":"QA_ONLY",
                "kis_external_api_calls":0,
                "trake_external_api_calls":0,
                "qa_remote_calls_per_query":2,
                "competition_input_mode":"MANUAL_ONLY",
                "auto_dataset_query_loader":False,
                "clip_text_max_tokens":77,
                "clip_long_query_policy":"TOKENIZER_TRUNCATION",
                "task_input_policy":"AUTO_DETECT_FROM_FILENAME_THEN_QUERY",
                "groq_qa_max_images_per_request":3,
                "retrieval_language_adapter":"REMOTE_LLM_WHEN_NEEDED",
                "kis_adapter_policy":"VI_OR_LONG_TO_CONCISE_ENGLISH",
                "trake_adapter_policy":"VI_EVENTS_TO_CONCISE_ENGLISH",
                "api_exhaustion_fallback":"AUTO_LOCAL_WHEN_ENABLED",
                "local_kis_fallback_backend":"ONLINE1_STAGE03_SIGLIP2",
                "local_qa_fallback":"EVIDENCE_ONLY_NO_FAKE_ANSWER",
                "qa_key_rotation_policy":"PHASE_AWARE_EXHAUST_ALL_KEYS_BEFORE_LOCAL",
                "qa_answer_recomputed_on_verifier_rotation":False,
                "hybrid_fallback_order":[
                    "REMOTE_API_MAGAZINE",
                    "KAGGLE_QWEN3_4B_TEXT_ONLY",
                    "FROZEN_LOCAL_BACKEND",
                ],
                "qwen_model_status":_qwen_model_status(),
                "qwen3_4b_vision_capable":False,
                "qwen3_4b_qa_scope":"OCR_ASR_TEXT_ONLY",
            })
        if path=="/schema":
            return self._send_json(200,_server_contract())
        if path=="/providers":
            return self._send_json(200,{
                "providers":{
                    k:{"kind":v["kind"],"default_base_url":v["base_url"]}
                    for k,v in AGENT_PROVIDER_PRESETS.items()
                }
            })
        return self._send_json(404,{"error":"not_found"})

    def do_POST(self):
        path=urlparse(self.path).path
        try:
            body=self._read_json()

            if path=="/search":
                task=str(body.get("task_type","")).strip().upper()
                query=str(body.get("query","")).strip()
                if task not in {"KIS","TRAKE","QA"}:
                    raise ValueError("task_type must be KIS, TRAKE or QA")
                if not query:
                    raise ValueError("query is empty")
                events=body.get("events")
                with SERVER_EXEC_LOCK:
                    result=_execute_retrieval(task,query,events)
                _btc=_final_btc_from_result(task,result,None)
                response_payload={
                    "ok":True,"task_type":task,"query":query,**result,
                    "btc_ready":bool(_btc.get("ready")),
                    "btc_reason":_btc.get("reason"),
                    "btc_rows":_btc.get("rows",[]),
                    "btc_csv_text":_final_btc_csv_text(_btc.get("rows",[])) if _btc.get("ready") else "",
                }
                response_payload["query_json_file"]=_persist_query_exchange(
                    "/search",body,response_payload
                )
                return self._send_json(200,response_payload)



            if path=="/query/adapt/qwen":
                task=str(
                    body.get("task_type","")
                ).upper().strip()
                query=str(
                    body.get("query","")
                ).strip()
                events=body.get("events")

                if task not in {"KIS","TRAKE"}:
                    raise ValueError(
                        "query/adapt/qwen supports KIS or TRAKE"
                    )
                if not query:
                    raise ValueError("query is empty")

                client=_qwen_local_client()
                adapted=_retrieval_adapter(
                    task,
                    query,
                    events,
                    client,
                )
                sane,why=_qwen_adapter_output_sane(
                    task,
                    query,
                    adapted,
                )
                if not sane:
                    raise RuntimeError(
                        ("QWEN_ADAPTER_NGAO",why,adapted)
                    )

                return self._send_json(
                    200,
                    {
                        "ok":True,
                        "tier":"KAGGLE_QWEN3_4B_TEXT_ONLY",
                        **adapted,
                        "qwen_validation":why,
                    },
                )

            if path=="/qa/qwen/full":
                query=str(
                    body.get("query","")
                ).strip()
                if not query:
                    raise ValueError("query is empty")

                client=_qwen_local_client()

                with SERVER_EXEC_LOCK:
                    result=_execute_retrieval(
                        "QA",
                        query,
                        None,
                    )

                prepared=_final_qa_prepare_answer(
                    query,
                    result,
                    client,
                )
                verification=_final_qa_verify_prepared(
                    prepared,
                    client,
                )
                _btc=_final_btc_from_result(
                    "QA",
                    result,
                    verification,
                )

                response_payload={
                    "ok":True,
                    "task_type":"QA",
                    "tier":"KAGGLE_QWEN3_4B_TEXT_ONLY",
                    "agent_verification":verification,
                    **result,
                    "btc_ready":bool(_btc.get("ready")),
                    "btc_reason":_btc.get("reason"),
                    "btc_rows":_btc.get("rows",[]),
                    "btc_csv_text":(
                        _final_btc_csv_text(
                            _btc.get("rows",[])
                        )
                        if _btc.get("ready")
                        else ""
                    ),
                    "provider_api_key_persisted":False,
                }

                response_payload[
                    "query_json_file"
                ]=_persist_query_exchange(
                    "/qa/qwen/full",
                    {
                        "task_type":"QA",
                        "tier":"KAGGLE_QWEN3_4B_TEXT_ONLY",
                    },
                    response_payload,
                )
                return self._send_json(
                    200,
                    response_payload,
                )

            if path=="/qa/phase/verify-qwen":
                phase_id=str(
                    body.get("phase_id","")
                ).strip()
                if not phase_id:
                    raise ValueError("phase_id is empty")

                prepared=_qa_phase_get(phase_id)
                client=_qwen_local_client()

                verification=_final_qa_verify_prepared(
                    prepared,
                    client,
                )
                result=prepared["retrieval_response"]
                _btc=_final_btc_from_result(
                    "QA",
                    result,
                    verification,
                )

                response_payload={
                    "ok":True,
                    "task_type":"QA",
                    "phase":"QWEN_VERIFIER_COMPLETE",
                    "tier":"KAGGLE_QWEN3_4B_TEXT_ONLY",
                    "phase_id":phase_id,
                    "candidate_answer":
                        prepared["candidate_answer"],
                    "agent_verification":verification,
                    **result,
                    "btc_ready":bool(_btc.get("ready")),
                    "btc_reason":_btc.get("reason"),
                    "btc_rows":_btc.get("rows",[]),
                    "btc_csv_text":(
                        _final_btc_csv_text(
                            _btc.get("rows",[])
                        )
                        if _btc.get("ready")
                        else ""
                    ),
                    "provider_api_key_persisted":False,
                }

                QA_PHASE_STORE.pop(
                    phase_id,
                    None,
                )

                response_payload[
                    "query_json_file"
                ]=_persist_query_exchange(
                    "/qa/phase/verify-qwen",
                    {
                        "task_type":"QA",
                        "phase_id":phase_id,
                        "tier":"KAGGLE_QWEN3_4B_TEXT_ONLY",
                    },
                    response_payload,
                )

                return self._send_json(
                    200,
                    response_payload,
                )

            if path=="/search/local-fallback":
                task=str(body.get("task_type","")).strip().upper()
                query=str(body.get("query","")).strip()
                events=body.get("events")

                if task not in {"KIS","TRAKE","QA"}:
                    raise ValueError(
                        "task_type must be KIS, TRAKE or QA"
                    )
                if not query:
                    raise ValueError("query is empty")

                with SERVER_EXEC_LOCK:
                    result=_execute_emergency_local_fallback(
                        task,query,events
                    )

                _btc=_final_btc_from_result(task,result,None)

                response_payload={
                    "ok":True,
                    "task_type":task,
                    "query":query,
                    "execution_mode":"LOCAL_FALLBACK",
                    "local_fallback_backend":(
                        "ONLINE1_STAGE03_SIGLIP2"
                        if task=="KIS"
                        else "FINAL_LOCAL_RUNTIME"
                    ),
                    **result,
                    "btc_ready":bool(_btc.get("ready")),
                    "btc_reason":_btc.get("reason"),
                    "btc_rows":_btc.get("rows",[]),
                    "btc_csv_text":(
                        _final_btc_csv_text(_btc.get("rows",[]))
                        if _btc.get("ready")
                        else ""
                    ),
                }

                response_payload["query_json_file"]=_persist_query_exchange(
                    "/search/local-fallback",
                    body,
                    response_payload,
                )
                return self._send_json(200,response_payload)

            if path=="/query/adapt":
                task=str(body.get("task_type","")).upper().strip()
                query=str(body.get("query","")).strip()
                events=body.get("events")

                if task not in {"KIS","TRAKE"}:
                    raise ValueError("query/adapt supports KIS or TRAKE only")
                if not query:
                    raise ValueError("query is empty")

                cfg=_request_agent_config(body,require_model=True)
                if cfg.kind=="local":
                    raise ValueError("query/adapt requires remote provider")

                client=RequestAgentClient(cfg)
                adapted=_retrieval_adapter(task,query,events,client)

                return self._send_json(
                    200,
                    {
                        "ok":True,
                        "external_api_scope":"RETRIEVAL_LANGUAGE_ADAPTER",
                        "provider_api_key_persisted":False,
                        **adapted,
                    },
                )

            if path=="/qa/phase/answer":
                query=str(body.get("query","")).strip()
                if not query:
                    raise ValueError("query is empty")

                cfg=_request_agent_config(
                    body,
                    require_model=True,
                )
                if cfg.kind=="local":
                    raise ValueError(
                        "QA answer phase requires remote provider"
                    )

                client=RequestAgentClient(cfg)

                # Retrieval happens ONCE here.
                with SERVER_EXEC_LOCK:
                    result=_execute_retrieval(
                        "QA",
                        query,
                        None,
                    )

                prepared=_final_qa_prepare_answer(
                    query,
                    result,
                    client,
                )
                phase_id=_qa_phase_put(prepared)

                return self._send_json(
                    200,
                    {
                        "ok":True,
                        "task_type":"QA",
                        "phase":"ANSWER_COMPLETE",
                        "phase_id":phase_id,
                        "candidate_answer":prepared[
                            "candidate_answer"
                        ],
                        "answer_evidence_ids":prepared[
                            "answer_evidence_ids"
                        ],
                        "answer_provider":prepared[
                            "answer_provider"
                        ],
                        "answer_model":prepared[
                            "answer_model"
                        ],
                        "provider_api_key_persisted":False,
                    },
                )

            if path=="/qa/phase/verify":
                phase_id=str(
                    body.get("phase_id","")
                ).strip()
                if not phase_id:
                    raise ValueError("phase_id is empty")

                cfg=_request_agent_config(
                    body,
                    require_model=True,
                )
                if cfg.kind=="local":
                    raise ValueError(
                        "QA verifier phase requires remote provider"
                    )

                client=RequestAgentClient(cfg)
                prepared=_qa_phase_get(phase_id)

                verification=_final_qa_verify_prepared(
                    prepared,
                    client,
                )
                result=prepared["retrieval_response"]

                _btc=_final_btc_from_result(
                    "QA",
                    result,
                    verification,
                )

                response_payload={
                    "ok":True,
                    "task_type":"QA",
                    "phase":"VERIFIER_COMPLETE",
                    "phase_id":phase_id,
                    "candidate_answer":prepared[
                        "candidate_answer"
                    ],
                    "agent_verification":verification,
                    **result,
                    "btc_ready":bool(_btc.get("ready")),
                    "btc_reason":_btc.get("reason"),
                    "btc_rows":_btc.get("rows",[]),
                    "btc_csv_text":(
                        _final_btc_csv_text(
                            _btc.get("rows",[])
                        )
                        if _btc.get("ready")
                        else ""
                    ),
                    "provider_api_key_persisted":False,
                }

                # Successful completion: context no longer needed.
                QA_PHASE_STORE.pop(phase_id,None)

                response_payload[
                    "query_json_file"
                ]=_persist_query_exchange(
                    "/qa/phase/verify",
                    {
                        "task_type":"QA",
                        "phase_id":phase_id,
                        "provider":body.get("provider"),
                        "model":body.get("model"),
                    },
                    response_payload,
                )

                return self._send_json(
                    200,
                    response_payload,
                )

            if path=="/qa/answer":
                task="QA"
                query=str(body.get("query","")).strip()
                if not query:
                    raise ValueError("query is empty")

                cfg=_request_agent_config(body,require_model=True)
                if cfg.kind=="local":
                    raise ValueError(
                        "QA final answer requires remote provider/api_key/model"
                    )
                client=RequestAgentClient(cfg)

                # IMPORTANT: no external LLM planner here.
                # Retrieval is local first; external model only sees the small
                # grounded QA evidence pack.
                with SERVER_EXEC_LOCK:
                    result=_execute_retrieval("QA",query,None)

                verification_error=None
                verification=None
                try:
                    verification=_final_qa_agent_verify(query,result,client)
                except Exception as e:
                    verification_error=_redact_secret_text(
                        f"{type(e).__name__}: {e}"
                    )

                _btc=_final_btc_from_result("QA",result,verification)
                response_payload={
                    "ok":True,
                    "task_type":"QA",
                    "query":query,
                    "agent_mode":"QA_GROUNDED_ANSWER_VERIFY_ONLY",
                    "planner_used":False,
                    "external_api_scope":"QA_ONLY",
                    "provider_api_key_persisted":False,
                    **result,
                    "agent_verification":verification,
                    "verification_error":verification_error,
                    "btc_ready":bool(_btc.get("ready")),
                    "btc_reason":_btc.get("reason"),
                    "btc_rows":_btc.get("rows",[]),
                    "btc_csv_text":_final_btc_csv_text(
                        _btc.get("rows",[])
                    ) if _btc.get("ready") else "",
                }
                response_payload["query_json_file"]=_persist_query_exchange(
                    "/qa/answer",body,response_payload
                )
                return self._send_json(200,response_payload)

            if path=="/agent/search":
                task=str(body.get("task_type","")).strip().upper()
                query=str(body.get("query","")).strip()
                mode=_validate_agent_mode(body.get("agent_mode","local"))
                if task not in {"KIS","TRAKE","QA"}:
                    raise ValueError("task_type must be KIS, TRAKE or QA")
                if not query:
                    raise ValueError("query is empty")

                planner_error=None
                verification_error=None
                verification=None
                plan={"search_query":query,"events":None,"notes":["local_identity_plan"]}
                client=None

                if mode!="local":
                    cfg=_request_agent_config(body,require_model=True)
                    if cfg.kind=="local":
                        raise ValueError(
                            "remote agent_mode requires provider/api_key/model in this request"
                        )
                    client=RequestAgentClient(cfg)
                    try:
                        plan=_agent_plan(task,query,client)
                    except Exception as e:
                        planner_error=_redact_secret_text(f"{type(e).__name__}: {e}")
                        plan={"search_query":query,"events":None,"notes":["planner_fallback_original_query"]}

                with SERVER_EXEC_LOCK:
                    result=_execute_retrieval(
                        task,plan["search_query"],plan.get("events")
                    )

                if mode=="llm_plan_verify" and client is not None:
                    try:
                        if task=="QA":
                            verification=_final_qa_agent_verify(query,result,client)
                        else:
                            verification=_agent_verify(task,query,result,client)
                    except Exception as e:
                        verification_error=_redact_secret_text(f"{type(e).__name__}: {e}")

                _btc=_final_btc_from_result(task,result,verification)
                response_payload={
                    "ok":True,
                    "task_type":task,
                    "original_query":query,
                    "agent_mode":mode,
                    "agent_config_source":"REQUEST_ONLY" if mode!="local" else "LOCAL_DEFAULT",
                    "provider_api_key_persisted":False,
                    "plan":plan,
                    "planner_error":planner_error,
                    **result,
                    "agent_verification":verification,
                    "verification_error":verification_error,
                    "btc_ready":bool(_btc.get("ready")),
                    "btc_reason":_btc.get("reason"),
                    "btc_rows":_btc.get("rows",[]),
                    "btc_csv_text":_final_btc_csv_text(_btc.get("rows",[])) if _btc.get("ready") else "",
                }
                response_payload["query_json_file"]=_persist_query_exchange(
                    "/agent/search",body,response_payload
                )
                return self._send_json(200,response_payload)

            if path=="/models":
                cfg=_request_agent_config(body,require_model=False)
                if cfg.kind=="local":
                    return self._send_json(200,{
                        "ok":True,
                        "provider":"local",
                        "models":["local-rule-based"],
                    })
                client=RequestAgentClient(cfg)
                return self._send_json(200,{
                    "ok":True,
                    "provider":cfg.provider,
                    "models":client.list_models(),
                    "provider_api_key_persisted":False,
                })

            if path=="/agent/test":
                cfg=_request_agent_config(body,require_model=True)
                if cfg.kind=="local":
                    return self._send_json(200,{
                        "ok":True,
                        "provider":"local",
                        "model":"local-rule-based",
                        "response":{"ok":True},
                    })
                client=RequestAgentClient(cfg)
                resp=client.generate_json(
                    "Return only JSON.",
                    '{"instruction":"Return an object with ok=true"}',
                )
                return self._send_json(200,{
                    "ok":True,
                    "provider":cfg.provider,
                    "model":cfg.model,
                    "response":resp,
                    "provider_api_key_persisted":False,
                })

            if path=="/btc/save":
                task=str(body.get("task_type","")).strip().upper()
                query_file=str(body.get("query_file","")).strip()
                rows=body.get("rows")
                saved=_final_save_btc(query_file,task,rows)
                return self._send_json(200,{
                    "ok":True,
                    "task_type":task,
                    "saved":saved,
                    "provider_api_key_persisted":False,
                })

            if path=="/btc/build":
                built=_final_build_submission_zip()
                return self._send_json(200,{
                    "ok":True,
                    "submission":built,
                    "provider_api_key_persisted":False,
                })

            if path=="/preview":
                vid=str(body.get("video_id","")).strip()
                pts=float(body.get("pts_time"))
                if not vid or not math.isfinite(pts) or pts<0:
                    raise ValueError("invalid video_id/pts_time")
                with SERVER_EXEC_LOCK:
                    row=decode_raw_first_at_or_after(
                        vid,pts,
                        request_id="server_preview_"+hashlib.sha256(
                            f"{vid}|{pts:.6f}".encode()
                        ).hexdigest()[:18],
                        persist_jpeg=False,
                    )
                    image=row["_image"]
                    buf=io.BytesIO()
                    image.save(buf,format="JPEG",quality=88,optimize=True)
                    data=buf.getvalue()
                    try:
                        image.close()
                    except Exception:
                        pass
                return self._send_jpeg(200,data)

            return self._send_json(404,{"error":"not_found"})

        except Exception as e:
            return self._send_json(400,{
                "ok":False,
                "error_type":type(e).__name__,
                "error":_redact_secret_text(str(e)),
            })


def _read_pid(path):
    try:
        return int(Path(path).read_text(encoding="utf-8").strip())
    except Exception:
        return None

def _pid_cmdline(pid):
    try:
        raw=Path(f"/proc/{int(pid)}/cmdline").read_bytes()
        return raw.replace(b"\x00",b" ").decode("utf-8",errors="replace").strip()
    except Exception:
        return ""

def _tracked_cloudflared_alive():
    pid=_read_pid(SERVER_TUNNEL_PID_PATH)
    if not pid:
        return None
    cmd=_pid_cmdline(pid)
    if "cloudflared" in cmd and "tunnel" in cmd:
        return pid
    return None

def _stop_tracked_cloudflared():
    """
    Stop only the cloudflared PID created/tracked by this notebook.
    Never pkill every cloudflared process on the machine.
    """
    global SERVER_TUNNEL_PROCESS,SERVER_PUBLIC_URL
    pid=_tracked_cloudflared_alive()
    stopped=False

    # Prefer child-process handle when this kernel owns it.
    proc=SERVER_TUNNEL_PROCESS
    if proc is not None and proc.poll() is None:
        try:
            proc.terminate()
            proc.wait(timeout=5)
            stopped=True
        except Exception:
            try:
                proc.kill()
                proc.wait(timeout=3)
                stopped=True
            except Exception:
                pass

    # Kernel may have reset while child cloudflared survived.
    if not stopped and pid:
        try:
            os.kill(pid,signal.SIGTERM)
            for _ in range(20):
                if not Path(f"/proc/{pid}").exists():
                    stopped=True
                    break
                time.sleep(0.1)
            if not stopped and Path(f"/proc/{pid}").exists():
                os.kill(pid,signal.SIGKILL)
                stopped=True
        except ProcessLookupError:
            stopped=True
        except Exception:
            pass

    SERVER_TUNNEL_PROCESS=None
    SERVER_PUBLIC_URL=None
    try:
        SERVER_TUNNEL_PID_PATH.unlink(missing_ok=True)
    except Exception:
        pass
    return stopped

def _read_current_public_url():
    try:
        u=SERVER_URL_PATH.read_text(encoding="utf-8").strip().splitlines()[0].strip()
    except Exception:
        return None
    if re.fullmatch(r"https://[a-zA-Z0-9-]+\.trycloudflare\.com/?",u):
        return u.rstrip("/")
    return None

def _probe_public_url(public_url,retries=SERVER_PUBLIC_HEALTH_RETRIES):
    """
    Validate the actual public path, not merely cloudflared stdout.

    v0.5.18:
    - Never let a non-JSON HTML response collapse diagnostics into JSONDecodeError.
    - Preserve HTTP status/content-type/body preview in the probe record.
    """
    if not public_url:
        return False,None

    health_url=public_url.rstrip("/")+"/health"
    last=None

    for attempt in range(max(1,int(retries))):
        try:
            r=requests.get(
                health_url,
                timeout=8,
                headers={"Cache-Control":"no-cache"},
            )
            content_type=str(r.headers.get("content-type") or "")
            body_preview=str(r.text or "")[:300].replace("\n"," ")

            last={
                "attempt":attempt+1,
                "status_code":int(r.status_code),
                "content_type":content_type,
                "body_preview":body_preview,
            }

            if r.status_code==200:
                try:
                    data=r.json()
                except Exception as parse_error:
                    last={
                        **last,
                        "json_parse_error_type":type(parse_error).__name__,
                        "json_parse_error":str(parse_error)[:200],
                    }
                else:
                    last={**last,"payload":data}
                    if (
                        isinstance(data,dict)
                        and data.get("status")=="ok"
                        and data.get("build_version")==BUILD_VERSION
                    ):
                        return True,last

        except Exception as e:
            last={
                "attempt":attempt+1,
                "error_type":type(e).__name__,
                "error":str(e)[:300],
            }

        time.sleep(SERVER_PUBLIC_HEALTH_DELAY_S)

    return False,last

def _reuse_existing_public_tunnel_if_healthy():
    """
    Reuse the same Quick Tunnel inside the current Kaggle container when possible.
    This is the key anti-garbage rule: rerunning the final cell does not spawn
    another random hostname if the tracked tunnel still serves /health.
    """
    global SERVER_PUBLIC_URL
    url=_read_current_public_url()
    pid=_tracked_cloudflared_alive()
    if not url or not pid:
        return None

    ok,probe=_probe_public_url(url,retries=2)
    if not ok:
        return None

    SERVER_PUBLIC_URL=url
    return {
        "status":"REUSED_EXISTING_HEALTHY_TUNNEL",
        "public_url":url,
        "pid":pid,
        "probe":probe,
    }


def _probe_local_aic_server(port,timeout=1.5):
    """
    Probe whether a local port already serves this AIC server contract.
    This handles a surviving HTTP thread even if the Python variable was overwritten.
    """
    url=f"http://{SERVER_HOST}:{int(port)}"
    try:
        r=requests.get(url+"/health",timeout=timeout,headers={"Cache-Control":"no-cache"})
        if r.status_code!=200:
            return False,None
        data=r.json()
        if data.get("status")!="ok":
            return False,data
        # Handler reads BUILD_VERSION from current notebook globals, so a reused
        # in-kernel server reports the current build after notebook rerun.
        return True,data
    except Exception:
        return False,None

def _server_object_local_url(server):
    try:
        addr=server.server_address
        port=int(addr[1])
        return f"http://{SERVER_HOST}:{port}",port
    except Exception:
        return None,None

def _shutdown_server_object(server):
    if server is None:
        return
    try:
        server.shutdown()
    except Exception:
        pass
    try:
        server.server_close()
    except Exception:
        pass

def start_local_server():
    """
    Idempotent local HTTP server startup.

    Order:
      1. Reuse the previous in-kernel ThreadingHTTPServer if it is still healthy.
      2. If the requested port already serves an AIC server, reuse that listener.
      3. Otherwise bind the first free port in a bounded range.
      4. Never fail immediately with EADDRINUSE just because 8765 is occupied.
    """
    global SERVER_HTTPD,SERVER_THREAD,SERVER_PORT,SERVER_LOCAL_REUSE_STATUS

    # 1) Strongest reuse path: we still have the actual prior server object.
    if SERVER_HTTPD is not None:
        previous_url,previous_port=_server_object_local_url(SERVER_HTTPD)
        if previous_port is not None:
            ok,_diag=_probe_local_aic_server(previous_port)
            if ok:
                SERVER_PORT=previous_port
                SERVER_LOCAL_REUSE_STATUS="REUSED_PREVIOUS_SERVER_OBJECT"
                print("[PASS] REUSED LOCAL API SERVER",previous_url)
                return previous_url

        # Stale object: close it before proceeding.
        _shutdown_server_object(SERVER_HTTPD)
        SERVER_HTTPD=None
        SERVER_THREAD=None

    # 2) The object reference may be lost while its daemon thread/listener survives.
    ok,_diag=_probe_local_aic_server(SERVER_REQUESTED_PORT)
    if ok:
        SERVER_PORT=SERVER_REQUESTED_PORT
        SERVER_LOCAL_REUSE_STATUS="REUSED_EXISTING_AIC_LISTENER"
        url=f"http://{SERVER_HOST}:{SERVER_PORT}"
        print("[PASS] REUSED EXISTING LOCAL AIC LISTENER",url)
        return url

    # 3) Bounded failover. If 8765 is owned by another service, try 8766...
    last_error=None
    for candidate_port in range(
        SERVER_REQUESTED_PORT,
        SERVER_REQUESTED_PORT+max(1,SERVER_PORT_SCAN_LIMIT),
    ):
        # Do not accidentally start a second AIC server if one is already there.
        ok,_diag=_probe_local_aic_server(candidate_port)
        if ok:
            SERVER_PORT=candidate_port
            SERVER_LOCAL_REUSE_STATUS="REUSED_EXISTING_AIC_LISTENER"
            url=f"http://{SERVER_HOST}:{SERVER_PORT}"
            print("[PASS] REUSED EXISTING LOCAL AIC LISTENER",url)
            return url

        try:
            httpd=ThreadingHTTPServer(
                (SERVER_HOST,candidate_port),AIC2026ServerHandler
            )
        except OSError as e:
            last_error=e
            if getattr(e,"errno",None)==98:
                continue
            raise

        SERVER_HTTPD=httpd
        SERVER_PORT=candidate_port
        SERVER_THREAD=threading.Thread(
            target=SERVER_HTTPD.serve_forever,
            name=f"aic2026-online-o2-http-{candidate_port}",
            daemon=True,
        )
        SERVER_THREAD.start()
        SERVER_LOCAL_REUSE_STATUS=(
            "BOUND_REQUESTED_PORT"
            if candidate_port==SERVER_REQUESTED_PORT
            else "BOUND_FALLBACK_PORT"
        )
        url=f"http://{SERVER_HOST}:{SERVER_PORT}"
        print("[PASS] LOCAL API SERVER",{
            "url":url,
            "requested_port":SERVER_REQUESTED_PORT,
            "actual_port":SERVER_PORT,
            "status":SERVER_LOCAL_REUSE_STATUS,
        })
        return url

    raise RuntimeError({
        "message":"NO_FREE_LOCAL_SERVER_PORT",
        "host":SERVER_HOST,
        "requested_port":SERVER_REQUESTED_PORT,
        "scan_limit":SERVER_PORT_SCAN_LIMIT,
        "last_error":repr(last_error),
    })

def _get_cloudflared():
    p=shutil.which("cloudflared")
    if p:
        return Path(p)
    target=Path("/kaggle/working/cloudflared")
    if target.is_file() and os.access(target,os.X_OK):
        return target
    if os.environ.get("AIC2026_ALLOW_CLOUDFLARED_DOWNLOAD","1")=="0":
        raise RuntimeError("cloudflared not installed and download disabled")
    src="https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64"
    tmp=target.with_suffix(".tmp")
    urllib.request.urlretrieve(src,tmp)
    tmp.chmod(0o755)
    tmp.replace(target)
    return target

TRYCLOUDFLARE_URL_PATTERN=re.compile(
    r"https://[a-zA-Z0-9-]+\.trycloudflare\.com"
)

TRYCLOUDFLARE_RESERVED_HOSTS={
    "api.trycloudflare.com",
    "www.trycloudflare.com",
}

def _normalize_trycloudflare_candidate(url):
    try:
        candidate=str(url or "").strip().rstrip("/")
        parsed=urlparse(candidate)
        host=str(parsed.hostname or "").lower()
    except Exception:
        return None

    if parsed.scheme!="https":
        return None
    if not host.endswith(".trycloudflare.com"):
        return None
    if host in TRYCLOUDFLARE_RESERVED_HOSTS:
        return None

    # Real Quick Tunnel hostnames are delegated subdomains, never the service API.
    label=host.removesuffix(".trycloudflare.com").strip(".")
    if not label or label in {"api","www"}:
        return None

    return "https://"+host

def _quick_tunnel_urls_from_log_line(line):
    out=[]
    for m in TRYCLOUDFLARE_URL_PATTERN.finditer(str(line or "")):
        candidate=_normalize_trycloudflare_candidate(m.group(0))
        if candidate and candidate not in out:
            out.append(candidate)
    return out

def start_public_tunnel(local_url,timeout_s=45):
    """
    Start one cloudflared Quick Tunnel and wait for the *actual* random hostname.

    Important:
    cloudflared may log the Quick Tunnel service API endpoint
    `https://api.trycloudflare.com` before it logs the assigned public hostname.
    v0.5.17 returned the first regex match and therefore sometimes published the
    API service URL instead of the tunnel. v0.5.18 explicitly rejects reserved
    hosts and continues reading stdout.
    """
    global SERVER_TUNNEL_PROCESS

    binary=_get_cloudflared()
    proc=subprocess.Popen(
        [str(binary),"tunnel","--no-autoupdate","--url",str(local_url)],
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
    )
    SERVER_TUNNEL_PROCESS=proc
    SERVER_TUNNEL_PID_PATH.write_text(str(proc.pid)+"\n",encoding="utf-8")
    q=queue.Queue()

    def reader():
        try:
            for line in proc.stdout:
                q.put(line)
        finally:
            q.put(None)

    threading.Thread(target=reader,daemon=True).start()

    deadline=time.time()+float(timeout_s)
    tail=[]
    rejected_urls=[]

    while time.time()<deadline:
        try:
            line=q.get(timeout=0.5)
        except queue.Empty:
            if proc.poll() is not None:
                break
            continue

        if line is None:
            break

        line_text=line.rstrip()
        tail.append(line_text)
        tail=tail[-30:]

        # Keep rejected trycloudflare URLs for diagnostics, but never publish them.
        for raw_match in TRYCLOUDFLARE_URL_PATTERN.findall(line_text):
            normalized=_normalize_trycloudflare_candidate(raw_match)
            if normalized:
                print("[TUNNEL] detected Quick Tunnel URL",normalized)
                return normalized,tail
            if raw_match not in rejected_urls:
                rejected_urls.append(raw_match)
                print("[TUNNEL] ignored reserved/non-tunnel URL",raw_match)

    if proc.poll() is None:
        proc.terminate()

    raise RuntimeError({
        "message":"TRYCLOUDFLARE_PUBLIC_URL_NOT_DETECTED",
        "rejected_urls":rejected_urls[-10:],
        "log_tail":tail[-20:],
    })

def publish_online_server(force_new=False):
    global SERVER_PUBLIC_URL
    local_url=start_local_server()

    # Local origin must be healthy first.
    health=requests.get(local_url+"/health",timeout=10).json()
    if health.get("status")!="ok":
        raise RuntimeError(("LOCAL_SERVER_HEALTH_FAILED",health))

    contract=_server_contract()
    atomic_json(SERVER_CONTRACT_PATH,contract)

    base_status={
        "build_version":BUILD_VERSION,
        "local_url":local_url,
        "public_url":None,
        "url_txt":str(SERVER_URL_PATH),
        "server_contract_json":str(SERVER_CONTRACT_PATH),
        "server_state_json":str(SERVER_STATE_PATH),
        "tunnel_pid_txt":str(SERVER_TUNNEL_PID_PATH),
        "cors_origin":SERVER_CORS_ORIGIN,
        "provider_api_key_persisted":False,
        "singleton_policy":SERVER_SINGLETON_POLICY,
        "quick_tunnel_url_policy":"IGNORE_RESERVED_HOSTS_WAIT_FOR_RANDOM_HOSTNAME",
        "local_server_requested_port":SERVER_REQUESTED_PORT,
        "local_server_actual_port":SERVER_PORT,
        "local_server_reuse_status":SERVER_LOCAL_REUSE_STATUS,
    }

    # Competition notebook UI runs in the same Kaggle kernel.
    # Do NOT send its own HTTP calls out through TryCloudflare and back in.
    # That hairpin path can fail with "Network is unreachable" even while
    # localhost and provider outbound Internet are healthy.
    if not SERVER_REQUIRE_PUBLIC:
        SERVER_PUBLIC_URL=None
        status={
            **base_status,
            "status":"PASS_LOCAL_API_SERVER_ONLY",
            "public_url":None,
            "tunnel":"DISABLED_BY_DEFAULT_FOR_IN_NOTEBOOK_UI",
        }
        atomic_json(SERVER_STATUS_PATH,status)
        atomic_json(SERVER_STATE_PATH,status)
        try:
            SERVER_URL_PATH.write_text(local_url+"\n",encoding="utf-8")
        except Exception:
            pass
        print("[PASS] LOCAL API SERVER",local_url)
        print("[INFO] TryCloudflare disabled by default for in-notebook competition UI")
        return status

    # Rerun-safe: reuse one already healthy tunnel instead of creating another.
    if not force_new:
        reused=_reuse_existing_public_tunnel_if_healthy()
        if reused:
            status={
                **base_status,
                **reused,
                "public_url":reused["public_url"],
                "tunnel":"TRYCLOUDFLARE_TEMPORARY_REUSED",
            }
            atomic_json(SERVER_STATUS_PATH,status)
            atomic_json(SERVER_STATE_PATH,status)
            SERVER_URL_PATH.write_text(reused["public_url"]+"\n",encoding="utf-8")
            print("[PASS] REUSED PUBLIC API SERVER",reused["public_url"])
            print("[SINGLETON] no new cloudflared process created")
            return status

    # Existing tracked tunnel is unhealthy or caller explicitly requested replace.
    _stop_tracked_cloudflared()

    # Never leave a stale URL file looking authoritative while replacement starts.
    try:
        SERVER_URL_PATH.unlink(missing_ok=True)
    except Exception:
        pass

    try:
        candidate_url,tunnel_tail=start_public_tunnel(local_url)

        # Critical: do not publish url.txt until the public /health returns our build.
        ok,probe=_probe_public_url(candidate_url)
        if not ok:
            _stop_tracked_cloudflared()
            raise RuntimeError({
                "message":"PUBLIC_TUNNEL_HEALTH_CHECK_FAILED",
                "candidate_url":candidate_url,
                "probe":probe,
            })

        SERVER_PUBLIC_URL=candidate_url.rstrip("/")
        SERVER_URL_PATH.write_text(SERVER_PUBLIC_URL+"\n",encoding="utf-8")
        status={
            **base_status,
            "status":"PASS_PUBLIC_API_SERVER",
            "public_url":SERVER_PUBLIC_URL,
            "tunnel":"TRYCLOUDFLARE_TEMPORARY_SINGLETON",
            "tunnel_pid":_tracked_cloudflared_alive(),
            "public_health_probe":probe,
            "tunnel_log_tail":tunnel_tail[-8:],
        }
        atomic_json(SERVER_STATUS_PATH,status)
        atomic_json(SERVER_STATE_PATH,status)
        print("[PASS] PUBLIC API SERVER",SERVER_PUBLIC_URL)
        print("[PASS] PUBLIC ROOT 200",SERVER_PUBLIC_URL+"/")
        print("[PASS] PUBLIC HEALTH 200",SERVER_PUBLIC_URL+"/health")
        print("[SAVED]",SERVER_URL_PATH)
        print("[SAVED]",SERVER_CONTRACT_PATH)
        return status

    except Exception as e:
        _stop_tracked_cloudflared()
        status={
            **base_status,
            "status":"PUBLIC_TUNNEL_FAILED",
            "error_type":type(e).__name__,
            "error":_redact_secret_text(str(e)),
        }
        atomic_json(SERVER_STATUS_PATH,status)
        atomic_json(SERVER_STATE_PATH,status)
        if SERVER_REQUIRE_PUBLIC:
            raise RuntimeError(status) from e
        print("[WARN] public tunnel failed; local API remains ready",status)
        return status

def replace_public_tunnel():
    """Explicit manual replacement. Normal reruns should call publish_online_server()."""
    return publish_online_server(force_new=True)

def stop_public_tunnel():
    """Explicit cleanup for the one tracked tunnel."""
    stopped=_stop_tracked_cloudflared()
    try:
        SERVER_URL_PATH.unlink(missing_ok=True)
    except Exception:
        pass
    state={
        "status":"STOPPED",
        "build_version":BUILD_VERSION,
        "stopped":bool(stopped),
        "created_at_utc":utc_now(),
    }
    atomic_json(SERVER_STATE_PATH,state)
    print("[STOPPED] tracked public tunnel",state)
    return state

SERVER_RUNTIME_STATUS=publish_online_server()
if SERVER_RUNTIME_STATUS.get("status") not in {
    "PASS_LOCAL_API_SERVER_ONLY",
    "PASS_PUBLIC_API_SERVER",
    "REUSED_EXISTING_HEALTHY_TUNNEL",
} and SERVER_REQUIRE_PUBLIC:
    raise AssertionError(SERVER_RUNTIME_STATUS)

print("[PASS] ONLINE O2 + SERVER READY",{
    "technical_build":BUILD_VERSION,
    "local_url":SERVER_RUNTIME_STATUS.get("local_url"),
    "local_server_reuse_status":SERVER_LOCAL_REUSE_STATUS,
    "public_url":SERVER_RUNTIME_STATUS.get("public_url"),
    "url_txt":str(SERVER_URL_PATH),
    "frontend":"IN_NOTEBOOK_UI_PLUS_EXTERNAL_API",
})
````

````python
# ── FINAL USER UI v1.0.21 — MANUAL QUERY + AUTO TASK ─────────────────────────
#
# No dataset query loader. No batch AUTO.
#
# User only enters:
#   - official query filename
#   - query text
#
# Task is detected automatically:
#   *-kis.txt   -> KIS
#   *-qa.txt    -> QA
#   *-trake.txt -> TRAKE
#
# Query-content heuristics are fallback only.
# ──────────────────────────────────────────────────────────────────────────────

from pathlib import Path
from IPython.display import display, HTML
import ipywidgets as widgets
import requests, json, math, html, hashlib, re, csv, io, os, time

PROVIDER_OPTIONS=[
    ("Local — không API","local"),
    ("Groq","groq"),
    ("Gemini","gemini"),
    ("OpenAI","openai"),
    ("OpenRouter","openrouter"),
    ("Anthropic","anthropic"),
    ("Mistral","mistral"),
    ("DeepSeek","deepseek"),
    ("xAI","xai"),
    ("Together","together"),
    ("Fireworks","fireworks"),
    ("Cerebras","cerebras"),
]

_QUERY_EVENT_RE=re.compile(r"(?im)^\s*E(\d+)\s*:\s*(.+?)\s*$")

# ==============================================================================
# LOCAL BACKEND
# ==============================================================================
def _be_base():
    port=globals().get("SERVER_PORT")
    if port is not None:
        return f"http://127.0.0.1:{int(port)}"
    status_obj=globals().get("SERVER_RUNTIME_STATUS")
    if isinstance(status_obj,dict) and status_obj.get("local_url"):
        return str(status_obj["local_url"]).strip().rstrip("/")
    return "http://127.0.0.1:8765"

def _be_get(path,timeout=30):
    r=requests.get(_be_base()+path,timeout=timeout)
    if r.status_code>=400:
        try:
            detail=r.json()
        except Exception:
            detail=r.text[:1000]
        raise RuntimeError(f"HTTP {r.status_code}: {detail}")
    return r.json()

def _be_post(path,payload,timeout=360):
    r=requests.post(
        _be_base()+path,
        headers={"Content-Type":"application/json"},
        json=payload,
        timeout=timeout,
    )
    if r.status_code>=400:
        try:
            detail=r.json()
        except Exception:
            detail=r.text[:1000]
        raise RuntimeError(f"HTTP {r.status_code}: {detail}")
    return r.json()

def _assert_server_ready():
    h=_be_get("/health",timeout=30)
    expected="aic2026-final-competition-server-v1.0.21"
    got=str(h.get("build_version",""))
    if got!=expected:
        raise RuntimeError(
            f"SERVER_BUILD_MISMATCH: got={got or '<unknown>'}, expected={expected}. "
            "Restart Session rồi Run All."
        )
    if h.get("competition_input_mode")!="MANUAL_ONLY":
        raise RuntimeError("SERVER_MANUAL_ONLY_PATCH_MISSING")
    if h.get("repeat_query_safe") is not True:
        raise RuntimeError("SERVER_REPEAT_QUERY_PATCH_MISSING")
    return h

# ==============================================================================
# API KEY MAGAZINE — only QA uses it
# ==============================================================================
def _parse_keys(text):
    vals=[
        x.strip()
        for x in re.split(r"[\s,;]+",str(text or ""))
        if x.strip()
    ]
    out=[]
    seen=set()
    for x in vals:
        if x not in seen:
            seen.add(x)
            out.append(x)
    return out

class KeyMagazine:
    def __init__(self):
        self.keys=[]
        self.idx=0
        self.fp=None

    def load(self,text,force=False):
        keys=_parse_keys(text)
        if not keys:
            raise ValueError("Băng API key đang trống.")
        fp=hashlib.sha256("\n".join(keys).encode()).hexdigest()
        if force or fp!=self.fp:
            self.keys=keys
            self.idx=0
            self.fp=fp
        return self.status()

    def current(self):
        if self.idx>=len(self.keys):
            raise RuntimeError(
                "HẾT BĂNG API KEY — dán list mới rồi NẠP / THAY BĂNG."
            )
        return self.keys[self.idx]

    def drop(self):
        self.idx+=1

    def status(self):
        total=len(self.keys)
        rem=max(0,total-self.idx)
        return {
            "total":total,
            "remaining":rem,
            "current":self.idx+1 if rem else None,
            "dropped":self.idx,
        }

KEY_MAG=KeyMagazine()

_PERMANENT_KEY_MARKERS=(
    "401","invalid api key","invalid_api_key","api key invalid",
    "unauthorized","authentication_error","authentication failed",
    "insufficient_quota","insufficient quota","quota exhausted",
    "billing hard limit","account disabled",
)
_TEMP_RATE_MARKERS=(
    "429","rate limit","rate_limit","too many requests",
    "tokens per minute","requests per minute","tpm","rpm",
)

def _err_class(text):
    s=str(text or "").lower()
    if any(x in s for x in _PERMANENT_KEY_MARKERS):
        return "PERMANENT"
    if any(x in s for x in _TEMP_RATE_MARKERS):
        return "RATE"
    return "OTHER"

def _mag_text():
    st=KEY_MAG.status()
    if not st["total"]:
        return "Băng đạn: chưa nạp"
    if not st["remaining"]:
        return f"Băng đạn: HẾT · {st['dropped']}/{st['total']} key đã bỏ"
    return (
        f"Băng đạn: slot {st['current']}/{st['total']} · "
        f"còn {st['remaining']}"
    )

# ==============================================================================
# PROVIDER MODEL — lazy, QA only
# ==============================================================================
AUTO_MODEL={"provider":None,"model":None,"slot":None}

def _auto_model_for_current_key():
    p=str(provider.value).strip()
    if p=="local":
        return "local-rule-based"

    st=KEY_MAG.status()

    if (
        AUTO_MODEL.get("provider")==p
        and AUTO_MODEL.get("model")
        and AUTO_MODEL.get("slot")==st.get("current")
    ):
        return AUTO_MODEL["model"]

    if p=="groq":
        m="qwen/qwen3.6-27b"
        AUTO_MODEL.update(
            {"provider":p,"model":m,"slot":st.get("current")}
        )
        return m

    resp=_be_post(
        "/models",
        {
            "provider":p,
            "api_key":KEY_MAG.current(),
            "model":"",
            "base_url":"",
        },
        timeout=60,
    )
    models=[str(x) for x in resp.get("models",[]) if str(x).strip()]
    if not models:
        raise RuntimeError(f"{p}: /models không trả model.")

    markers=(
        "vision","multimodal","-vl","_vl","/vl","gemini",
        "gpt-4o","gpt-4.1","gpt-5","claude","llama-4",
        "qwen3.6","qwen3-vl","qwen2.5-vl","pixtral",
    )
    filtered=[
        m for m in models
        if any(x in m.lower() for x in markers)
        and not any(
            x in m.lower()
            for x in ("embed","whisper","tts","guard","moderation")
        )
    ]
    if not filtered:
        raise RuntimeError(
            f"Không tự chọn được model vision cho {p}. Hãy đổi Provider."
        )

    m=sorted(filtered)[0]
    AUTO_MODEL.update(
        {"provider":p,"model":m,"slot":st.get("current")}
    )
    return m

# ==============================================================================
# TASK AUTO-DETECT
# ==============================================================================
def _filename_task(filename):
    name=Path(str(filename or "").strip()).name.lower()
    stem=re.sub(r"\.(txt|csv)$","",name)

    if stem.endswith("-trake"):
        return "TRAKE","FILENAME_SUFFIX"
    if stem.endswith("-qa"):
        return "QA","FILENAME_SUFFIX"
    if stem.endswith("-kis"):
        return "KIS","FILENAME_SUFFIX"

    return None,None

def _query_task_hint(query_text):
    """
    Fallback only. Official filename suffix always wins.
    """
    text=str(query_text or "").strip()
    if not text:
        return None,None

    # TRAKE is strongly identifiable by multiple event lines.
    if len(list(_QUERY_EVENT_RE.finditer(text)))>=2:
        return "TRAKE","QUERY_EVENT_STRUCTURE"

    low=text.lower().lstrip()
    qa_prefixes=(
        "ai ","ai là ","cái gì","gì ","ở đâu","khi nào",
        "bao nhiêu","bao lâu","màu gì","tên gì","vì sao","tại sao",
        "who ","what ","where ","when ","how many","how much",
        "why ","which ",
    )

    if "?" in text or any(low.startswith(x) for x in qa_prefixes):
        return "QA","QUERY_QUESTION_HEURISTIC"

    return "KIS","QUERY_DEFAULT_KIS"

def _detect_task(filename,query_text):
    ft,freason=_filename_task(filename)
    if ft:
        qh,_=_query_task_hint(query_text)

        # Only strong TRAKE structural conflict blocks execution.
        # QA-vs-KIS language heuristics are not reliable enough to override
        # an official BTC filename.
        conflict=None
        if qh=="TRAKE" and ft!="TRAKE":
            conflict=(
                f"Query có nhiều dòng E1/E2... nhưng filename được nhận là {ft}"
            )

        return ft,freason,conflict

    qt,qreason=_query_task_hint(query_text)
    if qt:
        return qt,qreason,None

    return None,"UNRESOLVED",None

# ==============================================================================
# TIER-2 QWEN KAGGLE
# ==============================================================================
def _refresh_qwen_status():
    try:
        st=_qwen_model_status()

        if st.get("found"):
            qwen_status.value=(
                "<b style='color:#66d18f'>"
                "Qwen tier: READY · Qwen3-4B TEXT-ONLY · "
                +html.escape(
                    str(st.get("model_dir"))
                )
                +"</b>"
            )
        else:
            qwen_status.value=(
                "<b style='color:#ff7777'>"
                "Qwen tier: MISSING · hãy Add Input model "
                "qwen-lm/qwen-3 → Transformers → 4b → version 1. "
                "Nếu Tier 2 cần mà vẫn thiếu, hệ thống đi Local."
                "</b>"
            )

    except Exception as e:
        qwen_status.value=(
            "<b style='color:#ff7777'>"
            "Qwen tier status error · "
            +html.escape(str(e))
            +"</b>"
        )


def _qwen_adapter(
    task_name,
    query_text,
    events=None,
):
    payload={
        "task_type":str(task_name).upper(),
        "query":str(query_text),
    }
    if events is not None:
        payload["events"]=events

    try:
        resp=_be_post(
            "/query/adapt/qwen",
            payload,
            timeout=420,
        )
    except Exception as e:
        return {
            "fallback_local":True,
            "reason":
                "QWEN_RUNTIME_OR_VALIDATION_FAIL",
            "qwen_error":str(e),
        }

    if _adapter_result_is_degenerate(
        resp,
        task_name,
    ):
        return {
            "fallback_local":True,
            "reason":"QWEN_DEGENERATE_OUTPUT",
        }

    return resp


def _qwen_full_qa(query_text):
    try:
        return _be_post(
            "/qa/qwen/full",
            {
                "task_type":"QA",
                "query":str(query_text),
            },
            timeout=720,
        )
    except Exception as e:
        return {
            "fallback_local":True,
            "reason":"QWEN_QA_NGAO_OR_RUNTIME_FAIL",
            "qwen_error":str(e),
        }


def _qwen_verify_phase(phase_id):
    try:
        return _be_post(
            "/qa/phase/verify-qwen",
            {
                "task_type":"QA",
                "phase_id":str(phase_id),
            },
            timeout=720,
        )
    except Exception as e:
        return {
            "fallback_local":True,
            "reason":
                "QWEN_VERIFIER_NGAO_OR_RUNTIME_FAIL",
            "qwen_error":str(e),
        }

# ==============================================================================
# API -> LOCAL FALLBACK POLICY
# ==============================================================================
def _local_fallback_enabled():
    try:
        return bool(fallback_local.value)
    except Exception:
        return True

def _use_local_fallback(task_name,query_text,events=None,reason="API_UNAVAILABLE"):
    adapted_query.value=(
        f"[LOCAL_FALLBACK · {reason}]\n"
        + (
            "KIS backend = Online1 Stage03 SigLIP2\n"
            if str(task_name).upper()=="KIS"
            else "backend = final local runtime\n"
        )
        + str(query_text)
    )

    payload={
        "task_type":str(task_name).upper(),
        "query":str(query_text),
    }
    if events is not None:
        payload["events"]=events

    return _be_post(
        "/search/local-fallback",
        payload,
        timeout=600,
    )

def _adapter_result_is_degenerate(resp,task_name):
    if not isinstance(resp,dict):
        return True

    t=str(task_name).upper()

    if t=="KIS":
        x=str(resp.get("search_query_en","")).strip()
        if x.lower() in {"","...","…",".","-","n/a","none","null"}:
            return True
        return len(re.findall(r"[A-Za-z0-9]+",x))<3

    if t=="TRAKE":
        arr=resp.get("events_en")
        if not isinstance(arr,list) or not arr:
            return True
        for x in arr:
            z=str(x).strip()
            if z.lower() in {"","...","…",".","-","n/a","none","null"}:
                return True
        return False

    return False

# ==============================================================================
# RETRIEVAL LANGUAGE ADAPTER
# ==============================================================================
_VI_DIACRITICS=set(
    "ăâđêôơưĂÂĐÊÔƠƯ"
    "áàảãạấầẩẫậắằẳẵặéèẻẽẹếềểễệ"
    "íìỉĩịóòỏõọốồổỗộớờởỡợ"
    "úùủũụứừửữựýỳỷỹỵ"
    "ÁÀẢÃẠẤẦẨẪẬẮẰẲẴẶÉÈẺẼẸẾỀỂỄỆ"
    "ÍÌỈĨỊÓÒỎÕỌỐỒỔỖỘỚỜỞỠỢ"
    "ÚÙỦŨỤỨỪỬỮỰÝỲỶỸỴ"
)

_VI_COMMON=re.compile(
    r"(?i)\b("
    r"một|những|các|đang|được|với|trong|trên|dưới|người|con|"
    r"tại|ở|và|của|có|không|đây|này|sau|trước|giới thiệu|"
    r"miền nam|miền bắc|khoảng|vừa|thêm"
    r")\b"
)

def _looks_vietnamese(text):
    s=str(text or "")
    return (
        any(ch in _VI_DIACRITICS for ch in s)
        or bool(_VI_COMMON.search(s))
    )

def _clip_token_count(text):
    try:
        proc=FINAL_CLIP_WORKERS[0]["processor"]
        tok=getattr(proc,"tokenizer",None)
        if tok is None:
            return None
        out=tok(str(text),add_special_tokens=True,truncation=False)
        ids=out.get("input_ids",[])
        return len(ids)
    except Exception:
        return None

def _needs_retrieval_adapter(text):
    n=_clip_token_count(text)
    return _looks_vietnamese(text) or (n is not None and n>70)

def _call_adapter_with_magazine(
    task_name,
    query_text,
    events=None,
):
    """
    Tier 1 only: remote API magazine.

    Returns:
      normal adapter response on success
      {"fallback_qwen": True, ...} when the API tier is unavailable/exhausted

    It NEVER jumps straight to Local unless the user explicitly selected
    Provider=Local. Tier 2 Qwen gets the next chance.
    """
    if str(provider.value).strip()=="local":
        return {
            "fallback_local":True,
            "reason":"FORCED_LOCAL_PROVIDER",
        }

    keys=_parse_keys(api_keys.value)

    if not keys:
        return {
            "fallback_qwen":True,
            "reason":"NO_API_KEYS",
        }

    try:
        KEY_MAG.load(
            api_keys.value,
            force=False,
        )
    except Exception:
        return {
            "fallback_qwen":True,
            "reason":"KEY_MAGAZINE_EMPTY",
        }

    while KEY_MAG.status()["remaining"]>0:
        key=KEY_MAG.current()
        model_id=_auto_model_for_current_key()

        payload={
            "task_type":str(task_name).upper(),
            "query":str(query_text),
            "events":events,
            "provider":str(provider.value),
            "api_key":key,
            "model":model_id,
            "base_url":"",
        }

        same_key_attempt=0

        while True:
            same_key_attempt+=1

            try:
                resp=_be_post(
                    "/query/adapt",
                    payload,
                    timeout=120,
                )

                if _adapter_result_is_degenerate(
                    resp,
                    task_name,
                ):
                    return {
                        "fallback_qwen":True,
                        "reason":"API_DEGENERATE_OUTPUT",
                    }

                return resp

            except Exception as e:
                msg=str(e)
                cls=_err_class(msg)

                if (
                    cls=="RATE"
                    and same_key_attempt<=3
                ):
                    wait=min(
                        12,
                        2**same_key_attempt,
                    )
                    run_status.value=(
                        "<b style='color:#ffc35b'>"
                        "API rate-limit · slot "
                        f"{KEY_MAG.status().get('current')} "
                        f"· retry sau {wait}s"
                        "</b>"
                    )
                    time.sleep(wait)
                    continue

                if cls in {"RATE","PERMANENT"}:
                    KEY_MAG.drop()
                    AUTO_MODEL.update({
                        "provider":None,
                        "model":None,
                        "slot":None,
                    })
                    _update_key_status()
                    break

                low=msg.lower()
                transport=any(
                    x in low
                    for x in (
                        "connection",
                        "timeout",
                        "temporarily unavailable",
                        "service unavailable",
                        "http 500",
                        "http 502",
                        "http 503",
                        "http 504",
                    )
                )
                if transport:
                    return {
                        "fallback_qwen":True,
                        "reason":"API_TRANSPORT_FAIL",
                    }

                if (
                    "adapter_degenerate" in low
                    or
                    "adapter_trake_invalid_output"
                    in low
                ):
                    return {
                        "fallback_qwen":True,
                        "reason":"API_INVALID_ADAPTER_OUTPUT",
                    }

                raise

    return {
        "fallback_qwen":True,
        "reason":"ALL_API_KEYS_EXHAUSTED",
    }


def _adapt_via_hybrid(
    task_name,
    query_text,
    events=None,
):
    """
    API -> Qwen Kaggle -> Local.
    """
    a=_call_adapter_with_magazine(
        task_name,
        query_text,
        events=events,
    )

    if a.get("fallback_local"):
        return a

    if a.get("fallback_qwen"):
        run_status.value=(
            "<b style='color:#ffc35b'>"
            "API tier unavailable · chuyển Qwen3-4B Kaggle TEXT"
            "</b>"
        )
        q=_qwen_adapter(
            task_name,
            query_text,
            events=events,
        )
        if q.get("fallback_local"):
            q["api_reason"]=a.get("reason")
        return q

    a["tier"]="REMOTE_API"
    return a


# ==============================================================================
# TRAKE EVENT PARSER
# ==============================================================================
def _parse_trake_events(query_text):
    text=str(query_text or "").strip()
    matches=list(_QUERY_EVENT_RE.finditer(text))

    if len(matches)<2:
        return None

    prefix=text[:matches[0].start()].strip()
    events=[]

    for i,m in enumerate(matches):
        et=str(m.group(2)).strip()

        if prefix:
            et=f"{prefix} {et}".strip()

        events.append({
            "event_index":i,
            "event_id":f"e{i}",
            "text":et,
            "optional":False,
            "relation_from_prev":"STRICT_BEFORE",
        })

    return events

# ==============================================================================
# RUN ONE QUERY
# ==============================================================================
def _run_kis(query_text):
    original=str(query_text).strip()

    if str(provider.value).strip()=="local":
        return _use_local_fallback(
            "KIS",
            original,
            reason="FORCED_LOCAL_PROVIDER",
        )

    search_query=original
    reason="DIRECT_CLIP"

    if _needs_retrieval_adapter(original):
        a=_adapt_via_hybrid(
            "KIS",
            original,
        )

        if a.get("fallback_local"):
            return _use_local_fallback(
                "KIS",
                original,
                reason=(
                    "QWEN_NGAO_TO_LOCAL · "
                    +str(a.get("reason","UNKNOWN"))
                ),
            )

        search_query=str(
            a.get("search_query_en","")
        ).strip()

        if _adapter_result_is_degenerate(
            a,
            "KIS",
        ):
            return _use_local_fallback(
                "KIS",
                original,
                reason="HYBRID_ADAPTER_DEGENERATE_TO_LOCAL",
            )

        reason=(
            "QWEN_KAGGLE_VI_TO_EN"
            if a.get("tier")=="KAGGLE_QWEN3_4B_TEXT"
            else "API_VI_TO_EN"
        )

    n=_clip_token_count(search_query)

    adapted_query.value=(
        f"[{reason}]"
        +(f" · CLIP tokens={n}" if n is not None else "")
        +"\n"
        +search_query
    )

    return _be_post(
        "/search",
        {
            "task_type":"KIS",
            "query":search_query,
        },
        timeout=360,
    )


def _run_trake(query_text):
    original=str(query_text).strip()
    events=_parse_trake_events(original)

    if str(provider.value).strip()=="local":
        return _use_local_fallback(
            "TRAKE",
            original,
            events=events,
            reason="FORCED_LOCAL_PROVIDER",
        )

    payload={
        "task_type":"TRAKE",
        "query":original,
    }

    if events is not None:
        final_events=events

        if any(
            _needs_retrieval_adapter(z["text"])
            for z in events
        ):
            a=_adapt_via_hybrid(
                "TRAKE",
                original,
                events=events,
            )

            if a.get("fallback_local"):
                return _use_local_fallback(
                    "TRAKE",
                    original,
                    events=events,
                    reason=(
                        "QWEN_NGAO_TO_LOCAL · "
                        +str(a.get("reason","UNKNOWN"))
                    ),
                )

            got=a.get("events_en")

            if (
                not isinstance(got,list)
                or len(got)!=len(events)
                or _adapter_result_is_degenerate(
                    a,
                    "TRAKE",
                )
            ):
                return _use_local_fallback(
                    "TRAKE",
                    original,
                    events=events,
                    reason="HYBRID_TRAKE_INVALID_TO_LOCAL",
                )

            final_events=[
                {
                    **events[i],
                    "text":str(got[i]).strip(),
                }
                for i in range(len(events))
            ]

            tier=(
                "QWEN_KAGGLE"
                if a.get("tier")=="KAGGLE_QWEN3_4B_TEXT"
                else "REMOTE_API"
            )
            adapted_query.value=(
                f"[{tier} · TRAKE]\n"
                +"\n".join(
                    f"E{i+1}: {z['text']}"
                    for i,z in enumerate(final_events)
                )
            )

        else:
            adapted_query.value=(
                "[DIRECT_TRAKE]\n"
                +"\n".join(
                    f"E{i+1}: {z['text']}"
                    for i,z in enumerate(final_events)
                )
            )

        payload["events"]=final_events

    return _be_post(
        "/search",
        payload,
        timeout=600,
    )


def _qa_phase_remote_call(path,payload,phase_label):
    """
    Try the current key with short backoff.
    If the key is permanently invalid OR still rate-limited after retries,
    move to the NEXT backup key.

    Return response on success.
    Return None only after ALL remaining keys are exhausted.
    """
    while KEY_MAG.status()["remaining"]>0:
        key=KEY_MAG.current()
        model_id=_auto_model_for_current_key()

        request_payload={
            **payload,
            "provider":str(provider.value),
            "api_key":key,
            "model":model_id,
            "base_url":"",
        }

        same_key_attempt=0

        while True:
            same_key_attempt+=1

            try:
                return _be_post(
                    path,
                    request_payload,
                    timeout=360,
                )

            except Exception as e:
                msg=str(e)
                cls=_err_class(msg)

                if cls=="RATE" and same_key_attempt<=3:
                    wait=min(12,2**same_key_attempt)
                    run_status.value=(
                        f"<b style='color:#ffc35b'>"
                        f"{phase_label} · slot "
                        f"{KEY_MAG.status().get('current')} "
                        f"rate-limit · retry sau {wait}s"
                        f"</b>"
                    )
                    time.sleep(wait)
                    continue

                if cls in {"RATE","PERMANENT"}:
                    # Crucial v1.0.16 behavior:
                    # do NOT jump Local here.
                    # Burn/skip this unusable slot and try the next backup key.
                    KEY_MAG.drop()
                    AUTO_MODEL.update(
                        {"provider":None,"model":None,"slot":None}
                    )
                    _update_key_status()

                    run_status.value=(
                        f"<b style='color:#ffc35b'>"
                        f"{phase_label} · chuyển sang API key kế tiếp"
                        f"</b>"
                    )
                    break

                # Semantic verifier rejection / bad citations / malformed
                # answer are NOT key failures. Do not rotate blindly.
                raise

    return None


def _run_qa(query_text):
    original=str(query_text).strip()

    if str(provider.value).strip()=="local":
        return _use_local_fallback(
            "QA",
            original,
            reason="FORCED_LOCAL_PROVIDER_EVIDENCE_ONLY",
        )

    keys=_parse_keys(api_keys.value)

    # ------------------------------------------------------------------
    # TIER 1 — REMOTE API MAGAZINE
    # ------------------------------------------------------------------
    answer_resp=None
    if keys:
        KEY_MAG.load(
            api_keys.value,
            force=False,
        )
        answer_resp=_qa_phase_remote_call(
            "/qa/phase/answer",
            {
                "task_type":"QA",
                "query":original,
            },
            "QA ANSWER",
        )

    # No API keys / all keys exhausted before answer:
    # try FULL local Qwen answer+verify.
    if answer_resp is None:
        run_status.value=(
            "<b style='color:#ffc35b'>"
            "API QA hết/không khả dụng · chuyển Qwen3-4B Kaggle TEXT"
            "</b>"
        )
        q=_qwen_full_qa(original)

        if not q.get("fallback_local"):
            adapted_query.value=(
                "[QA · QWEN3-4B TEXT PASS]\n"
                "Qwen3-4B TEXT answer + verifier từ OCR/ASR hoàn tất."
            )
            return q

        # Qwen is missing/OOM/malformed/rejected => final local evidence only.
        return _use_local_fallback(
            "QA",
            original,
            reason=(
                "QWEN_NGAO_TO_LOCAL · "
                +str(q.get("reason","UNKNOWN"))
            ),
        )

    phase_id=str(
        answer_resp.get("phase_id","")
    ).strip()
    candidate=str(
        answer_resp.get("candidate_answer","")
    ).strip()

    if not phase_id or not candidate:
        # Remote API produced malformed phase output:
        # Qwen gets a fresh full attempt before Local.
        q=_qwen_full_qa(original)
        if not q.get("fallback_local"):
            adapted_query.value=(
                "[QA · REMOTE MALFORMED → QWEN PASS]"
            )
            return q
        return _use_local_fallback(
            "QA",
            original,
            reason="REMOTE_AND_QWEN_INVALID_TO_LOCAL",
        )

    adapted_query.value=(
        "[QA API ANSWER COMPLETE]\n"
        f"Candidate: {candidate}\n"
        "Verifier đang xoay API keys."
    )

    verify_resp=_qa_phase_remote_call(
        "/qa/phase/verify",
        {
            "task_type":"QA",
            "phase_id":phase_id,
        },
        "QA VERIFIER",
    )

    if verify_resp is not None:
        adapted_query.value=(
            "[QA · REMOTE API PASS]\n"
            f"Answer: {candidate}"
        )
        return verify_resp

    # ------------------------------------------------------------------
    # TIER 2 — KAGGLE QWEN VERIFIES THE SAME API CANDIDATE
    # ------------------------------------------------------------------
    run_status.value=(
        "<b style='color:#ffc35b'>"
        "API verifier hết băng · chuyển Qwen3-4B verifier"
        "</b>"
    )
    qv=_qwen_verify_phase(phase_id)

    if not qv.get("fallback_local"):
        adapted_query.value=(
            "[QA · API ANSWER + QWEN VERIFIER PASS]\n"
            f"Answer: {candidate}"
        )
        return qv

    # ------------------------------------------------------------------
    # TIER 3 — FROZEN LOCAL EVIDENCE ONLY
    # ------------------------------------------------------------------
    local=_use_local_fallback(
        "QA",
        original,
        reason=(
            "QWEN_NGAO_TO_LOCAL · "
            +str(qv.get("reason","UNKNOWN"))
        ),
    )
    adapted_query.value=(
        "[QA · API ANSWER EXISTS · QWEN VERIFIER FAILED]\n"
        f"Candidate chưa được verify: {candidate}\n"
        "Đã chuyển Local evidence-only; không lưu CSV giả."
    )
    return local


def _run_one(task_name,query_text):
    t=str(task_name).upper()

    if t=="KIS":
        return _run_kis(query_text)
    if t=="TRAKE":
        return _run_trake(query_text)
    if t=="QA":
        return _run_qa(query_text)

    raise ValueError(t)

# ==============================================================================
# PREVIEW
# ==============================================================================
def _num(v):
    try:
        x=float(v)
        return x if math.isfinite(x) else None
    except Exception:
        return None

def _pts(row):
    for k in (
        "actual_pts_time","pts_time","target_pts_time",
        "pts_time_technical","middle_pts_time","center_pts_time",
    ):
        x=_num(row.get(k))
        if x is not None and x>=0:
            return x
    return None

def _walk(obj,parent_video=None):
    if isinstance(obj,dict):
        vid=obj.get("video_id") or parent_video
        yield obj,vid
        for v in obj.values():
            yield from _walk(v,vid)

    elif isinstance(obj,list):
        for v in obj:
            yield from _walk(v,parent_video)

def _preview_items(resp,limit=10):
    out=[]
    seen=set()

    for row,vid in _walk(resp):
        if not vid:
            continue

        t=_pts(row)
        if t is None:
            continue

        k=(str(vid),round(float(t),3))
        if k in seen:
            continue

        seen.add(k)
        out.append((str(vid),float(t)))

        if len(out)>=limit:
            break

    return out

def _render_gallery(resp):
    cards=[]

    for rank,(vid,t) in enumerate(_preview_items(resp,10),1):
        try:
            r=requests.post(
                _be_base()+"/preview",
                headers={"Content-Type":"application/json"},
                json={"video_id":vid,"pts_time":t},
                timeout=60,
            )

            if r.status_code>=400:
                raise RuntimeError(r.status_code)

            img=widgets.Image(
                value=r.content,
                format="jpeg",
                layout=widgets.Layout(
                    width="190px",
                    height="120px",
                    object_fit="cover",
                ),
            )

        except Exception:
            img=widgets.HTML(
                "<div style='width:190px;height:120px;"
                "display:grid;place-items:center;border:1px solid #555'>"
                "NO IMAGE</div>"
            )

        cards.append(
            widgets.VBox(
                [
                    img,
                    widgets.HTML(
                        f"<div style='font-size:11px'>"
                        f"<b>#{rank}</b> {html.escape(vid)} @ {t:.3f}s"
                        f"</div>"
                    ),
                ],
                layout=widgets.Layout(width="200px"),
            )
        )

    gallery.children=tuple(cards)

# ==============================================================================
# UI WIDGETS
# ==============================================================================
style={"description_width":"105px"}

provider=widgets.Dropdown(
    options=PROVIDER_OPTIONS,
    value="groq",
    description="API Provider:",
    style=style,
    layout=widgets.Layout(width="420px"),
)

api_keys=widgets.Textarea(
    value="",
    placeholder=(
        "Dùng cho KIS/TRAKE khi cần chuẩn hóa tiếng Việt → English, và cho QA."
    ),
    description="API keys:",
    style=style,
    layout=widgets.Layout(width="980px",height="120px"),
)
api_keys.add_class("keys-hidden")

show_keys=widgets.Checkbox(
    value=False,
    description="Hiện key",
    indent=False,
)

load_keys=widgets.Button(
    description="NẠP / THAY BĂNG",
    icon="refresh",
    layout=widgets.Layout(width="210px",height="42px"),
)

key_status=widgets.HTML(value="Băng đạn: chưa nạp")

qwen_status=widgets.HTML()



fallback_local=widgets.Checkbox(
    value=True,
    description="Hybrid: API → Qwen3-4B Kaggle → Local",
    indent=False,
)

query_file=widgets.Text(
    value="",
    placeholder="Ví dụ: query-p1-12-qa.txt",
    description="Query file:",
    style=style,
    layout=widgets.Layout(width="700px"),
)

query=widgets.Textarea(
    value="",
    placeholder="Dán nguyên query BTC vào đây…",
    description="Query:",
    style=style,
    layout=widgets.Layout(width="980px",height="180px"),
)

detected_task=widgets.HTML(
    value="<b style='color:#ffc35b'>Task tự nhận diện: chưa đủ thông tin</b>"
)

adapted_query=widgets.Textarea(
    value="",
    description="Query model:",
    disabled=True,
    style={"description_width":"105px"},
    layout=widgets.Layout(width="980px",height="85px"),
)

run_btn=widgets.Button(
    description="CHẠY QUERY + LƯU CSV",
    icon="play",
    layout=widgets.Layout(width="300px",height="50px"),
)

build_btn=widgets.Button(
    description="BUILD submission.zip",
    icon="archive",
    layout=widgets.Layout(width="260px",height="50px"),
)

run_status=widgets.HTML()

btc_csv=widgets.Textarea(
    value="",
    description="BTC CSV:",
    style={"description_width":"80px"},
    layout=widgets.Layout(width="980px",height="150px"),
)

gallery=widgets.HBox(
    children=[],
    layout=widgets.Layout(
        width="100%",
        overflow_x="auto",
        border="1px solid #333",
        padding="8px",
    ),
)

zip_status=widgets.HTML()

# ==============================================================================
# UI EVENTS
# ==============================================================================
def _update_key_status():
    key_status.value=(
        f"<b style='color:#66d18f'>{html.escape(_mag_text())}</b>"
    )

def _load_keys(_=None):
    try:
        KEY_MAG.load(api_keys.value,force=True)
        AUTO_MODEL.update(
            {"provider":None,"model":None,"slot":None}
        )
        _update_key_status()

    except Exception as e:
        key_status.value=(
            f"<b style='color:#ff7777'>NẠP BĂNG FAIL · "
            f"{html.escape(str(e))}</b>"
        )

def _provider_changed(change=None):
    is_local=str(provider.value)=="local"
    api_keys.disabled=is_local
    load_keys.disabled=is_local
    show_keys.disabled=is_local

    if is_local:
        key_status.value=(
            "<b style='color:#66d18f'>LOCAL MODE · không dùng API</b>"
        )
    else:
        _update_key_status()

def _toggle_keys(change):
    if change["new"]:
        api_keys.remove_class("keys-hidden")
        api_keys.add_class("keys-visible")
    else:
        api_keys.remove_class("keys-visible")
        api_keys.add_class("keys-hidden")

def _refresh_detected_task(change=None):
    t,reason,conflict=_detect_task(query_file.value,query.value)

    if t:
        conflict_html=(
            f" · <span style='color:#ff7777'>{html.escape(conflict)}</span>"
            if conflict else ""
        )

        detected_task.value=(
            f"<b style='color:#66d18f'>Task tự nhận diện: {t}</b>"
            f" <span style='color:#999'>({reason})</span>"
            f"{conflict_html}"
        )

    else:
        detected_task.value=(
            "<b style='color:#ffc35b'>Task tự nhận diện: chưa đủ thông tin</b>"
        )

def _rows_to_csv(rows):
    sio=io.StringIO(newline="")
    csv.writer(
        sio,
        delimiter=",",
        lineterminator="\n",
        quoting=csv.QUOTE_MINIMAL,
    ).writerows(rows)
    return sio.getvalue()

def _run_manual(_):
    # Critical UX fix: do not show stale rows/gallery from the previous query
    # if the new query fails.
    btc_csv.value=""
    gallery.children=tuple()
    adapted_query.value=""
    run_status.value=""
    zip_status.value=""

    try:
        _assert_server_ready()

        q=query.value.strip()
        if not q:
            raise ValueError("Query đang trống.")

        filename=query_file.value.strip()
        if not filename:
            raise ValueError(
                "Nhập tên file query BTC, ví dụ query-p1-12-qa.txt."
            )

        task_name,reason,conflict=_detect_task(filename,q)

        if not task_name:
            raise ValueError(
                "Không tự nhận diện được KIS / QA / TRAKE."
            )

        if conflict:
            raise ValueError(
                f"TASK_CONFLICT: {conflict}. "
                "Kiểm tra lại filename trước khi chạy."
            )

        run_status.value=(
            f"<b style='color:#ffc35b'>"
            f"Đang chạy {task_name} · {reason}…"
            f"</b>"
        )

        resp=_run_one(task_name,q)

        if not resp.get("btc_ready"):
            # Local QA fallback is intentionally evidence-only. Show images and
            # stop without writing a fake/invalid QA CSV.
            if (
                task_name=="QA"
                and resp.get("execution_mode")=="LOCAL_FALLBACK"
            ):
                _render_gallery(resp)
                run_status.value=(
                    "<b style='color:#ffc35b'>"
                    "LOCAL QA FALLBACK · evidence đã tìm được nhưng "
                    "không có VLM local để sinh đáp án → KHÔNG lưu CSV giả."
                    "</b>"
                )
                return

            raise RuntimeError(
                f"BTC_NOT_READY: {resp.get('btc_reason') or 'unknown'}"
            )

        rows=resp.get("btc_rows",[])

        saved=_be_post(
            "/btc/save",
            {
                "query_file":filename,
                "task_type":task_name,
                "rows":rows,
            },
            timeout=60,
        )

        btc_csv.value=_rows_to_csv(rows)
        _render_gallery(resp)

        saved_info=saved.get("saved",{})

        run_status.value=(
            "<b style='color:#66d18f'>PASS · "
            f"{task_name} · "
            f"{html.escape(str(saved_info.get('filename',filename)))} · "
            f"{len(rows)} row(s)</b>"
        )

    except Exception as e:
        run_status.value=(
            f"<b style='color:#ff7777'>FAIL · "
            f"{html.escape(type(e).__name__+': '+str(e))}</b>"
        )

def _build_zip(_):
    try:
        out=_be_post("/btc/build",{},timeout=60)
        s=out.get("submission",{})

        zip_status.value=(
            "<b style='color:#66d18f'>SUBMISSION READY · "
            f"{html.escape(str(s.get('zip_path','/kaggle/working/submission.zip')))} · "
            f"{s.get('file_count','?')} CSV</b>"
        )

    except Exception as e:
        zip_status.value=(
            f"<b style='color:#ff7777'>BUILD FAIL · "
            f"{html.escape(str(e))}</b>"
        )

load_keys.on_click(_load_keys)
show_keys.observe(_toggle_keys,names="value")
provider.observe(_provider_changed,names="value")
_provider_changed()
_refresh_qwen_status()
query_file.observe(_refresh_detected_task,names="value")
query.observe(_refresh_detected_task,names="value")
run_btn.on_click(_run_manual)
build_btn.on_click(_build_zip)

# ==============================================================================
# DISPLAY
# ==============================================================================
css=r"""
<style>
.aic-title{font-size:27px;font-weight:900;margin:4px 0 8px}
.aic-section{font-size:22px;font-weight:850;margin:18px 0 8px}
.aic-note{color:#bbb;line-height:1.55;margin:3px 0 9px}
.keys-hidden textarea{-webkit-text-security:disc !important;}
.keys-visible textarea{-webkit-text-security:none !important;}
</style>
"""

display(HTML(css))

display(
    widgets.VBox(
        [
            widgets.HTML(
                "<div class='aic-title'>AIC26 FINAL — MANUAL + AUTO TASK</div>"
                "<div class='aic-note'>"
                "Không còn dataset loader. Không cần chọn KIS/QA/TRAKE bằng tay: "
                "notebook tự nhận diện từ filename BTC."
                "</div>"
            ),

            widgets.HTML(
                f"<div class='aic-note'><b>Backend local:</b> "
                f"{html.escape(_be_base())}</div>"
            ),

            widgets.HTML(
                "<div class='aic-section'>1 · MÔ HÌNH ĐANG DÙNG</div>"
                "<div class='aic-note'>"
                "<b>KIS:</b> API adapter → CLIP; hết API → Online1 Stage03 SigLIP2 local<br>"
                "<b>TRAKE:</b> API event adapter khi cần; hết API → local temporal/raw PTS<br>"
                "<b>QA:</b> API VISION → Qwen3-4B chỉ đọc OCR/ASR TEXT → local evidence-only"                "</div>"
            ),

            widgets.HTML(
                "<div class='aic-note'>"
                "<b>Hybrid mặc định bật: xoay hết API keys → Qwen3-4B trong Kaggle → nếu Qwen lỗi/ngáo/OOM thì Local.</b>"
                "</div>"
            ),

            provider,
            api_keys,
            widgets.HBox([load_keys,show_keys,key_status]),
            fallback_local,
            qwen_status,

            widgets.HTML(
                "<div class='aic-section'>2 · NHẬP QUERY THỦ CÔNG</div>"
            ),

            query_file,
            query,
            detected_task,
            adapted_query,

            widgets.HTML(
                "<div class='aic-note'>"
                "Ưu tiên filename: <b>-kis.txt / -qa.txt / -trake.txt</b>. "
                "Nếu chạy lại cùng tên query, CSV cũ sẽ được <b>ghi đè</b>; "
                "không cần nút xóa riêng."
                "</div>"
            ),

            widgets.HBox([run_btn,build_btn]),

            widgets.HTML("<div class='aic-section'>KẾT QUẢ</div>"),
            run_status,
            btc_csv,
            gallery,
            zip_status,
        ],
        layout=widgets.Layout(width="100%",gap="8px"),
    )
)

print("[READY] v1.0.21 QWEN3-4B INTERIM HYBRID")
print("KIS   = API -> Qwen3-4B Kaggle TEXT -> Online1 Stage03 local")
print("TRAKE = API -> Qwen3-4B Kaggle TEXT -> local/raw PTS")
print("QA    = API vision -> Qwen3-4B OCR/ASR text-only -> local evidence-only")
print("Task detect = filename suffix first; query heuristic fallback")
print("Final output = /kaggle/working/submission.zip")
````

## v1.0.21 — interim Qwen3-4B attached Kaggle Model

Exact input:

```text
https://www.kaggle.com/models/qwen-lm/qwen-3/Transformers/4b/1
```

Recommended temporary role:

```text
API
→ Qwen3-4B text-only
→ frozen Local
```

Why 4B instead of other Qwen3 dense sizes:

```text
Qwen3-1.7B : lighter, but weaker as a Vietnamese→English adapter/reasoner
Qwen3-4B   : best temporary balance for this 2×T4 retrieval runtime
Qwen3-8B   : stronger text model, but ~16.4 GB weights and materially higher OOM risk
             while CLIP/SigLIP/retrieval are already resident
```

Qwen3-4B is text-only:

```text
KIS   : good tier-2 adapter
TRAKE : good tier-2 event adapter
QA    : only OCR/ASR/text-evidence fallback
```

For QA, if there is no OCR/ASR evidence, Qwen3-4B is forced to fail/abstain and the
system falls through to Local evidence-only. It is never allowed to pretend it saw images.

This version also removes the extra Qwen pip-install path: the Qwen subprocess reuses the
already-frozen Transformers 5.0.0 environment, which is new enough for Qwen3 and avoids a
network/package mutation during the competition.
